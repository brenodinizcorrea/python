# Class and Static Methods

Imagine that you are building a game.

The game creates different players:

```python
player_1 = Player("Ana")
player_2 = Player("Bruno")
player_3 = Player("Carlos")
```

Each player has its own data, such as a name and score.

But the program also needs to answer:

```text
How many players have been created?
```

This information does not belong to one specific player. It belongs to the Player class as a whole.

Now imagine that the same game also needs to validate whether a score is valid before creating a player.

For example:

```python
Player.is_valid_score(100)
```

This validation does not depend on:

- one specific player
- the data of an existing player
- data stored in the Player class

It only needs the value being validated.

## Instance Methods

An instance method operates on a specific object.

It receives `self`, which refers to the instance that called the method.

Instance methods can access and modify that object's attributes.

````python
class Player:

    def __init__(self, name, score):
        self.name = name
        self.score = score

    def add_score(self, points):
        self.score += points


player = Player("Ana", 10)

player.add_score(5)

print(player.score)
````

When:

`player.add_score(5)`

is called, `self` refers to `player`.

The method therefore operates on the state of that specific object.

## Data Shared by a Class

Some data belongs to the class rather than to one specific instance.

For example, a program may want to keep track of how many `Player` objects have been created.

````python
class Player:

    total_players = 0

    def __init__(self, name):
        self.name = name

        Player.total_players += 1


player_1 = Player("Ana")
player_2 = Player("Bruno")

print(Player.total_players)
````

`total_players` belongs to the class.

There is only one shared value:

`Player.total_players`

It is not a separate value stored in each player.

## Class Methods

A class method operates on the class itself.

Instead of receiving `self`, it receives `cls`.

`cls` refers to the class that called the method.

Class methods are created using the `@classmethod` decorator.

````python
class Player:

    total_players = 0

    def __init__(self, name):
        self.name = name

        Player.total_players += 1

    @classmethod
    def show_total_players(cls):
        print(cls.total_players)


player_1 = Player("Ana")
player_2 = Player("Bruno")

Player.show_total_players()
````

When:

`Player.show_total_players()`

is called, `cls` refers to `Player`.

This allows the method to access data belonging to the class.

The main difference is:

`self`

→ current instance

`cls`

→ current class

## Behavior Without Object or Class State

Some behavior may be related to a class but does not need:

- a specific instance
- instance attributes
- class attributes

In this case, the method does not need `self` or `cls`.

````python
class Temperature:

    @staticmethod
    def is_valid(value):
        return value >= -273.15


print(Temperature.is_valid(20))

print(Temperature.is_valid(-300))
````

`is_valid()` is related to the `Temperature` concept.

However, it does not need to know:

- which temperature object is calling the method
- anything stored in the `Temperature` class

It only needs the value provided as an argument.

## `self` Is a Convention

Without @staticmethod, a method accessed through an instance receives that instance as its first argument automatically.

The name `self` is only a convention. It is not a Python keyword and does not cause the object to be passed.

````python
class Temperature: 
    def show(self, value): 
        print(self) 
        print(value) 

temperature = Temperature() 

temperature.show(20)
````

Conceptually, Python treats the call as:

```python
Temperature.show(temperature, 20)
```

The first parameter could have any valid name:

```python
class Temperature:
    def show(object, value):
        print(object, value)
```

Using self is simply the standard convention because it makes the code immediately understandable to other Python programmers.

Key idea: Python passes the instance automatically; self is the conventional name for the parameter that receives it.

## Comparing the Three Method Types

An instance method receives:

`self`

It operates on a specific object.

A class method receives:

`cls`

It operates on the class.

A static method receives neither `self` nor `cls`.

It behaves like a regular function placed inside a class because it is logically related to that class.

````python
class Example:

    def instance_method(self):
        print("Instance method")

    @classmethod
    def class_method(cls):
        print("Class method")

    @staticmethod
    def static_method():
        print("Static method")


example = Example()

example.instance_method()

Example.class_method()

Example.static_method()
````

The method type depends on what the behavior needs.

If the method needs data from one specific object:

→ instance method

If the method needs data from the class:

→ class method

If the method needs neither:

→ static method

## A Practical Example

A class method can also be used as an alternative way to create objects.

The method belongs to the class and can create an instance with a predefined configuration.

````python
class User:

    def __init__(self, name, active):
        self.name = name
        self.active = active

    @classmethod
    def create_guest(cls, name):
        return cls(name, False)


guest = User.create_guest("Ana")

print(guest.name)
print(guest.active)
````

`create_guest()` does not operate on an existing user.

Instead, it belongs to the `User` class and creates a new instance.

`cls` makes the class method reusable as an alternative constructor.

## Summary

- Instance methods receive `self` and operate on a specific object.
- Class methods receive `cls` and operate on the class.
- `@classmethod` defines a class method.
- Static methods receive neither `self` nor `cls`.
- `@staticmethod` defines a static method.
- Static methods are regular functions logically grouped inside a class.
- The type of method depends on whether the behavior needs an instance, the class, or neither.
- self is only the conventional name for the first parameter of an instance method.
