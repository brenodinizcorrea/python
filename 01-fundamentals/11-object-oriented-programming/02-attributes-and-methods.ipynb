# Attributes and Methods

A program can now create objects from custom classes.

But an empty object is not very useful.

For example, a bank account needs information such as the account holder and balance. It may also need operations such as depositing or withdrawing money.

How can an object store its own data and perform actions related to that data?

## Instance Attributes

An instance attribute is data stored in an individual object.

Different objects from the same class can have different attribute values.

```python
class BankAccount:
    pass


account_1 = BankAccount()
account_2 = BankAccount()

account_1.owner = "Ana"
account_1.balance = 1000

account_2.owner = "Bruno"
account_2.balance = 500

print(account_1.owner)
print(account_2.owner)
```

Each instance has its own attributes.

Changing an attribute of one object does not change the attribute of another object.

```python
account_1.balance = 1200

print(account_1.balance)
print(account_2.balance)
```

The same class can therefore create objects with different states.

The state of an object is the data stored in its attributes.

```python
print(account_1.owner)
print(account_1.balance)

print(account_2.owner)
print(account_2.balance)
```

## Instance Methods

Objects can also have behavior.

An instance method is a function defined inside a class that operates on an instance.

```python
class BankAccount:

    def show_balance(self):
        print(self.balance)


account = BankAccount()

account.balance = 1000

account.show_balance()
```

When we call:

`account.show_balance()`

Python uses the method with `account` as the instance.

The method can access data belonging to that object through `self`.

## `self`

`self` refers to the current instance.

It allows a method to access the attributes and other methods belonging to that instance.

```python
class BankAccount:

    def show_owner(self):
        print(self.owner)


account = BankAccount()

account.owner = "Ana"

account.show_owner()
```

The same method can work with different objects.

Each object becomes `self` when the method is called through that object.

```python
class BankAccount:

    def show_owner(self):
        print(self.owner)


account_1 = BankAccount()
account_2 = BankAccount()

account_1.owner = "Ana"
account_2.owner = "Bruno"

account_1.show_owner()
account_2.show_owner()
```

The method definition uses `self`:

`def show_owner(self):`

But we do not provide `self` when calling the method:

`account_1.show_owner()`

Python provides the current instance automatically.

```python
class BankAccount:

    def show_owner(self):
        print(self.owner)


account = BankAccount()
account.owner = "Ana"

account.show_owner()
```

A method can also modify the current instance.

Because `self` refers to the current object, assigning to `self.balance` changes that object's attribute.

```python
class BankAccount:

    def increase_balance(self, amount):
        self.balance = self.balance + amount


account = BankAccount()

account.balance = 1000

account.increase_balance(500)

print(account.balance)
```

The method operates on the object that called it.

`account.increase_balance(500)`

means that the method modifies `account` through `self`.

```python
class BankAccount:

    def increase_balance(self, amount):
        self.balance = self.balance + amount


account_1 = BankAccount()
account_2 = BankAccount()

account_1.balance = 1000
account_2.balance = 500

account_1.increase_balance(200)
account_2.increase_balance(100)

print(account_1.balance)
print(account_2.balance)
```

## Attributes and Methods

Attributes represent data belonging to an instance.

Methods represent behavior that can operate on an instance.

`self` connects the method to the current instance.

A useful mental model is:

Object
→ attributes = its data

Object
→ methods = its behavior

Method
→ `self` = the current object

## Summary

- An instance attribute stores data in an individual object.
- Different instances can have different attribute values.
- An instance method is a function defined inside a class that operates on an instance.
- `self` refers to the current instance.
- `self.attribute` accesses an attribute belonging to that instance.
- `self.method()` calls another method on the current instance.
- Python provides the instance automatically when an instance method is called.
