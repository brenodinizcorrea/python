# Polymorphism

Imagine that a program needs to process different types of notifications.

An email, an SMS, and a push notification all need to be sent, but each one is sent differently.

Now imagine that the program receives several notifications and needs to send all of them.

Without a common interface, the program needs to know the type of each object before deciding which method to call.

As more notification types are added, this code becomes increasingly dependent on the specific classes.

Can we make all notification objects provide the same method, while allowing each class to implement that method in its own way?

````python
class Email: 
    def send_email(self): 
        print("Sending email") 

class SMS: 
    def send_sms(self): 
        print("Sending SMS") 

notifications = [Email(), SMS()] 

for notification in notifications: 
    if isinstance(notification, Email): 
        notification.send_email() 

    elif isinstance(notification, SMS): 
        notification.send_sms()
````

## A Common Interface 

Instead of giving each class a different method, we can give all notification classes the same method: `send()`. 

The implementation can still be different for each class. 

Now the code that processes the notifications only needs to know that every object provides `send()`.

````python
class Email: 
    
    def send(self): 
        print("Sending email") 

class SMS: 
    def send(self): 
        print("Sending SMS") 

notifications = [Email(), SMS()] 

for notification in notifications: 
    notification.send()
````

## Same Interface, Different Behavior 

The objects provide the same operation, but each class implements it differently. 

This is the basic idea of polymorphism: the same interface can take different forms depending on the object.

````python
class Email: 
    def send(self): 
        print("Sending email") 

class SMS: 
    def send(self): 
        print("Sending SMS") 

class Push: 
    def send(self): 
        print("Sending push notification") 

notifications = [Email(), SMS(), Push()] 

for notification in notifications: 
    notification.send()
````

## Polymorphism Through Inheritance 

Inheritance is one way to create this common interface. 

Different classes can inherit from the same parent and override the same method. 

The parent defines the common operation, while each child provides its own implementation.

````python
class Notification: 
    def send(self): 
        print("Sending notification") 

class Email(Notification): 
    def send(self): 
        print("Sending email") 

class SMS(Notification): 
    def send(self): 
        print("Sending SMS") 

notifications = [Email(), SMS()] 

for notification in notifications: 
    notification.send()
````

The loop does not need to know whether the object is an `Email` or an `SMS`. 

It only needs to know that the object provides `send()`. 

Each object determines what happens when `send()` is called.

````python
email = Email() 
sms = SMS() 

email.send() 
sms.send()
````

## Duck Typing 

Polymorphism does not require inheritance. 

Different classes can provide the same method even when they are completely unrelated. 

Python calls this idea **duck typing**. 

The name comes from the idea: 

> "If it walks like a duck and it quacks like a duck, then it must be a duck." 

The important point is not whether the object really belongs to the `Duck` class. 

What matters is whether it provides the behavior we need. 

In Python: 

- we do not need to check the object's class first 
- we can simply try to use the required operation 
- if the object provides that operation, the code can use it

````python
class Email: 
    def send(self): 
        print("Sending email") 

class SMS: 
    def send(self): 
        print("Sending SMS") 

class Printer: 
    def send(self): 
        print("Sending document") 

objects = [Email(), SMS(), Printer()] 

for object in objects: 
    object.send()
````

Here, `Email` and `SMS` do not need to inherit from the same class. 

The loop does not ask: 

- "Is this an `Email`?" 
- "Is this an `SMS`?" 

It only relies on the fact that both objects provide `send()`. 

This is duck typing: **the behavior matters more than the object's type.**

## When the Interface Is Missing 

Duck typing also means that Python does not automatically check whether an object provides the required method. 

If the method does not exist, the call fails when the code reaches it.

````python
class Email: 
    def send(self): 
        print("Sending email") 

class Report: 
    def show(self): 
        print("Showing report") 

objects = [Email(), Report()] 

for object in objects: 
    object.send()
````

The `Email` object provides `send()`.

The `Report` object does not.

When Python reaches `Report`, it raises an `AttributeError` because that object does not provide the required operation.

## `isinstance()`

Sometimes the program really needs to know the type of an object.

`isinstance()` checks whether an object is an instance of a class or one of its subclasses.

````python
class Employee: 
    pass 

class Manager(Employee): 
    pass 

employee = Employee() 
manager = Manager() 

print(isinstance(employee, Employee)) 
print(isinstance(employee, Manager)) 

print(isinstance(manager, Manager)) 
print(isinstance(manager, Employee))
````

Because `Manager` inherits from `Employee`, a `Manager` object is also considered an instance of `Employee`.

````python
class Employee: 
    pass 

class Manager(Employee): 
    pass 

class Developer(Employee): 
    pass 

employee = Employee() 
manager = Manager() 
developer = Developer() 

print(isinstance(manager, (Manager, Developer))) 
print(isinstance(developer, (Manager, Developer))) 
print(isinstance(employee, (Manager, Developer))) 
````

## Polymorphism and `isinstance()`

Polymorphism focuses on **behavior**.

`isinstance()` focuses on **type**.

For example, polymorphism allows the code to simply call:

```python
notification.send()
```

instead of deciding which method to call based on the object's class.

isinstance() is useful when the program genuinely needs different logic for different types.

````python
class Email: 
    def send(self): 
        print("Sending email") 

class SMS: 
    def send(self): 
        print("Sending SMS") 

notifications = [Email(), SMS()] 

for notification in notifications: 
    notification.send()
````

## Summary 

- Polymorphism means **many forms**. 
- Different objects can provide the same operation with different implementations. 
- A common interface allows code to work with different object types uniformly. 
- Inheritance can be used to create a common interface. 
- Python also supports polymorphism through duck typing. - Duck typing focuses on whether an object provides the required behavior. 
- `isinstance()` explicitly checks an object's type. 
- Polymorphism focuses on behavior; `isinstance()` focuses on type.
