# Special Methods

So far, we have created objects with their own data and methods.

But custom objects do not automatically know how to behave in many situations.

For example, what should happen when we:

- print an object?
- use `len()`?
- compare two objects?
- add two objects?
- use an object with `with`?

Python already knows how these operations work for built-in objects.

How can we define these behaviors for our own objects?

```python
class Playlist:

    def __init__(self, songs):
        self.songs = songs


playlist = Playlist(["Song A", "Song B", "Song C"])

print(playlist)
```

The object can be printed, but Python does not know how we want our `Playlist` to be represented.

Python provides **special methods** that allow a class to define how an object behaves in specific situations.

Special methods usually have names surrounded by double underscores.

```python
class Playlist:

    def __init__(self, songs):
        self.songs = songs

    def __str__(self):
        return f"Playlist with {len(self.songs)} songs"


playlist = Playlist(["Song A", "Song B", "Song C"])

print(playlist)
```

## `__str__`

`__str__` defines the string representation of an object.

When `print()` receives the object, Python uses its `__str__()` method.

```python
class Product:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name}: R$ {self.price}"


product = Product("Keyboard", 150)

print(product)
```

## `__len__`

`__len__` defines what `len()` should return for an object.

The method must return an integer.

```python
class Playlist:

    def __init__(self, songs):
        self.songs = songs

    def __len__(self):
        return len(self.songs)


playlist = Playlist(["Song A", "Song B", "Song C"])

print(len(playlist))
```

The important idea is that we do not normally call `__len__()` ourselves.

We use:

`len(playlist)`

and Python uses the special method defined by the class.

```python
class Team:

    def __init__(self, players):
        self.players = players

    def __len__(self):
        return len(self.players)


team = Team(["Ana", "Bruno", "Carlos"])

print(len(team))
print(team.__len__())
```

## `__eq__`

`__eq__` defines how objects of a class behave when compared with `==`.

The method receives the other object being compared.

```python
class Product:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __eq__(self, other):
        return self.name == other.name and self.price == other.price


product_1 = Product("Keyboard", 150)
product_2 = Product("Keyboard", 150)
product_3 = Product("Mouse", 80)

print(product_1 == product_2)
print(product_1 == product_3)
```

Without `__eq__`, `==` does not automatically mean that two custom objects have the same data.

By defining `__eq__`, we decide what equality means for our objects.

```python
class Product:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __eq__(self, other):
        return self.name == other.name and self.price == other.price


product_1 = Product("Keyboard", 150)
product_2 = Product("Keyboard", 150)

print(product_1 is product_2)
print(product_1 == product_2)
```

`is` and `==` answer different questions.

- `is` checks whether two variables refer to the same object.
- `==` checks whether two objects are considered equal.

`__eq__` controls the behavior of `==`.

## `__add__`

Special methods can also define how operators work with custom objects.

`__add__` defines the behavior of the `+` operator.

```python
class Money:

    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Money(self.value + other.value)


money_1 = Money(100)
money_2 = Money(50)

total = money_1 + money_2

print(total.value)
```

Here, Python can use:

`money_1 + money_2`

because the class defines `__add__()`.

The operator provides a natural way to express the operation between the objects.

```python
class Money:

    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Money(self.value + other.value)

    def __str__(self):
        return f"R$ {self.value}"


money_1 = Money(100)
money_2 = Money(50)

total = money_1 + money_2

print(total)
```

## Special Methods and `with`

We have already seen another example of special methods in context managers.

An object used with `with` can define:

- `__enter__()` for entering the context
- `__exit__()` for leaving the context

```python
class MyContext:

    def __enter__(self):
        print("Entering the context")

    def __exit__(self, exc_type, exc_value, traceback):
        print("Leaving the context")


with MyContext():
    print("Inside the context")
```

The same general idea appears in all these examples:

Python performs an operation, and the object's special methods define how the object participates in that operation.

For example:

- `print(obj)` → `__str__()`
- `len(obj)` → `__len__()`
- `obj1 == obj2` → `__eq__()`
- `obj1 + obj2` → `__add__()`
- `with obj` → `__enter__()` and `__exit__()`

## Special Methods Are Not Called Like Regular Methods

Special methods can be called directly because they are normal methods.

However, their main purpose is to let Python call them automatically through its syntax and built-in operations.

Usually, we write:

`len(playlist)`

instead of:

`playlist.__len__()`

The special method is the mechanism behind the operation.

## Summary

- Special methods are methods with names such as `__str__`, `__len__`, and `__eq__`.
- Python recognizes these methods for specific operations and protocols.
- `__str__` defines the string representation used by `print()`.
- `__len__` defines the behavior of `len()`.
- `__eq__` defines the behavior of `==`.
- `__add__` defines the behavior of `+`.
- `__enter__` and `__exit__` define the behavior of a context manager.
- Special methods allow custom objects to participate naturally in Python operations.
- They are usually used indirectly through Python syntax or built-in functions.
