# Classes and Objects

Your program needs to represent multiple customers.

Each customer exists as an individual entity in the system.

At first, you might represent each customer using separate dictionaries:

```python
customer_1 = {
    "name": "Ana",
    "age": 25
}

customer_2 = {
    "name": "Bruno",
    "age": 30
}
```

This works, but as the program grows, you may need to represent many customers, products, orders, and other entities with their own data and behavior.

How can we define a reusable way to create objects that represent the same type of entity?

## Objects

In Python, an object is an entity that exists in memory.

Python values are objects.

For example, a number, a string, a list, and a dictionary are all objects.

````python
number = 10
name = "Ana"
products = ["Laptop", "Mouse"]

print(type(number))
print(type(name))
print(type(products))
````

An object has a type.

The type defines what kind of object it is.

For example:

10 → int object

"Ana" → str object

["Laptop", "Mouse"] → list object

## Classes

A class defines a new type of object.

It describes a category of objects that can be created.

For example, Python already provides classes such as:

int

str

list

dict

We can also create our own classes.

````python
class Customer:
    pass
````

The `class` keyword creates a new class.

`Customer` is now a type that Python knows about.

However, the class itself is not yet a customer.

It is the definition used to create customer objects.

## Creating Objects

An object created from a class is called an instance.

We create an instance by calling the class.

````python
class Customer:
    pass


customer_1 = Customer()

print(customer_1)
print(type(customer_1))
````

`Customer` is the class.

`customer_1` is an object created from that class.

More than one object can be created from the same class.

````python
customer_1 = Customer()

customer_2 = Customer()

customer_3 = Customer()

print(customer_1)
print(customer_2)
print(customer_3)
````

Each variable refers to a different object.

All of these objects are instances of the same class.

````python
print(type(customer_1))
print(type(customer_2))

print(customer_1 is customer_2)
````

## Class vs Object

A class defines a type.

An object is an individual instance of that type.

For example:

Class:
- Customer

Objects:
- customer_1
- customer_2
- customer_3

We can think about the relationship like this:

Class → defines what kind of object can exist

Object → an individual instance created from the class

````python
class Product:
    pass


product_1 = Product()

product_2 = Product()

print(type(product_1))
print(type(product_2))
````

## Summary

- An object is an entity that exists in memory.
- Python values such as integers, strings, lists, and dictionaries are objects.
- Every object has a type.
- A class defines a type of object.
- The `class` keyword creates a class.
- Calling a class creates an object.
- An object created from a class is called an instance.
- Multiple instances can be created from the same class.
