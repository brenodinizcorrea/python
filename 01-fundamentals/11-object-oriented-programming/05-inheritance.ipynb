# Inheritance

Imagine that you are building a system for different types of employees.

A manager and a developer are both employees.

They may share information such as:

- name
- salary

They may also share behaviors such as displaying their information.

However, each type of employee can have additional characteristics and behaviors.

For example:

```text
Employee
├── Manager
│   └── team_size
└── Developer
    └── programming_language
```

## The Parent Class

A class can be used as a more general starting point for other classes.

In this example, `Employee` represents the general concept shared by different types of employees.

````python
class Employee:

    def __init__(self, name):
        self.name = name

    def show_name(self):
        print(self.name)
````

## Creating a Child Class

A new class can inherit from another class.

The parent class is placed inside parentheses when defining the child class.

````python
class Employee:

    def __init__(self, name):
        self.name = name

    def show_name(self):
        print(self.name)


class Manager(Employee):

    pass

manager = Manager('Ana')

manager.show_name()
````

`Manager` inherits from `Employee`.

This means that a `Manager` object can use the behavior defined by `Employee`.

The child class does not need to redefine everything from the parent class.

The `Manager` class does not define `__init__()` or `show_name()`.

Python looks for these methods in `Manager`.

When they are not found there, Python can use the inherited behavior from `Employee`.

## Extending the Parent Class

A child class can add information that does not belong to every object of the parent class.

For example, a manager can have a `team_size`.

A regular employee does not necessarily need this information.

````python
class Employee:

    def __init__(self, name):
        self.name = name


class Manager(Employee):

    def __init__(self, name, team_size):
        self.name = name
        self.team_size = team_size

manager = Manager('Laura', 8)

print(manager.name)
print(manager.team_size)
````

````python
class Employee:

    def __init__(self, name):
        self.name = name


class Manager(Employee):

    def __init__(self, name, team_size):
        #self.name = name
        self.team_size = team_size

manager = Manager('Laura', 8)

print(manager.name)
print(manager.team_size)
````

The code works, but the child class is repeating part of the parent's initialization:

```python
self.name = name
```

As classes become more complex, repeating parent logic can become difficult to maintain.

Python provides `super()` to access behavior from the parent class.

## Reusing Parent Behavior

`super()` gives access to the parent class behavior.

The child class can use the parent's `__init__()` and then add its own state.

````python
class Employee:

    def __init__(self, name):
        self.name = name


class Manager(Employee):

    def __init__(self, name, team_size):
        super().__init__(name)

        self.team_size = team_size

manager = Manager('Laura', 8)

print(manager.name)
print(manager.team_size)
````

When creating a `Manager`:

```python
manager = Manager("Ana", 5)
```

the initialization happens conceptually in two steps:

1. `super().__init__(name)` initializes the part inherited from `Employee`.
2. `self.team_size = team_size` initializes the additional state of `Manager`.

This avoids repeating the parent's initialization logic.

## Parent Constructor Changes

If the parent constructor changes its parameters, the child constructor may also need to change.

````python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary


class Manager(Employee):

    def __init__(self, name, salary, team_size):
        super().__init__(name, salary)

        self.team_size = team_size

manager = Manager('Laura', 8000, 10)

print(manager.name)
print(manager.salary)
print(manager.team_size)
````

````python
class Employee:

    def __init__(self, name, salary, email):
        self.name = name
        self.salary = salary
        self.email = email


class Manager(Employee):

    def __init__(self, name, salary, email, team_size):
        super().__init__(name, salary, email)

        self.team_size = team_size

manager = Manager('Laura', 8000, 'laura@email.com', 10)

print(manager.name)
print(manager.salary)
print(manager.email)
print(manager.team_size)
````

## Reusing Parent Methods

A child class can also use methods inherited from its parent.

````python
class Employee:

    def __init__(self, name):
        self.name = name

    def show_info(self):
        print(self.name)


class Developer(Employee):

    def __init__(self, name, programming_language):
        super().__init__(name)

        self.programming_language = programming_language


developer = Developer('Bruno', 'Python')

developer.show_info()
````

`Developer` does not define `show_info()`.

The method is inherited from `Employee`.

The child class can therefore reuse behavior without copying the method.

## Changing Inherited Behavior

Sometimes a child class should behave differently from its parent.

In this case, the child class can define a method with the same name as an inherited method.

This replaces the inherited version for objects of that child class.

This is called **overriding**.

````python
class Employee:

    def show_info(self):
        print("Employee")


class Manager(Employee):

    def show_info(self):
        print("Manager")


employee = Employee()
manager = Manager()

employee.show_info()
manager.show_info()
````

## Reusing and Extending Behavior

Overriding does not always mean completely replacing the parent behavior.

A child class can reuse the parent method with `super()` and then add additional behavior.

````python
class Employee:
    
    def show_info(self):
        print("Employee information")


class Manager(Employee):

    def show_info(self):
        super().show_info()

        print("Manager information")


manager = Manager()

manager.show_info()
````

````python
class Employee:
    
    def show_info():
        print("Employee information")


class Manager(Employee):

    @staticmethod
    def show_info():
        Employee.show_info()

        print("Manager information")


manager = Manager()
manager.show_info()
````

## Another Example - Specialized Objects

Inheritance is useful when one class represents a more specialized version of another.

A `Car` can be considered a specialized type of `Vehicle`.

The child class inherits the general behavior and adds its own characteristics.

````python
class Vehicle:

    def __init__(self, brand):
        self.brand = brand

    def show_brand(self):
        print(self.brand)


class Car(Vehicle):

    def __init__(self, brand, doors):
        super().__init__(brand)

        self.doors = doors

    def show_info(self):
        self.show_brand()

        print(self.doors)


car = Car("Toyota", 4)

car.show_info()
````

`Car` reuses the general concept represented by `Vehicle`.

It inherits:

- `brand`
- `show_brand()`

And adds:

- `doors`
- `show_info()`

Inheritance helps organize classes when specialized objects share common characteristics and behavior.

## Summary

- Inheritance allows a class to reuse behavior from another class.
- The more general class is commonly called the parent or base class.
- The specialized class is commonly called the child or subclass.
- A child class can inherit attributes and methods from its parent.
- A child class can add its own attributes and methods.
- A child class can override an inherited method by defining a method with the same name.
- `super()` allows a child class to access behavior from its parent.
- `super()` is useful when a child class extends parent behavior instead of repeating it.
