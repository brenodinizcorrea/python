# Exercise 1 - Common Interface

Create three classes:

- `Email`
- `SMS`
- `Push`

Each class must have a `send()` method with a different message.

Create a list containing one object of each class.

Use one `for` loop to call `send()` for every object.

```python
class Email():
    def send(self):
        print("Sending email...")

class SMS():
    def send(self):
        print('Sending SMS...')

class Push():
    def send(self):
        print('Sending push notification...')

notifications = [Email(), SMS(), Push()]

for notification in notifications:
    notification.send()
```

# Exercise 2 — Polymorphism Through Inheritance

Create a parent class called `Notification` with a `send()` method.

Create two child classes:

- `Email`
- `SMS`

Override `send()` in each child class with a different message.

Create a list containing an `Email` and an `SMS`.

Use one `for` loop to call `send()` for every object.

```python
class Notification():
    def send(self):
        pass

class Email(Notification):
    def send(self):
        print('Sending Email...')

class SMS(Notification):
    def send(self):
        print('Sending SMS...')

notifications = [Email(), SMS()]

for notification in notifications:
    notification.send()
```

# Exercise 3 — Duck Typing

Create two unrelated classes:

- `Printer`
- `Email`

Both classes must have a `send()` method.

Do not use inheritance.

Create a list containing one `Printer` and one `Email`.

Use one `for` loop to call `send()` for every object.

The loop should not check the type of the objects.

```python
class Printer():
    def send(self):
        print('Printing...')

class Email():
    def send(self):
        print('Email...')

objects = [Printer(), Email()]

for obj in objects:
    obj.send()
```

# Exercise 4 - isinstance()

Create these classes:

- `Vehicle`
- `Car(Vehicle)`
- `Motorcycle(Vehicle)`

Create one object of each class.

Use `isinstance()` to check:

- whether the car is a `Car`
- whether the car is a `Vehicle`
- whether the motorcycle is a `Car`
- whether the motorcycle is a `Vehicle`

Print all four results.

```python
class Vehicle():
    pass

class Car(Vehicle):
    pass

class Motorcycle(Vehicle):
    pass

vehicle = Vehicle()
car = Car()
motorcycle = Motorcycle()

print(isinstance(car, Car))
print(isinstance(car, Vehicle))

print(isinstance(motorcycle, Car))
print(isinstance(motorcycle, Vehicle))
```

# Exercise 5 - Polymorphism + isinstance()

Create these classes:

- `Employee`
- `Developer(Employee)`
- `Manager(Employee)`

Both `Developer` and `Manager` must have a `show_role()` method with a different message.

Create one developer and one manager.

First, use polymorphism to call `show_role()` for both objects using one loop.

Then, use `isinstance()` to check whether each object is a `Manager`.

Print the results.

```python
class Employee():
    pass

class Developer(Employee):
    def show_role(self):
        print('Developer')

class Manager(Employee):
    def show_role(self):
        print('Manager')

developer = Developer()
manager = Manager()

for user in (developer, manager):
    user.show_role()

print(isinstance(developer, Manager))
print(isinstance(manager, Manager))
```
