# Exercise 1 — Instance Method

Create a `Playlist` class.

Each playlist should store a `name` and a `songs` count.

Create a `show_info()` instance method that prints the playlist name and number of songs.

Create two playlists with different values and call the method on both objects.

```python
class Playlist():

    def __init__(self, name, count_songs):
        self.name = name
        self.count_songs = count_songs

    def show_info(self):
        print(self.name)
        print(self.count_songs)

playlist_01 = Playlist('playlist_01', 21)
playlist_02 = Playlist('playlist_02', 5)

playlist_01.show_info()
playlist_02.show_info()
```

# Exercise 2 — Class Method

Create an `Employee` class with a class attribute called `total_employees`, initially set to `0`.

Use `__init__` to increment `total_employees` every time a new employee is created.

Create a class method called `get_total_employees()` that returns the total number of employees.

Create three employees and print the total using the class method.

```python
class Employee():

    total_employees = 0

    @staticmethod
    def __init__():
        Employee.total_employees += 1

    @classmethod
    def get_total_employees(cls):
        print(cls.total_employees)

employee_1 = Employee()
employee_2 = Employee()

print(employee_1)

Employee.get_total_employees()
```

# Exercise 3 — Static Method

Create a `Password` class with a static method called `is_valid()`.

The method should receive a password and return `True` if it has at least 8 characters, otherwise return `False`.

Call the method through the class with two different passwords.

```python
class Password():

    @staticmethod
    def is_valid(password):
        return len(password) >= 8

print(Password.is_valid('aaaaaaaa'))
print(Password.is_valid('aaaaaaa'))
```

# Exercise 4 — Understanding self

Create a `Score` class with an instance method called `show()`.

The method should receive a score as an argument and print both the object and the score.

Create an instance and call the method.

Then create another version of the class where the first parameter is called `object` instead of `self`. Verify that the method still works.

```python
class Score():

    def show(self, score):
        print(self)
        print(score)

score_1 = Score()

score_1.show(9)

class Score():

    def show(object, score):
        print(object)
        print(score)

score_1 = Score()

score_1.show(9)
```

# Exercise 5 — Choosing the Method Type

Create a `LibraryBook` class.

The class should:

- store the title of each book;
- have an instance method that prints the title;
- keep track of how many books have been created;
- provide a class method that returns the total number of books;
- provide a static method that checks whether a given page count is valid (`pages > 0`).

Create two books and demonstrate all three types of methods.

```python
class LibraryBook():

    count_books = 0

    def __init__(self, title):
        self.title = title

        LibraryBook.count_books += 1

    def show_title(self):
        print(self.title)

    @classmethod
    def show_count_books(cls):
        return cls.count_books

    @staticmethod
    def count_pages_is_valid(count_pages):
        return count_pages > 0

book_1 = LibraryBook('title_1')
book_2 = LibraryBook('title_2')
book_3 = LibraryBook('title_3')

book_3.show_title()

print(LibraryBook.show_count_books())

print(LibraryBook.count_pages_is_valid(9))
print(LibraryBook.count_pages_is_valid(-9))
```
