# Exercise 1 — Instance Attributes

A program needs to represent books.

Create a class called `Book`.

Create two instances:

- `book_1`
- `book_2`

Give each object its own `title` and `pages` attributes with different values.

Print the attributes of both objects.

```python
class Book():
    pass

book_1 = Book()
book_2 = Book()

book_1.title = 'title_1'
book_1.pages = 123

book_2.title = 'title_2'
book_2.pages = 321

print(book_1.title)
print(book_1.pages)
print(book_2.title)
print(book_2.pages)
```

# Exercise 2 — Independent State

Create a class called `Player`.

Create two instances:

- `player_1`
- `player_2`

Give each player a `score` attribute with a different value.

Then change only `player_1.score`.

Print both scores to verify that changing one object does not change the other.

```python
class Player():
    pass

player_1 = Player()
player_2 = Player()

player_1.score = 23
player_2.score = 45

print(player_1.score)
print(player_2.score)

player_1.score = 98

print(player_1.score)
print(player_2.score)
```

# Exercise 3 — Instance Method

Create a class called `Temperature`.

Create an instance called `temperature`.

Give the object a `value` attribute.

Create an instance method called `show_value()` that prints the object's temperature value.

Call the method.

```python
class Temperature():

    def show_value(self):
        print(self.value)

temperature = Temperature()

temperature.value = 32

temperature.show_value()
```

# Exercise 4 — Using `self`

Create a class called `Employee`.

Create an instance method called `show_name()`.

The method should print the employee's `name` attribute using `self`.

Create two employees with different names and call `show_name()` for both.

```python
class Employee():

    def show_name(self):
        print(self.name)

employee_1 = Employee()
employee_2 = Employee()

employee_1.name = 'Joao'
employee_2.name = 'Jose'

employee_1.show_name()
employee_2.show_name()
```

# Exercise 5 — Modify Instance State

Create a class called `Counter`.

Each counter should have a `value` attribute.

Create an instance method called `increase()` that adds `1` to the current value using `self`.

Create two counters with different starting values.

Call `increase()` twice on the first counter and once on the second counter.

Print both values.

```python
class Counter():

    def increase(self):

        self.value += 1

counter_1 = Counter()
counter_2 = Counter()

counter_1.value = 10
counter_2.value = 100

print(counter_1.value)
print(counter_2.value)

counter_1.increase()
counter_1.increase()
counter_2.increase()

print(counter_1.value)
print(counter_2.value)
```
