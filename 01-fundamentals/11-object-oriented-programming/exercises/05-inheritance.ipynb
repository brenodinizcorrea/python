# Exercise 1 — Reusing a Parent Method

Create a class `Animal` with a `show_type()` method that prints `"Animal"`.

Create a class `Dog` that overrides `show_type()` and uses `super()` to execute the parent's method first.

Then print `"Dog"`.

````python
class Animal():

    def show_type(self):
        print('Animal')

class Dog(Animal):

    def show_type(self):
        super().show_type()

        print('Dog')

dog_1 = Dog()

dog_1.show_type()
````

# Exercise 2 — `super()` and `self`

Create a class `Employee` with a `show_name()` method that prints the employee's name.

Create a class `Manager` that overrides `show_name()` and uses `super()`.

Create a manager named `"Ana"` and call `show_name()`.

Do not pass `self` manually to `super()`.

````python
class Employee():

    def __init__(self, name):
        self.name = name

    def show_name(self):
        print(self.name)

class Manager(Employee):

    def show_name(self):
        super().show_name()

        print('Another print just to check')

manager = Manager('Ana')
manager.show_name()
````

# Exercise 3 — Additional Parameters

Create a class `Employee` with a `show_info(self, department)` method.

The method should print the employee's name and department.

Create a class `Manager` that overrides `show_info()` and reuses the parent's method with `super()`.

Create a manager named `"Carlos"` in the `"Sales"` department.

````python
class Employee():

    def __init__(self, name):
        self.name = name

    def show_info(self, department):
        print(f'{self.name}: {department}')

class Manager(Employee):

    def __init__(self, name):
        super().__init__(name)

    def show_info(self, department):
        super().show_info(department)

manager_1 = Manager('Carlos')

manager_1.show_info('Sales')
````

# Exercise 4 — Fixing a `super()` Call

A `Report` class has a method that receives a format.

A `SalesReport` overrides this method, but its call to `super()` is incomplete.

Fix the code so that the parent method receives the format correctly.

```python
class Report:

    def show(self, format):
        print(f"Report format: {format}")


class SalesReport(Report):

    def show(self, format):
        super().show()
        print("Sales report")


report = SalesReport()

report.show("PDF")
```

````python
class Report:

    def show(self, format):
        print(f"Report format: {format}")


class SalesReport(Report):

    def show(self, format):
        super().show(format)
        print("Sales report")


report = SalesReport()

report.show("PDF")
````

# Exercise 5 — Extending Parent Behavior

Create a class `Notification` with a `send()` method that prints:

`Sending notification`

Create a class `EmailNotification` that overrides `send()`.

The `EmailNotification` should:

1. Use `super()` to execute the parent's behavior.
2. Then print the email address being used.

Create an `EmailNotification` for `"ana@example.com"` and call `send()`.

````python
class Notification():

    def send(self):
        print('Sending notification')

class EmailNotification(Notification):

    def __init__(self, email):
        self.email = email

    def send(self):
        super().send()

        print(self.email)

email_notification_01 = EmailNotification('ana@example.com')

email_notification_01.send()
````
