# Exercise 1 — Initial State

A program needs to represent movies.

Create a class called `Movie`.

Use `__init__` to receive:

- `title`
- `year`

Store both values as instance attributes.

Create a movie and print its attributes.

```python
class Movie():

    def __init__(self, title, year):
        self.title = title
        self.year = year

movie_example = Movie('movie_title', 2021)

print(movie_example.title)
print(movie_example.year)
```

# Exercise 2 — Different Initial States

Create a class called `Course`.

Use `__init__` to receive:

- `name`
- `duration`

Store both values as instance attributes.

Create two course objects with different values.

Print the attributes of both objects.

```python
class Course():

    def __init__(self, name, duration):
        self.name = name
        self.duration = duration

course_1 = Course('course_1', 205)
course_2 = Course('course_2', 35)

print(course_1.name)
print(course_1.duration)
print(course_2.name)
print(course_2.duration)
```

# Exercise 3 — Changing Object State

Create a class called `InventoryItem`.

Use `__init__` to receive:

- `name`
- `quantity`

Create an object with an initial quantity.

Then change the `quantity` attribute after the object has been created.

Print the quantity before and after the change.

```python
class InventoryItem():

    def __init__(self, name, quantity):
        self.name = name
        self.quantity = quantity

item_1 = InventoryItem('item_a', 2)

print(item_1.name)
print(item_1.quantity)

item_1.quantity = 31

print(item_1.name)
print(item_1.quantity)
```

# Exercise 4 — Local Variable vs Instance Attribute

Create a class called `Circle`.

Use `__init__` to receive a `radius`.

Inside `__init__`:

1. Create a local variable called `diameter`.
2. Calculate the diameter using the radius.
3. Store the radius as an instance attribute.
4. Store the calculated diameter as an instance attribute.

Create a `Circle` object and print its `radius` and `diameter`.

```python
class Circle():

    def __init__(self, radius):
        self.radius = radius

        diameter = 2 * radius

        self.diameter = diameter

circle_1 = Circle(2.3)

print(circle_1.radius)
print(circle_1.diameter)
```

# Exercise 5 — Temporary Calculation

Create a class called `ShoppingItem`.

Use `__init__` to receive:

- `product`
- `price`
- `quantity`

Inside `__init__`:

1. Calculate the total price using a local variable.
2. Store `product`, `price`, and `quantity` as instance attributes.
3. Store the calculated total as an instance attribute called `total`.

Create an object and print all its instance attributes.

```python
class ShoppingItem():

    def __init__(self, product, price, quantity):
        total_price = price * quantity

        self.product = product
        self.price = price
        self.quantity = quantity

        self.total = total_price

item_1 = ShoppingItem('item_a', 3.9, 8)

print(item_1.product)
print(item_1.price)
print(item_1.quantity)
print(item_1.total)
```
