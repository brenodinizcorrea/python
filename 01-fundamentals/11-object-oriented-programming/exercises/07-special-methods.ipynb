# Exercise 1 - `__str__`

Create a `Book` class with:

- `title`
- `author`

Implement `__str__()` so that printing the object displays:

`Title by Author`

Create one book and print it.


```python
class Book():

    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __str__(self):
        return f'{self.title} by {self.author}'

book_1 = Book('title_1', 'author_1')

print(book_1)
```

# Exercise 2 - `__len__`

Create a `ShoppingCart` class with a `products` attribute containing a list.

Implement `__len__()` so that `len(cart)` returns the number of products.

Create a cart with four products and print its length.


```python
class ShoppingCart():

    def __init__(self, products):
        self.products = products

    def __len__(self):
        return len(self.products)

cart = ShoppingCart(['product_1', 'product_2', 'product_3'])

print(len(cart))
```

# Exercise 3 - `__eq__`

Create a `Movie` class with:

- `title`
- `year`

Implement `__eq__()` so that two movies are considered equal when both attributes are equal.

Create three movies and compare them using `==`.


```python
class Movie():

    def __init__(self, title, year):
        self.title = title
        self.year = year

    def __eq__(self, other):
        return self.title == other.title and self.year == other.year

movie_1 = Movie('title_1', 2025)
movie_2 = Movie('title_1', 2025)
movie_3 = Movie('title_2', 2025)
movie_4 = Movie('title_1', 2024)

print(movie_1 == movie_1)
print(movie_1 == movie_2)
print(movie_1 == movie_3)
print(movie_1 == movie_4)
```

# Exercise 4 - `__add__`

Create a `Score` class with a `value` attribute.

Implement `__add__()` so that adding two `Score` objects returns a new `Score` containing the sum of their values.

Create two scores, add them, and print the resulting value.

```python
class Score():

    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value

score_1 = Score(20)
score_2 = Score(23)

print(score_1 + score_2)
```

# Exercise 5 - Combine Special Methods

Create a `Playlist` class with a `songs` attribute.

Implement:

- `__str__()` to display the number of songs
- `__len__()` to return the number of songs

Create a playlist and use both:

- `print(playlist)`
- `len(playlist)`

```python
class Playlist():

    def __init__(self, songs):
        self.songs = songs

    def __str__(self):
        return str(len(self.songs))

    def __len__(self):
        return len(self.songs)

playlist = Playlist(['song_1', 'song_2', 'song_3'])

print(playlist)
len(playlist)
```
