# Exercise 2 — Multiple Instances

A program needs to represent different vehicles.

Create a class called `Vehicle`.

Then create four instances from this class:

- `car`
- `motorcycle`
- `truck`
- `bus`

Print each instance.

```python
class Vehicle():
    pass

car = Vehicle()
motorcycle = Vehicle()
truck = Vehicle()
bus = Vehicle()

print(car)
print(motorcycle)
print(truck)
print(bus)
```

# Exercise 3 — Identify the Type

Create a class called `Book`.

Create an instance called `book`.

Use `type()` to verify that `book` is an instance of the `Book` class.

```python
class Book():
    pass

book_1 = Book()

print(book_1)
print(type(book_1) is Book)
```

# Exercise 4 — Independent Instances

Create a class called `Ticket`.

Create two instances:

- `ticket_a`
- `ticket_b`

Check whether they refer to the same object using `is`.

Print the result.

```python
class Ticket():
    pass

ticket_a = Ticket()
ticket_b = Ticket()

print(ticket_a is ticket_b)
```

# Exercise 5 — One Class, Several Instances

A delivery system needs to represent individual packages.

Create a class called `Package`.

Create three instances:

- `package_1`
- `package_2`
- `package_3`

Use `type()` to verify that all three objects belong to the same class.

Then use `is` to verify that `package_1` and `package_2` are different objects.

```python
class Package():
    pass

package_1 = Package()
package_2 = Package()
package_3 = Package()

print(type(package_1) is type(package_2))
print(type(package_1) is type(package_3))
print(package_1 is package_2)
```
