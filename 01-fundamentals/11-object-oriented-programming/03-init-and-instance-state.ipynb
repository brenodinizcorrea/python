# `__init__` and Instance State

Objects can store their own data through instance attributes.

However, creating an object and then assigning its initial attributes separately can be inconvenient:

```python
account = BankAccount()

account.owner = "Ana"
account.balance = 1000
```

A bank account should not need to exist without an owner or an initial balance before it can be properly used.

How can an object receive its initial state when it is created?

# Initializing an Object

Python provides the `__init__` method to define what should happen when a new instance is created.

`__init__` can receive values and use them to create the initial attributes of an object.

````python
class BankAccount:

    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance


account = BankAccount("Ana", 1000)

print(account.owner)
print(account.balance)
````

When Python executes:

`BankAccount("Ana", 1000)`

a new instance is created.

The values provided when calling the class are passed to `__init__`.

Inside the method:

`self` refers to the new instance.

The values are then stored as instance attributes.

The process can be represented as:

`BankAccount("Ana", 1000)`

↓

new instance

↓

`__init__(self, "Ana", 1000)`

↓

`self.owner = "Ana"`

`self.balance = 1000`

## Each Instance Has Its Own State

The same class can create multiple objects.

Each object receives its own attribute values and therefore has its own state.

````python
class BankAccount:

    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance


account_1 = BankAccount("Ana", 1000)

account_2 = BankAccount("Bruno", 500)

print(account_1.owner, account_1.balance)

print(account_2.owner, account_2.balance)
````

The attributes belong to each individual object.

Changing the state of one instance does not change the state of another instance.

````python
account_1.balance = 1200

print(account_1.balance)

print(account_2.balance)
````

## `__init__` Is a Special Method

`__init__` follows Python's special method naming pattern.

The double underscores indicate that Python gives this method a special role.

In this case, Python automatically calls `__init__` when a new instance is created.

You do not call `__init__` directly when creating an object.

Instead, you call the class:

`BankAccount("Ana", 1000)`

Python creates the instance and automatically executes its `__init__` method.

## Attributes vs Local Variables

A variable created inside a method is not automatically stored in the object.

Consider:

````python
class Rectangle:

    def __init__(self, width, height):
        area = width * height


rectangle = Rectangle(10, 5)

print(rectangle.width)
````

The code fails because `area` is a local variable.

It exists only while `__init__` is executing.

It does not become an attribute of the object.

To store data in the instance, assign it through `self`.

````python
class Rectangle:

    def __init__(self, width, height):
        self.width = width
        self.height = height
        self.area = width * height


rectangle = Rectangle(10, 5)

print(rectangle.width)

print(rectangle.height)

print(rectangle.area)
````

## Local Variables Can Still Be Useful

A local variable can be used to calculate a value before storing the result in an attribute.

````python
class Rectangle:

    def __init__(self, width, height):
        calculated_area = width * height

        self.area = calculated_area


rectangle = Rectangle(10, 5)

print(rectangle.area)
````

`calculated_area` is used temporarily inside `__init__`.

`self.area` stores the result in the object.

After the method finishes, the local variable no longer exists, but the instance attribute remains available.

## Initial State and Object State

An object's state is represented by the values stored in its attributes.

`__init__` defines the initial state of an object when the object is created.

````python
class Player:

    def __init__(self, name, score):
        self.name = name
        self.score = score


player = Player("Ana", 10)

print(player.name)
print(player.score)

player.score = 25

print(player.score)
````

The initial state was:

`name = "Ana"`

`score = 10`

The state can later change:

`score = 25`

`__init__` defines the initial state, but it does not prevent the object's state from changing later.

## Summary

- `__init__` is a special method used when creating a new instance.
- Python automatically calls `__init__` when a class is used to create an object.
- Arguments passed when creating an object can be used to define its initial state.
- `self.attribute` creates or accesses an instance attribute.
- Each instance has its own state.
- A local variable belongs only to the execution of a method.
- An instance attribute belongs to the object and remains available after the method finishes.
- `__init__` defines the initial state of an object, but the state can change later.
