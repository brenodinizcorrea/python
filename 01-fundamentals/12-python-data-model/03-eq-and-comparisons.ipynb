# `__eq__` and Comparisons

Imagine that you are building a task management system.

Each task has information such as a title and a priority. You want to compare tasks directly:

```python
task_1 == task_2
task_1 < task_2
task_1 > task_2
```

Python already knows how these operations work for built-in types such as integers and strings.

For a custom class, however, Python needs to know what these comparisons should mean.

How can a custom object define equality and ordering so that comparison operators work naturally?

## Equality and Ordering

Comparison operators can represent different kinds of relationships.

For example:

```python
5 == 5
5 < 10
5 > 2
```

For custom objects, we can define what these operators mean through special methods.

## `__eq__`

The `==` operator is associated with the `__eq__()` special method.

Conceptually:

```python
object_1 == object_2
```

works like:

```python
object_1.__eq__(object_2)
```

The left object becomes `self`, while the right object is passed as the other value.

````python
class Task():
    def __init__(self, title, priority):
        self.title = title
        self.priority = priority

    def __eq__(self, other):
        return self.title == other.title


task_1 = Task("Study Python", 1)
task_2 = Task("Study Python", 3)

print(task_1 == task_2)
````

## What Defines Equality?

The class decides which attributes matter for equality.

In the previous example, two tasks are equal when their titles are equal.

The priority does not participate in the equality rule.

Equality is therefore a rule defined by the class, rather than an automatic comparison of every attribute.

## `is` vs `==`

These operators answer different questions.

```python
object_1 is object_2
```

checks whether both variables refer to the same object.

```python
object_1 == object_2
```

checks whether the objects are considered equal according to their equality behavior.

````python
class Task():
    def __init__(self, title, priority):
        self.title = title
        self.priority = priority

    def __eq__(self, other):
        return self.title == other.title

task_1 = Task("Study Python", 1)
task_2 = Task("Study Python", 1)
task_3 = task_1

print(task_1 is task_2)
print(task_1 == task_2)

print(task_1 is task_3)
print(task_1 == task_3)
````

## `<` and `__lt__`

The `<` operator is associated with `__lt__()`.

For example, a task management system could define that a task with a lower priority number comes before a task with a higher priority number.

Conceptually:

```python
task_1 < task_2
```

works like:

```python
task_1.__lt__(task_2)
```

````python
class Task():
    def __init__(self, title, priority):
        self.title = title
        self.priority = priority

    def __lt__(self, other):
        return self.priority < other.priority


task_1 = Task("Study Python", 1)
task_2 = Task("Study SQL", 3)

print(task_1 < task_2)
````

## `>` and `__gt__`

The `>` operator is associated with `__gt__()`.

We can use it to define when one custom object should be considered greater than another.

For example, a product could be greater than another product when its price is higher.

````python
class Product():
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __gt__(self, other):
        return self.price > other.price


product_1 = Product("Keyboard", 200)
product_2 = Product("Monitor", 900)

print(product_1 > product_2)
````

## Comparison Methods

Python provides special methods for the main comparison operators:

    ==  → __eq__()
    !=  → __ne__()
    <   → __lt__()
    <=  → __le__()
    >   → __gt__()
    >=  → __ge__()

Each method defines the behavior of one comparison operation.

## Defining Several Comparisons

A class can implement multiple comparison methods when its objects need to support different operators.

For example, an `Employee` class could define:

- `__eq__()` to compare employees by ID
- `__lt__()` to compare employees by salary
- `__gt__()` to compare employees by salary

The comparison rule should represent a meaningful relationship between the objects.

````python
class Employee():
    def __init__(self, employee_id, salary):
        self.employee_id = employee_id
        self.salary = salary

    def __eq__(self, other):
        return self.employee_id == other.employee_id

    def __lt__(self, other):
        return self.salary < other.salary

    def __gt__(self, other):
        return self.salary > other.salary


employee_1 = Employee(101, 5000)
employee_2 = Employee(102, 7000)

print(employee_1 == employee_2)
print(employee_1 < employee_2)
print(employee_1 > employee_2)
````

## A Repetition Problem

Suppose a class needs to support all ordering comparisons:

    <
    <=
    >
    >=

We could implement all four methods manually.

However, these methods are often logically related.

Python provides `functools.total_ordering` as a convenience for generating the remaining ordering methods from a smaller set of definitions.

## `@total_ordering`

`total_ordering` is a decorator from the `functools` module.

It can generate the remaining ordering methods when a class provides:

- `__eq__()`
- one ordering method, such as `__lt__()`

Example:

```python
from functools import total_ordering

@total_ordering
class Product:
    ...
```

The decorator handles the implementation of the other ordering comparisons.

````python
from functools import total_ordering


@total_ordering
class Product():
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __eq__(self, other):
        return self.price == other.price

    def __lt__(self, other):
        return self.price < other.price


product_1 = Product("Keyboard", 200)
product_2 = Product("Monitor", 200)

print(product_1 == product_2)
print(product_1 < product_2)
print(product_1 <= product_2)
print(product_1 > product_2)
print(product_1 >= product_2)
````

## When to Use `total_ordering`

`total_ordering` is useful when a class needs complete ordering behavior but implementing every comparison method manually would be repetitive.

It is a convenience, not a requirement.

A class can implement all comparison methods explicitly when more control is useful.

## Comparison Rules Are Part of the Class Design

A comparison only makes sense when the class has a clear rule for it.

For example:

- tasks can be ordered by priority
- products can be ordered by price
- employees can be ordered by salary
- dates can be ordered chronologically

The special method does not determine the meaning by itself.

The class defines the rule that makes the comparison meaningful.

## Summary

- `__eq__()` defines equality with `==`.
- `__lt__()` defines the `<` operation.
- `__gt__()` defines the `>` operation.
- Python provides special methods for the main comparison operators.
- `is` checks object identity, while `==` checks equality.
- `@total_ordering` can provide the remaining ordering methods from `__eq__()` and one ordering method.
- Comparison rules should represent a meaningful relationship between the objects.
