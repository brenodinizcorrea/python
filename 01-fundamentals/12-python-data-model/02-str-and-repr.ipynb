# `__str__` and `__repr__`

Imagine that you are building a system that works with many `Product` objects.

When a product is shown to a user, you may want:

```text
Laptop — R$ 5000
```

But when you are debugging the program, you may want more information about the object itself, such as:

```python
Product(name='Laptop', price=5000)
```

The same object can therefore need different representations depending on the situation.

How can Python provide a user-friendly representation and a developer-oriented representation for the same object?

`````python
class Product:

    def __init__(self, name, price):
        self.name = name
        self.price = price


product = Product("Laptop", 5000)

print(product)
`````

Python needs a way to represent an object as text.

Two special methods are commonly used for this:

- `__str__()` provides a user-friendly representation.
- `__repr__()` provides a developer-oriented representation.

`````python
class Product:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name} — R${self.price}"


product = Product("Laptop", 5000)

print(product)
`````

`__str__()` is intended to produce a readable representation of the object.

It is commonly used when the object is converted to a string for presentation.

`````python
class Product:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name} — R${self.price}"


product = Product("Laptop", 5000)

print(str(product))
`````

Calling `str(product)` explicitly requests the string representation.

`print(product)` also uses this representation when `__str__()` is available.

`````python
class Product:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __repr__(self):
        return f"Product(name='{self.name}', price={self.price})"


product = Product("Laptop", 5000)

print(product)
`````

`__repr__()` is intended to provide a more precise representation of an object.

It is especially useful when inspecting objects during development and debugging.

`````python
class Product:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __repr__(self):
        return f"Product(name='{self.name}', price={self.price})"


product = Product("Laptop", 5000)

print(repr(product))
`````

Just as `str(product)` explicitly requests the `__str__()` representation:

```python
repr(product)
````

explicitly requests the `__repr__()` representation.

## When Both Exist

A class can define both methods.

They can represent the same object in different ways.

`````python
class Product:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name} — R${self.price}"

    def __repr__(self):
        return f"Product(name='{self.name}', price={self.price})"


product = Product("Laptop", 5000)

print(product)
print(repr(product))
`````

The difference is mainly about purpose:

```text
__str__
  ↓
readable representation
  ↓
users / presentation

__repr__
  ↓
precise representation
  ↓
developers / debugging
```

## `__repr__` in Collections

The difference becomes especially useful when objects are inside collections.

When Python displays a list containing objects, it uses the objects' `__repr__()` representations.

`````python
class Product:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name} — R${self.price}"

    def __repr__(self):
        return f"Product(name='{self.name}', price={self.price})"


products = [
    Product("Laptop", 5000),
    Product("Mouse", 100)
]

print(products)
`````

Notice that the list uses `__repr__()` for its elements.

This makes `__repr__()` particularly useful when inspecting collections of custom objects.

## When `__str__` Is Missing

If a class does not define `__str__()`, Python can fall back to `__repr__()` when a string representation is requested.

This makes `__repr__()` useful as a general object representation, even when no separate user-facing representation is needed.

`````python
class Product:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __repr__(self):
        return f"Product(name='{self.name}', price={self.price})"


product = Product("Laptop", 5000)

print(product)
`````

A practical rule is:

- Define `__str__()` when you need a representation designed for users.
- Define `__repr__()` to make objects useful when inspecting and debugging them.
- If both are useful, define both.

## Summary

- `__str__()` defines a user-friendly string representation.
- `__repr__()` defines a developer-oriented representation.
- `str(object)` requests the `__str__()` representation.
- `repr(object)` requests the `__repr__()` representation.
- `print(object)` normally uses `__str__()`.
- Collections use `__repr__()` for their elements.
- When `__str__()` is not defined, Python can fall back to `__repr__()`.
