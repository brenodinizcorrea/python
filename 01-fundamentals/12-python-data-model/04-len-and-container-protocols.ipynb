# `__len__` and Container Protocols

Imagine that you are building a shopping cart.

The cart stores products internally, but you want the object to behave naturally when used by the rest of the program.

For example, you want to write:

```python
len(cart)
"Keyboard" in cart
cart[0]
```

Python already knows what these operations mean for lists, tuples, and other collections.

For a custom class, however, Python needs to know how these operations should work.

How can a custom object provide collection-like behavior?

## Container Protocols

Python defines protocols that describe how objects should behave for certain operations.

A class does not need to inherit from `list` to support operations commonly used with collections.

It can implement the special methods that Python expects for those operations.

## `__len__`

The `len()` function asks an object for its logical size.

Conceptually:

```python
len(object)
```

works through:

```python
object.__len__()
```

The method must return an integer.

````python
class ShoppingCart():

    def __init__(self, products):
        self.products = products

    def __len__(self):
        return len(self.products)


cart = ShoppingCart(["Keyboard", "Mouse", "Monitor"])

print(len(cart))
````

## `__contains__`

The `in` operator checks whether an item belongs to an object.

Conceptually:

    item in object

can use:

    object.__contains__(item)

A custom class can implement `__contains__()` to define what membership means for its objects.

````python
class ShoppingCart():

    def __init__(self, products):
        self.products = products

    def __contains__(self, product):
        return product in self.products


cart = ShoppingCart(["Keyboard", "Mouse", "Monitor"])

print("Mouse" in cart)
print("Laptop" in cart)
````

## `__getitem__`

The `[]` syntax is also connected to a special method.

Conceptually:

```python
object[index]
```

works through:

```python
object.__getitem__(index)
```

This allows a custom object to define how items are accessed using square brackets.

````python
class ShoppingCart():

    def __init__(self, products):
        self.products = products

    def __getitem__(self, index):
        return self.products[index]


cart = ShoppingCart(["Keyboard", "Mouse", "Monitor"])

print(cart[0])
print(cart[1])
print(cart[2])
````

## Indexing

When `__getitem__()` receives an integer, the class can interpret it as an index.

For example:

```python
cart[0]
cart[1]
cart[2]
```

The implementation can delegate the operation to the internal list.

````python
class Playlist():

    def __init__(self, songs):
        self.songs = songs

    def __getitem__(self, index):
        return self.songs[index]


playlist = Playlist(["Song A", "Song B", "Song C"])

print(playlist[0])
print(playlist[2])
````

## Slices

`__getitem__()` does not only receive integers.

When a slice is used, Python passes a `slice` object to `__getitem__()`.

For example:

```python
playlist[1:3]
```

can be handled by the same method.

The internal list can perform the actual slicing.

````python
class Playlist():

    def __init__(self, songs):
        self.songs = songs

    def __getitem__(self, index):
        return self.songs[index]


playlist = Playlist(["Song A", "Song B", "Song C", "Song D"])

print(playlist[1:3])
````

## One Method, Different Arguments

The same `__getitem__()` method can therefore receive different kinds of values.

For example:

```python
playlist[0]
```

passes an integer.

```python
playlist[1:3]
```

passes a `slice` object.

The method can delegate both operations to the internal collection.

## Combining the Protocols

A custom object becomes more collection-like when it supports several operations.

For example, a playlist could support:

```python
len(playlist)
"Song B" in playlist
playlist[0]
```

Each operation is connected to a different special method.

````python
class Playlist():

    def __init__(self, songs):
        self.songs = songs

    def __len__(self):
        return len(self.songs)

    def __contains__(self, song):
        return song in self.songs

    def __getitem__(self, index):
        return self.songs[index]


playlist = Playlist(["Song A", "Song B", "Song C"])

print(len(playlist))
print("Song B" in playlist)
print(playlist[0])
````

## The Object Defines the Behavior

The operations are the same, but their behavior depends on the object's type.

For example:

```python
len(list)
len(tuple)
len(custom_object)
```

All use the same `len()` function.

Python asks each object how it should respond to that operation.

## Protocols and Polymorphism

This is another example of polymorphic behavior.

Different types can respond to the same operation in different ways.

For example:

```python
len(object_1)
len(object_2)
len(object_3)
```

The caller does not need to know the internal implementation.

It only needs the object to provide the expected behavior.

## A Custom Collection Is Not a List

Implementing these methods does not turn the object into a `list`.

The object remains an instance of its own class.

It simply provides some of the behaviors that Python collections provide.

This allows a class to expose only the operations that make sense for its domain.

## Summary

- `__len__()` defines the behavior of `len()`.
- `__contains__()` defines membership behavior for `in`.
- `__getitem__()` defines access through `[]`.
- `__getitem__()` can receive indexes and slices.
- A class can implement several protocols without inheriting from `list`.
- Protocols allow different types to respond to the same operation.
- This is a form of polymorphic behavior.
