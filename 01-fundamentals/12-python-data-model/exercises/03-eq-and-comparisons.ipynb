# Exercise 1 — Compare Tasks

Create a `Task` class with:

- `title`
- `priority`

Implement `__lt__()` so that a task is considered smaller when its priority number is lower.

Create three tasks with different priorities.

Use `<` to compare them.

```python
class Task():

    def __init__(self, title, priority):
        self.title = title
        self.priority = priority

    def __lt__(self, other):
        return self.priority < other.priority

priority_1 = Task('task_1', 1)
priority_2 = Task('task_2', 2)
priority_3 = Task('task_3', 3)

print(priority_1 < priority_2)
print(priority_1 < priority_3)
print(priority_2 < priority_3)
```

# Exercise 2 — Compare Employees

Create an `Employee` class with:

- `name`
- `salary`

Implement `__gt__()` so that one employee is greater than another when their salary is higher.

Create two employees and compare them using `>`.

```python
class Employee():

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def __gt__(self, other):
        return self.salary > other.salary

employee_1 = Employee('name_1', 1000)
employee_2 = Employee('name_2', 3000)

print(employee_1 > employee_2)
```

## Exercise 3 — Complete Product Ordering

Create a `Product` class with:

- `name`
- `price`

Use `@total_ordering`.

Implement:

- `__eq__()` to compare products by price
- `__lt__()` to compare products by price

Create two products and test all of these operators:

    ==
    !=
    <
    <=
    >
    >=

```python
from functools import total_ordering

@total_ordering
class Product():

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __eq__(self, other):
        return self.price == other.price

    def __lt__(self, other):
        return self.price < other.price

product_1 = Product('product_1', 10)
product_2 = Product('product_2', 20)

print(product_1 == product_2)
print(product_1 < product_2)
print(product_1 <= product_2)
print(product_1 > product_2)
print(product_1 >= product_2)
```
