# Exercise 1 — User Representation

Create a `BankAccount` class with:

- `owner`
- `balance`

Implement `__str__()` so that:

```python
account = BankAccount("Ana", 2500)

print(account)
```

displays:

```text
Ana — Balance: R$2500
```

````python
class BankAccount():

    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def __str__(self):
        return f'{self.owner} - Balance: R${self.balance}'

account = BankAccount("Ana", 2500)

print(account)
````

# Exercise 2 — Developer Representation

Create a `User` class with:

- `username`
- `age`

Implement `__repr__()` so that:

```python
user = User("bruno", 30)

print(repr(user))
```

displays:

```text
User(username='bruno', age=30)
```

````python
class User():

    def __init__(self, username, age):
        self.username = username
        self.age = age

    def __repr__(self):
        return f'User(username="{self.username}", age={self.age})'

user = User("bruno", 30)

print(repr(user))
````

# Exercise 3 — Both Representations

Create a `Book` class with:

- `title`
- `author`
- `year`

Implement both:

- `__str__()` for a reader-friendly representation.
- `__repr__()` for a developer-oriented representation.

Then create a list with two books and print the list.

Observe which representation Python uses for the books inside the list.

````python
class Book():

    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year

    def __str__(self):
        return f'{self.title}, {self.author}, {self.year}'

    def __repr__(self):
        return f'Book(title="{self.title}", author="{self.author}", year={self.year})'

book_1 = Book('title_1', 'author_1', 2021)
book_2 = Book('title_2', 'author_2', 2022)

book_list = [book_1, book_2]

print(book_list)
````
