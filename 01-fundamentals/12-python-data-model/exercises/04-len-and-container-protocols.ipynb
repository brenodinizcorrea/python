# Exercise 1 — Course

Create a `Course` class that receives a list of students.

Implement:

- `__len__()` to return the number of students
- `__contains__()` to check whether a student belongs to the course

Create a course with at least four students.

Test both `len()` and `in`.

````python
class Course():

    def __init__(self, students):
        self.students = students

    def __len__(self):
        return len(self.students)

    def __contains__(self, person):
        return person in self.students

course_1 = Course(['student_1', 'student_2'])

print(len(course_1))
print('student_1' in course_1)
print('student_3' in course_1)
````

# Exercise 2 — Inventory

Create an `Inventory` class that receives a list of products.

Implement `__getitem__()` so that products can be accessed using an index.

Then test:

```python
inventory[0]
inventory[1]
```

Also test a slice that returns more than one product.

````python
class Inventory():

    def __init__(self, products):
        self.products = products

    def __getitem__(self, item):
        return self.products[item]

inventory_1 = Inventory(['product_1', 'product_2'])

print(inventory_1[0])
print(inventory_1[1])
print(inventory_1[:1])
````

# Exercise 3 — Team Collection

Create a `Team` class that receives a list of members.

Implement all three methods:

- `__len__()`
- `__contains__()`
- `__getitem__()`

Then make the following operations work:

    len(team)
    "Ana" in team
    team[0]

Finally, use a slice to retrieve part of the team.

````python
class Team():

    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)

    def __contains__(self, item):
        return item in self.members

    def __getitem__(self, item_index):
        return self.members[item_index]

team_1 = Team(['member_1', 'member_2'])

print(len(team_1))
print('member_1' in team_1)
print(team_1[0])
print(team_1[0:])
````
