# Exercise 1 — Custom Number

Create a `Number` class with a `value` attribute.

Implement `__add__()` so that two `Number` objects can be added with `+`.

Then create two objects and use the `+` operator to calculate their result.

````python
class Number():

    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value

number_1 = Number(3)
number_2 = Number(5)

print(number_1 + number_2)
````

# Exercise 2 — Custom Collection

Create a `Team` class that receives a list of members.

Implement `__len__()` so that the following works:

```python
team = Team(["Ana", "Bruno", "Carlos"])

print(len(team))
```

The result should be `3`.

Then add a second team with a different number of members and test it as well.

````python
class Team():

    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)

team = Team(['Ana', 'Bruno', 'Carlos'])

print(len(team))
````

# Exercise 3 — Custom Comparison

Create a `Product` class with `name` and `price` attributes.

Implement `__eq__()` so that two products are considered equal when both their name and price are equal.

Then create three products and compare them using `==`.

````python
class Product():

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __eq__(self, other):
        return self.name == other.name and self.price == other.price

product_1 = Product('product', 1000)
product_2 = Product('product', 1000)
product_3 = Product('product_3', 1000)
product_4 = Product('product', 2000)

print(product_1 == product_1)
print(product_1 == product_2)
print(product_1 == product_3)
print(product_1 == product_4)
````
