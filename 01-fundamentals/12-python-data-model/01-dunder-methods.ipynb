# Dunder Methods

Imagine that you are building a system that works with different kinds of objects.

You want to be able to write:

```python
print(product)
len(cart)
product_1 == product_2
score_1 + score_2
```

and have Python know what these operations should mean for each object.

For built-in types, Python already knows how to perform these operations.

But for a class that you create yourself, Python needs a way to know how the object should behave.

How can a custom class tell Python what to do when one of these built-in operations is used?

## Dunder Names

Dunder is short for **double underscore**.

Dunder methods are special methods whose names begin and end with double underscores.

Examples include:

- `__str__`
- `__len__`
- `__eq__`
- `__add__`

`````python
class Product:

    def __str__(self):
        return "Product"


product = Product()

print(product)
`````

The important part is not the name itself. 

Python gives specific meanings to certain special method names. 

For example, `__str__` defines how an object is converted to a user-readable string.

## Operations and Special Methods

Python operations are connected to specific special methods.

For example:

```python
print(product)
```

uses the object's string representation.

```python
len(product)
```

uses the object's length behavior.

```python
product_1 == product_2
```

uses the object's equality behavior.

`````python
class Playlist:

    def __init__(self, songs):
        self.songs = songs

    def __len__(self):
        return len(self.songs)


playlist = Playlist(["Song A", "Song B", "Song C"])

print(len(playlist))
`````

Conceptually, Python can be understood as connecting:

```python
len(playlist)
```

to:

```python
playlist.__len__()
```

The first form is the normal Python operation.

The second form exposes the special method involved.


## The Left Operand

Operators follow the same idea.

Consider:

```python
score_1 + score_2
```

Conceptually, Python can be understood as using:

```python
score_1.__add__(score_2)
```

The left operand becomes `self`.

The right operand is passed to the method.


`````python
class Score:

    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value


score_1 = Score(20)
score_2 = Score(30)

print(score_1 + score_2)
`````

The operation:

```python
score_1 + score_2
````

provides the two objects involved.

Conceptually:

```text
self  → score_1
other → score_2
```

This is why the method can use data from both objects.

## Protocols

A **protocol** is a set of behaviors that an object can provide.

For example, the length protocol allows Python to ask an object for its length through:

```python
len(object)
```

An object participates in this protocol by providing the expected special method:

```python
__len__()
```

This means that Python does not need to know the concrete class of every object in advance.

It can simply use the expected operation.

For example:

```python
len(value)
````

works with different types because those types provide length behavior.

`````python
print(len("hello"))
print(len([10, 20, 30]))
print(len({"name": "Ana"}))
`````

Different types can respond to the same operation in different ways.

This is one of the places where Python's polymorphism becomes visible.

The operation stays the same:

```python
len(value)
```

but the object's type determines how the operation behaves.

## Built-in Types Use Special Methods Too

Special methods are not only for classes we create.

Python's built-in types also implement them.

For example:

```python
10 + 5
```

can be understood conceptually as:

```python
(10).__add__(5)
```

The same idea applies to strings, lists, dictionaries, and other objects.


`````python
print((10).__add__(5))
print("hello".__len__())
print([1, 2].__add__([3, 4]))
`````

You normally should not write operations this way.

Prefer:

```python
10 + 5
len("hello")
[1, 2] + [3, 4]
```

The special methods are the mechanism behind these behaviors.

Python's syntax provides the convenient interface.

## Implicit Calls

Special methods are usually called **implicitly** by Python.

For example:

```python
print(product)
```

rather than:

```python
product.__str__()
```

And:

```python
len(cart)
```

rather than:

```python
cart.__len__()
```

This is what makes custom objects work naturally with Python's built-in operations.

The general idea is:

```text
Python operation
       ↓
expected protocol
       ↓
special method
       ↓
object-specific behavior
```

The object participates in the operation by implementing the corresponding special method.

## Summary

- Dunder means **double underscore**.
- Dunder methods are special methods recognized by Python.
- Python operations and syntax can trigger these methods implicitly.
- `len()` uses length behavior such as `__len__()`.
- `+` can use `__add__()`.
- `==` can use `__eq__()`.
- Different types can implement the same protocol differently.
- Built-in types also use special methods.
- Protocols allow objects to participate in Python's built-in operations.
