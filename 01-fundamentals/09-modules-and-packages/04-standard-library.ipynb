# Standard Library

You need to calculate a square root, work with dates, read a JSON file, or create a file path.

Should you install a package for each of these tasks?

Python already provides many modules for common programming needs. 

These modules are part of the Standard Library and are distributed with Python.

## Built-in functions vs Standard Library

Some functionality is available directly as built-in functions.

Other functionality is provided by modules from the Standard Library and must be imported before use.

```python
print(len([10, 20, 30]))
print(sum([10, 20, 30]))
```

```python
import math

print(math.sqrt(25))
```

The difference is how the functionality is accessed.

`len()` and `sum()` are built-ins.

`sqrt()` belongs to the `math` module, so the module must be imported first.

## Common Standard Library modules

The Standard Library contains modules for many different tasks.

Some commonly used examples are:

- `math` — mathematical operations
- `statistics` — basic statistics
- `datetime` — dates and times
- `pathlib` — filesystem paths
- `json` — JSON data
- `random` — random values

```python
import math
import statistics

values = [10, 20, 30, 40]

print(math.sqrt(100))
print(statistics.mean(values))
```

Different modules provide different groups of related functionality.

You do not need to install these modules separately because they are already part of Python's Standard Library.

## Import only what you need

As with other modules, you can import specific objects from a Standard Library module.

```python
from math import sqrt

print(sqrt(144))
```

The import style changes how the object is accessed, but the object still belongs to the `math` module.

## Working with dates

`datetime` provides objects and operations for dates and times.

```python
from datetime import date, timedelta

today = date.today()
next_week = today + timedelta(days=7)

print(today)
print(next_week)
```

## Working with file paths

`pathlib` provides objects for representing and manipulating filesystem paths.

```python
from pathlib import Path

data_path = Path("data")
file_path = data_path / "records.csv"

print(data_path)
print(file_path)
```

## Standard Library vs third-party packages

The Standard Library comes with Python.

Third-party packages are developed outside Python and normally need to be installed separately.

For example:

- `math` → Standard Library
- `datetime` → Standard Library
- `pathlib` → Standard Library
- `numpy` → third-party package
- `pandas` → third-party package

## Summary

- The Standard Library provides ready-to-use modules for common programming tasks.
- The main skill here is not memorizing every module.
- It is recognizing when Python already provides what you need and knowing which module to look for.
- When you need functionality beyond the Standard Library, third-party packages can provide additional tools.
