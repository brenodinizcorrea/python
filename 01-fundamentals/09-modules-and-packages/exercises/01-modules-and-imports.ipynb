# Exercise 1 — Generate a Random Number

Import the `random` module.

Use `random.randint()` to generate a number between 1 and 20.

Store the result in a variable named `number` and print it.

The output should be one integer between 1 and 20.

`````python
import random

number = random.randint(1, 20)

print(number)
`````

# Exercise 2 — Calculate a Circle

Import the `math` module.

Create a variable named `radius` with the value `4`.

Use `math.pi` to calculate the circle's area and store the result in a variable named `area`.

Print `area`.

The output should be approximately `50.26548245743669`.

`````python
import math

radius = 4

area = math.pi * (radius ** 2)

print(area)
`````

# Exercise 3 — Work with Numbers

Import the `math` module.

Create a variable named `number` with the value `81`.

Use `math.sqrt()` to calculate its square root.

Also use `math.pow()` to calculate 2 raised to the power of 5.

Store the results in `square_root` and `power`.

Print both values.

The expected output is:

9.0
32.0

`````python
import math

number = 81

square_root = math.sqrt(number)

power = math.pow(2, 5)

print(square_root)
print(power)
`````
