# Exercise 1 — Mathematical Operations

Use the `math` module to calculate the square root of 225 and round 7.3 up to the next integer.

Print both results.

````python
import math

print(math.sqrt(225))
print(math.ceil(7.3))
````

# Exercise 2 — Basic Statistics

Given the following values:

values = [12, 15, 18, 21, 24]

Use the `statistics` module to calculate the mean and median.

Print both results.

````python
import statistics

values = [12, 15, 18, 21, 24]

print(statistics.mean(values))
print(statistics.median(values))
````

# Exercise 3 — Dates

A project starts on September 3, 2026.

Use the `datetime` module to calculate the date 30 days after the start date.

Print the resulting date.

````python
import datetime

date_variable = datetime.date(2026, 9, 3)

print(date_variable + datetime.timedelta(days=30))
````

# Exercise 4 — File Paths

Use `pathlib` to build the following path:

```text
data/
└── reports/
    └── sales.csv
```

Create the path using `Path` objects and the `/` operator.

Print the resulting path.

````python
import pathlib

data_path = pathlib.Path('data')

file_path = data_path / 'reports' / 'sales.csv'

print(data_path)
print(file_path)
````

# Exercise 5 — Combine Standard Library Modules

A report should be saved inside a `reports` directory.

Use `datetime` to get today's date and `pathlib` to build a filename using that date.

For example, if today's date is 2026-09-03, the resulting path should be:

reports/report_2026-09-03.txt

Print the resulting path.

Do not create the file.

````python
import datetime
import pathlib

today_date = datetime.date(2026, 9, 3)

data_path = pathlib.Path('reports')

file_path = data_path / ('report_' + str(today_date) + '.txt')

print(today_date)
print(data_path)
print(file_path)
````
