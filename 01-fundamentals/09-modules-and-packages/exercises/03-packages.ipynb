# Exercise 5 — Fix the Package Imports

The package has the following structure:

```text
package_reports/
├── __init__.py
├── sales.py
└── customers.py
```

`sales.py` contains:

```python
def total(price, quantity):
    return price * quantity
```

`customers.py` contains:

```python
def count(customers):
    return len(customers)
```

The following imports are incorrect:

```python
from package_reports.sales import total
from package_reports.customers import count
```

The goal is to use the package interface:

```python
from package_reports import total, count
```

Fix the package so this import works, then calculate:

```python
total(50, 4)
count(["Ana", "Bruno", "Carlos"])
```

Print both results.

````python
from package_reports import total_multi, count

print(total_multi(50, 4))
print(count(["Ana", "Bruno", "Carlos"]))
````
