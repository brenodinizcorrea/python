# Exercise 1 — Shorten a Module Name

Import the `statistics` module using the alias `stats`.

Create a list named `scores` with the values `10, 20, 30, 40, 50`.

Use `stats.mean()` to calculate the average and store it in `average`.

Print `average`.

`````python
import statistics as stats

scores = [10, 20, 30, 40, 50]

average = stats.mean(scores)

print(average)
`````

# Exercise 2 — Calculate with Selected Functions

Import `ceil` and `floor` from the `math` module.

Create a variable named `number` with the value `7.8`.

Use `ceil()` and `floor()` to calculate the values above and below the number.

Store the results in `rounded_up` and `rounded_down`.

Print both values.

`````python
from math import ceil, floor

number = 7.8

rounded_up = ceil(number)
rounded_down = floor(number)

print(rounded_up)
print(rounded_down)
`````

# Exercise 3 — Avoid a Name Conflict

Your program already has a function named `sqrt`.

Import the `sqrt` function from the `math` module using the alias `math_sqrt`.

Then call your own `sqrt()` function and the imported `math_sqrt()` function.

Use this code:

```python
def sqrt(number):
    return f"Custom function: {number}"
````

Print both results.

`````python
from math import sqrt as math_sqrt

def sqrt(number):
    return f"Custom function: {number}"

print(sqrt(25))
print(math_sqrt(25))
`````

# Exercise 4 — Work with Dates

Import `date` and `timedelta` from the `datetime` module.

Create a variable named `start_date` with the date `2026-01-15`.

Add 30 days to it using `timedelta`.

Store the result in `end_date`.

Print `end_date`.

`````python
from datetime import date, timedelta

start_date = date(2026, 1, 15)

end_date = start_date + timedelta(30)

print(end_date)
`````

# Exercise 5 — Make the Import Explicit

The following code uses a wildcard import:

```python
from math import *

number = 16
result = sqrt(number)

print(result)
````

Rewrite the code using an explicit import instead of *.

`````python
from math import sqrt

number = 16
result = sqrt(number)

print(result)
`````
