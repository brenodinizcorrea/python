# Import Variations

## The Problem

You already know how to import a module and access its objects through the module name.

But imagine a module contains dozens of useful objects and your program only needs one of them. Writing the module name every time can make the code unnecessarily repetitive.

There is also another problem: sometimes the original name of an imported object is too long, unclear, or conflicts with a name already used in your program.

## Importing a Specific Object

Instead of importing the entire module and accessing an object through its namespace, Python allows you to import a specific object directly.

The syntax is `from module import object`.

After that, the imported object can be used directly by its name.

```python
from math import sqrt

number = 25
result = sqrt(number)

print(result)
```

The important difference is where the name comes from.

With `import math`, the function is accessed as `math.sqrt()`.

With `from math import sqrt`, the function itself is available as `sqrt()`.

```python
import math

print(math.sqrt(36))

from math import sqrt

print(sqrt(36))
```

## Importing Multiple Objects

You can import more than one object from the same module.

Separate the object names with commas.

Only the specified objects are imported directly into the current namespace.

```python
from math import sqrt, pi

radius = 3
area = pi * radius ** 2

print(sqrt(25))
print(area)
```

## Aliases with `as`

An alias gives an imported module or object another name.

This can make a name shorter or help avoid a naming conflict.

The alias exists only in the current program; it does not rename the original object inside the module.

```python
import math as m

number = 49
result = m.sqrt(number)

print(result)
```

The same idea works when importing a specific object.

The name after `as` becomes the name used in the program.

```python
from math import sqrt as square_root

number = 64
result = square_root(number)

print(result)
```

## Why Use an Alias?

Aliases are useful when a name is inconvenient, too long, or conflicts with another name.

However, aliases should make the code easier to understand. An unnecessarily short or cryptic alias can make the code harder to read.

```python
import math as mathematics

radius = 5
area = mathematics.pi * radius ** 2

print(area)
```

## Wildcard Imports

Python also allows importing everything from a module with `*`.

The syntax is `from module import *`.

This makes the module's imported names directly available without the module prefix.

```python
from math import *

number = 25
result = sqrt(number)

print(result)
```

## Why Wildcard Imports Are Usually Avoided

Wildcard imports can make it unclear where a name came from.

They can also introduce many names into the current namespace and potentially overwrite names that already exist.

Explicit imports make dependencies easier to see and make code easier to understand.

```python
from math import *

sqrt = "my value"

print(sqrt)
```

In this example, the name `sqrt` was first imported from `math`.

The later assignment replaces that name in the current namespace.

The original function inside the `math` module is not changed.

```python
import math

sqrt = "my value"

print(sqrt)
print(math.sqrt(25))
```

# Summary

- `from module import object` imports a specific object directly into the current namespace.
- Multiple objects can be imported from the same module using commas.
- `as` creates an alias for an imported module or object.
- An alias changes the name used in the current program, not the original object.
- Wildcard imports use `from module import *`.
- Wildcard imports can introduce many names and make dependencies less clear.
- Explicit imports are generally easier to read and maintain.
