# Modules and Imports

## The Problem

Imagine that you are building a project with several Python files. You need the same function in five of them, so you copy the code into each file.

At first, this seems harmless. But then you discover a bug in that function. Now you have to find every copy, change every one of them, and hope you did not miss any.

As the project grows, the same problem appears again and again: useful code is duplicated across files, changes have to be repeated manually, and keeping everything consistent becomes increasingly difficult.

There should be a way to write reusable code once and use it wherever it is needed.

## What Is a Module?

A module is a Python file containing code that can be used by another Python program.

A module can contain variables, functions, classes, and other Python definitions.

Python also provides many modules as part of its standard library.

## Importing a Module

The `import` statement makes a module available to your program.

After importing a module, you can access its contents using the module name followed by a dot.

```python
import math

print(math.pi)
print(math.sqrt(25))
```

The dot connects the module name to something defined inside that module.

For example, `math.sqrt` means the `sqrt` function provided by the `math` module.

```python
import math

number = 49
square_root = math.sqrt(number)

print(square_root)
```

## Importing Your Own Module

Modules do not have to come from Python's standard library.

A Python file can also be used as a module by another Python file.

For example, a file named `calculations.py` could contain reusable functions.

Another Python file can import `calculations` and use the function defined inside it.

The module name is based on the filename, without the `.py` extension.

```python
from libraries import calculations

result = calculations.add(10, 20)

print(result)
```

## Module Namespace

When a module is imported, its definitions are accessed through the module's namespace.

This keeps names organized and helps prevent unrelated variables and functions from being mixed together.

```python
import math

print(math.pi)
print(math.sqrt(16))
```

For example, both a program and a module could define a variable with the same name without automatically replacing each other.

The module name provides the context for accessing its variable.

```python
import math

pi = "my value"

print(pi)
print(math.pi)
```

## Importing a Module Once

Python does not normally execute the same module from scratch every time it is imported again during a program's execution.

Once imported, the module is available for reuse.

```python
import math
import math

print(math.sqrt(36))
```

## Module vs. Function

A module is a container for related Python code.

A function is a specific reusable operation defined inside a module or another part of a program.

For example, `math` is a module, while `math.sqrt` is a function provided by that module.

```python
import math

print(type(math))
print(type(math.sqrt))
```

## Importing a Missing Module

Python raises `ModuleNotFoundError` when it cannot find the module you are trying to import.

```python
import module_that_does_not_exist
```

# Summary

- A module is a Python file containing reusable code.
- The `import` statement makes a module available to a program.
- Module contents are accessed through the module's namespace.
- A dot connects a module name to a definition inside that module.
- Python's standard library provides many reusable modules.
- Your own `.py` files can also be imported as modules.
