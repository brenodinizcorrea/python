# Sets

## The Problem

Imagine that a program receives the names of people who entered an event:

Alice
Bob
Alice
Charlie
Bob

If we want to represent the people who entered, we may not want repeated names.

A `list` preserves every occurrence:

`names = ["Alice", "Bob", "Alice", "Charlie", "Bob"]`

We need a collection in which each value appears only once.

## Creating Sets

A set is created by placing its elements between curly braces `{}`.

```python
names = {"Alice", "Bob", "Charlie"}

print(names)
```

A set can contain values of different types.

```python
values = {10, "Python", 3.14}

print(values)
```

A set can also be empty.

There is an important detail: `{}` creates an empty dictionary, not an empty set.

An empty set is created using `set()`.

```python
empty = set()

print(empty)
```

## Sets Contain Unique Values

If repeated values are provided, the set keeps only one occurrence of each value.

```python
numbers = {1, 2, 2, 3, 3, 3}

print(numbers)
```

## Sets as Objects

A set is an object created from the `set` type.

As with the other types we have studied, we can inspect its type and identity.

```python
numbers = {1, 2, 3}

print(type(numbers))
print(id(numbers))
```

## Sets Are Not Indexed by Position

Lists and tuples are ordered collections whose elements can be accessed using positions.

A set does not provide this kind of positional access.

```python
numbers = {10, 20, 30}

print(numbers[0])
```

## Sets Are Mutable

Sets can be modified after they have been created.

For example, an existing element can be removed.

```python
numbers = {1, 2, 3}

numbers.remove(2)

print(numbers)
```

The set object itself was modified.

This is called **mutation**.

We can observe that the identity remains the same.

```python
numbers = {1, 2, 3}

original_id = id(numbers)

numbers.remove(2)

print(numbers)
print(id(numbers) == original_id)
```

## Mutation vs. Reassignment

As with lists and dictionaries, a set can change through mutation or reassignment.

Mutation changes the existing object.

```python
numbers = {1, 2, 3}

print(numbers)
print(id(numbers))

numbers.remove(2)

print(numbers)
print(id(numbers))
```

```python
numbers = {1, 2, 3}

print(numbers)
print(id(numbers))

numbers = {10, 20, 30}

print(numbers)
print(id(numbers))
```

Therefore:

- **Mutation → same object, different contents**
- **Reassignment → different object**

## Two Variables, One Set

Two variables can refer to the same set object.

```python
a = {1, 2, 3}
b = a

print(id(a))
print(id(b))
print(a is b)

a.add(4)

print(a)
print(b)
```

Both variables refer to the same object.

Therefore, a mutation through one variable is visible through the other.

## Two Equal Sets Are Not Necessarily the Same Object

Two sets can contain the same values while still being different objects.

```python
a = {1, 2, 3}
b = {1, 2, 3}

print(a == b)
print(a is b)

print(id(a))
print(id(b))
```

# Summary

- `set` is a Python type used to represent a collection of unique values.
- Sets are objects.
- Sets do not store duplicate values.
- Sets are not indexed by position.
- Sets are mutable.
- Mutation changes an existing set object.
- Reassignment makes a variable refer to another set object.
- Two variables can refer to the same set object.
- A mutation through one variable is visible through another variable referring to the same set.
- Two sets can contain the same values while still being different objects.
- `{}` creates an empty dictionary; `set()` creates an empty set.
