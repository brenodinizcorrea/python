# Boolean Values

## The Problem

Programs often need to represent situations that have only two possible states.

For example, a user may be logged in or not logged in:

logged_in = True

We need a way for Python to represent values such as "yes or no", "true or false", or "on or off".

## Representing Logical States

Python represents logical values using two values:

- `True`
- `False`

Both values belong to the `bool` type.

```python
is_logged_in = True
is_admin = False

print(is_logged_in)
print(is_admin)
```

## Booleans as Objects

`True` and `False` are not just special words.

They are objects created from the `bool` type.

As with the other types we have studied, we can inspect their type and identity.

```python
is_logged_in = True

print(type(is_logged_in))
print(id(is_logged_in))
```

## The Two Boolean Objects

Unlike types such as `int` or `str`, `bool` has only two possible values:

`True` and `False`.

These are the two boolean objects used by Python.

```python
a = True
b = True
c = False

print(id(a))
print(id(b))
print(id(c))
```

Because `a` and `b` both contain `True`, they refer to the same boolean object.

We can verify this using `is`.

```python
a = True
b = True

print(a is b)
```

## Booleans Are Immutable

Boolean objects are immutable.

There is no operation that changes the existing `True` object into `False`.

A variable can, however, be reassigned.

```python
status = True

print(status)
print(id(status))

status = False

print(status)
print(id(status))
```

The original `True` object was not modified.

The variable `status` was reassigned and now refers to the `False` object.

This is the same distinction we saw with other immutable types:

- **Mutation** changes an existing object.
- **Reassignment** makes a variable refer to another object.

For `bool`, mutation is not possible.

## Boolean Values and Types

A boolean is its own type.

It is not a string containing the characters `"True"` or `"False"`.

It is also not an integer value written as `1` or `0`.

```python
a = True
b = "True"
c = 1
d = 0

print(type(a))
print(type(b))
print(type(c))
print(type(d))
```

The values may represent related ideas, but Python treats them as different types.

This distinction becomes important when these values are used in expressions and program logic.

```python
a = True
b = "True"

print(type(a) == type(b))
```

# Summary

- `bool` is the Python type used to represent logical values.
- A boolean can have one of two values: `True` or `False`.
- `True` and `False` are objects.
- There are only two boolean objects.
- Boolean objects are immutable.
- Reassignment can make a variable refer to the other boolean object.
- `True` and `False` are different from the strings `"True"` and `"False"`.
- `bool` is different from `int`, even though boolean values are closely related to integer values in Python.
