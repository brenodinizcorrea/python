# Integers in Python

## The Problem

An order contains 4 boxes with 12 items each.

How can Python represent these quantities and calculate the total number of items?

We need a way to represent whole-number quantities and perform calculations with them.

```python
boxes = 4
items_per_box = 12

total_items = boxes * items_per_box

print(total_items)
```

## Representing Whole-Number Quantities

The quantities in our problem are whole numbers:

- 4 boxes
- 12 items per box
- 48 total items

Python represents whole-number values using the `int` type.

We can assign integer values to variables:

```python
boxes = 4
items_per_box = 12
total_items = 48
```

We can check the type of a value using `type()`

```python
print(type(boxes))
print(type(items_per_box))
print(type(total_items))
```

## Arithmetic Operations

Integers can be manipulated using arithmetic operations.

```python
10 + 5
```

```python
10 - 5
```

```python
4 * 12
```

### Integer Division vs. Regular Division

There is an important difference between `/` and `//`.

Regular division produces a floating-point result:

```python
12 / 4
```

Even though the result is mathematically a whole number, Python represents it as `3.0`.

Integer division produces the whole-number result:

```python
17 // 5
```

# Summary

- `int` represents whole-number values in Python.
- Integers can be stored in variables and used in arithmetic operations.
- `+`, `-`, `*`, and `**` perform common arithmetic operations.
- `/` performs regular division and returns a floating-point result.
- `//` performs floor division.
- The same operations can be applied to integer values stored in variables.
