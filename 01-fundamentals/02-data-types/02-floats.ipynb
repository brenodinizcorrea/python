# Floating-Point Numbers in Python

## The Problem

A product weighs 1.75 kilograms.

How can Python represent this quantity and use it in calculations?

A whole number is not enough to represent the weight accurately, because the quantity contains a fractional part.

````python
weight = 1.75

print(weight)
````

## Representing Fractional Quantities

Python uses the `float` type to represent numbers that can contain a fractional part.

For example:

````python
weight = 1.75
price = 19.90
temperature = 23.5

print(type(weight))
print(type(price))
print(type(temperature))
````

All three values are `float` objects.

Unlike `int`, a `float` can represent a numerical value with a fractional part.

## Arithmetic with Floating-Point Numbers

Floating-point values can be used in arithmetic operations just like integers.

For example, we can calculate the total price of two products:

````python
price = 19.90
quantity = 2

total = price * quantity

print(total)
````

We can also combine floating-point and integer values in the same calculation:

````python
price = 19.90
quantity = 3

total = price * quantity

print(total)
````

The result is a `float` because the calculation involves a floating-point value.

### Basic Operations

Floating-point numbers support the same basic arithmetic operators we saw with integers.

````python
1.5 + 2.5
````

````python
5.5 - 2.25
````

````python
2.5 * 4
````

````python
10.0 / 4
````

## Floating-Point Precision

Floating-point numbers have an important characteristic: some decimal numbers
cannot be represented exactly using a binary representation.

To understand why, it helps to look at how fractions are represented in binary.

### Decimal with Powers of 10

In decimal, the positions after the decimal point represent powers of 10:

```text
0.1 = 1 × 10⁻¹
0.01 = 1 × 10⁻²

### Decimal with Powers of 2

In binary, the positions after the decimal point represent powers of 2:

```text
0.1₂  = 1 × 2⁻¹  = 0.5
0.01₂ = 1 × 2⁻²  = 0.25

This means that some fractions can be represented exactly in binary.

However, some fractions can't be. 0.1 is an example.

Its binary representation is an infinitely repeating sequence:

0.1₁₀ = 0.000110011001100110011...₂

### Finite Amount of Storage

A floating-point number has a finite amount of storage, so Python cannot store this
infinite sequence. Instead, it stores the closest representable floating-point value.

```text
mathematical value:  0.1
stored float:        0.10000000000000000555...

````python
x = 0.1

print(x)
print(format(x, ".20f"))
````

### Why does Python still print `0.1`?

We know that the floating-point value stored for `0.1` is only an approximation.

However, when we use print() normally.

````python
x = 0.1

print(x)
````

This does not mean that the stored value changed to exactly 0.1.

The value stored in x is still the same floating-point approximation.

Python simply chooses a shorter decimal representation when displaying the value.

The important property is that this shorter representation can be converted back
into exactly the same floating-point value:

````python
x = 0.1

print(float("0.1") == x)
````

So there are two different things to keep in mind:

- Stored value: the floating-point approximation kept in memory.
- Displayed value: a short decimal representation chosen by Python.

The displayed 0.1 is therefore not claiming that the internal value is mathematically
exactly 0.1. It is simply the shortest representation that can be used to recover
the same stored floating-point value.

### Why does `0.1 + 0.2` produce `0.30000000000000004`?

````python
a = 0.1
b = 0.2

result = a + b

print(result)
print(result == 0.3)
````

#### The values `0.1` and `0.2` are stored as floating-point approximations.

0.1 → 0.000110011001100110011001100110...₂

0.2 → 0.001100110011001100110011001100...₂

so:

```text
0.1 + 0.2 = 0.000110011001100110011001... + 0.001100110011001100110011...

````python
import struct

def binary_float(value):
    bits = struct.unpack(">Q", struct.pack(">d", value))[0]
    return f"{bits:064b}"

print(binary_float(0.3))
print(binary_float(0.1 + 0.2))
````

Where the difference is:

````python
a = binary_float(0.3)
b = binary_float(0.1 + 0.2)

print(a)
print(b)
print("".join("^" if x != y else " " for x, y in zip(a, b)))
````

#### Why does Python print different values?

Python's `print()` displays the shortest decimal representation that can be
converted back to exactly the same stored floating-point value.

We can verify this using `0.3` and the result of `0.1 + 0.2`:

````python
a = 0.3
b = 0.1 + 0.2

print(float("0.3") == a)
print(float("0.3") == b)
````
