# Tuples

## Introduction

Programs often need to group values together.

For example, we may want to represent the coordinates of a fixed location

## Creating Tuples

A tuple is commonly created by placing its elements between parentheses.

Like lists, tuples can contain multiple values and can contain values of
different types.

```python
coordinates = (40.7, -74.0)
person = ("Alice", 30, "Brazil")
mixed = (10, "Python", 3.14)

print(coordinates)
print(person)
print(mixed)
```

```python
empty = ()

print(empty)
```

## Tuples as Objects

A tuple is an object created from the `tuple` type.

As with other objects, we can inspect its type and identity.

```python
coordinates = (40.7, -74.0)

print(type(coordinates))
print(id(coordinates))
```

## Tuples Are Immutable

Like `int`, `float`, and `str`, tuples are immutable.

Once a tuple has been created, its elements cannot be changed.

```python
coordinates = (40.7, -74.0)

coordinates[0] = 41.0
```

The operation raises an error because the tuple itself cannot be modified.

This makes tuples different from lists, which allow their elements to be
changed.

## Reassignment

Immutability does not prevent a variable from being reassigned.

The tuple object cannot be modified, but the variable can be made to refer to
another tuple.

```python
values = (1, 2, 3)

print(values)
print(id(values))

values = (10, 20, 30)

print(values)
print(id(values))
```

The original tuple was not modified.

Instead, `values` was reassigned to another tuple object.

This is the same distinction we saw with immutable types such as `str`.

## Two Variables, One Tuple

Two variables can refer to the same tuple object.

Because tuples are immutable, neither variable can modify that object.

```python
a = (1, 2, 3)
b = a

print(id(a))
print(id(b))
```

## Summary

- `tuple` is a Python type used to group multiple values.
- Tuples are ordered collections.
- Tuples are objects.
- Tuples are **immutable**.
- A variable can still be reassigned to another tuple.
- Lists and tuples can both group values, but lists are mutable while tuples
  are immutable.
- Two variables can refer to the same tuple object.
