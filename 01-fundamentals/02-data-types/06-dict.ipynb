# Dictionaries

## The Problem

A program needs to store information about a person:

- name: Alice
- age: 30
- country: Brazil

We could store these values in separate variables, but the values belong
together.

A list could group the values, but we would have to remember which position
represents each piece of information.

We need a way to associate each value with a name.

## Creating Dictionaries

A dictionary stores values as **key-value pairs**.

A dictionary is created using curly braces `{}`.

Each key is associated with a value using `:`.

The pairs are separated by commas.

````python
person = {
    "name": "Alice",
    "age": 30,
    "country": "Brazil"
}

print(person)
````

Mostrar também o vazio:

````python
empty = {}

print(empty)
````

## Dictionaries as Objects

A dictionary is an object created from the `dict` type.

As with other Python objects, we can inspect its type and identity.

````python
person = {
    "name": "Alice",
    "age": 30
}

print(type(person))
print(id(person))
````

## Keys and Values

A dictionary is composed of **keys** and **values**.

In the following dictionary:

```text
"name" → "Alice"
"age"  → 30
```

"name" and "age" are keys.

"Alice" and 30 are values.

The key identifies which value we want to associate with it.

````python
person = {
    "name": "Alice",
    "age": 30
}

print(person)
````

## Accessing Values by Key

A value in a dictionary can be accessed using its key.

The key is placed inside square brackets.

````python
person = {
    "name": "Alice",
    "age": 30
}

print(person["name"])
print(person["age"])
````

Unlike a list, we do not need to use a numeric position to identify the
information we want.

The key gives us a meaningful way to access the value.

## Dictionaries Are Mutable

Dictionaries are mutable.

After creating a dictionary, we can change the value associated with an
existing key.

````python
person = {
    "name": "Alice",
    "age": 30
}

person["age"] = 31

print(person)
````

Comprovar identidade:

````python
person = {
    "name": "Alice",
    "age": 30
}

original_id = id(person)

person["age"] = 31

print(person)
print(id(person) == original_id)
````

## Adding a New Key

A dictionary can also receive a new key-value pair after it has been created.

We can assign a value to a key that does not yet exist.

````python
person = {
    "name": "Alice",
    "age": 30
}

person["country"] = "Brazil"

print(person)
````

## Mutation vs. Reassignment

A dictionary can be changed through mutation or a variable can be reassigned.

Mutation changes the existing dictionary object.

Reassignment makes the variable refer to another dictionary object.

````python
person = {
    "name": "Alice",
    "age": 30
}

print(id(person))

person["age"] = 31

print(id(person))
````

````python
person = {
    "name": "Alice",
    "age": 30
}

print(id(person))

person = {
    "name": "Alice",
    "age": 31
}

print(id(person))
````

With mutation:

- the contents change;
- the object remains the same.

With reassignment:

- the variable refers to another object.

## Two Variables, One Dictionary

Two variables can refer to the same dictionary object.

If the dictionary is mutated through one variable, the change is visible
through the other variable.

````python
a = {
    "name": "Alice",
    "age": 30
}

b = a

print(id(a))
print(id(b))

a["age"] = 31

print(a)
print(b)
````

## Two Equal Dictionaries Are Not Necessarily the Same Object

Two dictionaries can contain the same key-value pairs while still being
different objects.

````python
a = {
    "name": "Alice",
    "age": 30
}

b = {
    "name": "Alice",
    "age": 30
}

print(a == b)
print(id(a) == id(b))
````

## Summary

- `dict` is the Python type used to represent key-value relationships.
- Dictionaries are objects.
- A dictionary stores values associated with keys.
- Values can be accessed using their keys.
- Dictionaries are **mutable**.
- Existing values can be changed through their keys.
- New key-value pairs can be added to an existing dictionary.
- Mutation changes the existing dictionary object.
- Reassignment makes a variable refer to another dictionary object.
- Two variables can refer to the same dictionary object.
- Two dictionaries can have the same contents while being different objects.
