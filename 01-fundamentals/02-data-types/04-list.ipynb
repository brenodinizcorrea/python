# Lists

## Introduction

Programs often need to work with multiple values that belong together.

For example, instead of creating a separate variable for every score, we may want to represent all scores as a single object.

## Creating Lists

A list is created by placing its elements between square brackets.

A list can contain multiple values, including values of different types.

```python
numbers = [1, 2, 3]
names = ["Alice", "Bob", "Charlie"]
mixed = [10, "Python", 3.14]

print(numbers)
print(names)
print(mixed)
```

A list can also be created without any elements.

This is called an **empty list**.

```python
empty = []

print(empty)
```

## Lists as Objects

A list is not just a collection of values.

It is an object created from the `list` type.

As with other objects, we can inspect its type and identity.

```python
numbers = [1, 2, 3]

print(type(numbers))
print(id(numbers))
```

## Lists Are Mutable

We have seen that `int`, `float`, and `str` objects are immutable.

Lists behave differently.

A list can be modified after it has been created.

For example, we can replace one of its elements.

```python
numbers = [1, 2, 3]

numbers[0] = 10

print(numbers)
```

The operation above changed the existing list.

This is called **mutation**.

The list object itself was modified rather than replaced by another object.

```python
numbers = [1, 2, 3]

original_id = id(numbers)

numbers[0] = 10

print(numbers)
print(id(numbers) == original_id)
```

## Mutation vs. Reassignment

A list can change in two fundamentally different ways.

**Mutation** changes the existing object.

**Reassignment** makes the variable refer to a different object.

Both operations can produce the same final value, but they have different
effects on the object itself.

```python
numbers = [1, 2, 3]

print(numbers)
print(id(numbers))

numbers[0] = 10

print(numbers)
print(id(numbers))
```

In this example, the list was mutated.

The contents changed, but the `id` remained the same.

The variable still refers to the same list object.

```python
numbers = [1, 2, 3]

print(numbers)
print(id(numbers))

numbers = [10, 2, 3]

print(numbers)
print(id(numbers))
```

Here, the variable was reassigned.

The original list was not changed. A different list object was created, and
`numbers` was made to refer to it.

Therefore:

- **Mutation → same object, different contents**
- **Reassignment → different object, variable points somewhere else**

This distinction becomes especially important when more than one variable
refers to the same object.

If two variables refer to the same mutable object, mutation through one
variable can affect what the other variable observes.

### Two Variables, One List

A particularly important consequence of mutability appears when two variables
refer to the same list.

Consider:

```python
a = [1, 2, 3]
b = a

print(a)
print(b)
print(id(a) == id(b))
```

Both variables refer to the **same list object**.

Therefore, modifying the object through one variable is visible through the
other variable.

```python
a.append(4)

print(a)
print(b)
```

### Two Equal Lists Are Not Necessarily the Same Object

Now consider two lists created independently with the same values.

They have the same contents, but they are different objects.

```python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)
print(id(a) == id(b))
```

```python
a.append(4)

print(a)
print(b)
```

## Summary

- `list` is the Python type used to represent an ordered collection of values.
- Lists are objects.
- Lists are **mutable**.
- Mutation changes an existing list object.
- Reassignment makes a variable refer to another object.
- Two variables can refer to the same list object.
- If two variables refer to the same mutable object, a mutation through one
  variable is visible through the other.
- Two lists can contain the same values while still being different objects.
