# Exercise 1 — Creating Sets

Create three sets:

- one containing three integers;
- one containing three strings;
- one containing values of different types.

Also create an empty set.

Print all four sets.

```python
three_integers = {1, 2, 3}
three_strings = {'string_1', 'string_2', 'string_3'}
three_different_types = {1, 'string', 10.25}
empty = set()

print(three_integers)
print(three_strings)
print(three_different_types)
print(empty)
```

# Exercise 3 — Set Objects

Create the following set:

numbers = {10, 20, 30}

Print:

- the set;
- its type;
- its `id()`.

Then create another variable that refers to the same set:

other = numbers

Print the `id()` of both variables.

Finally, use `is` to check whether they refer to the same object.

```python
numbers = {10, 20, 30}

print(numbers)
print(type(numbers))
print(id(numbers))

other = numbers

print(id(other))
print(other is numbers)
```

# Exercise 4 — Mutation

Create the following set:

numbers = {1, 2, 3}

Store its `id()` in a variable called `original_id`.

Then remove the value `2` from the set.

Print:

- the set before the mutation;
- the set after the mutation;
- the original `id()`;
- the current `id()`.

Did the contents change? Did the object change?

```python
numbers = {1, 2, 3}

original_id = id(numbers)

print(numbers)

numbers.remove(2)

print(numbers)
print(original_id)
print(id(numbers))
```

# Exercise 5 — Two Variables, One Set

Create the following set:

numbers = {1, 2, 3}

Create another variable that refers to the same set:

other = numbers

Print both sets.

Then add the value `4` to the set through `other`.

Print both sets again.

Finally, print the `id()` of both variables.

Observe what happened to `numbers` after the mutation was made through `other`.

```python
numbers = {1, 2, 3}

other = numbers

print(numbers)
print(other)

other.add(4)

print(numbers)
print(other)

print(id(numbers))
print(id(other))
```
