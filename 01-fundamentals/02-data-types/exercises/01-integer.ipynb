# Exercise 2 — Checking the Type

Create the following variables:

- `apples = 10`
- `students = 25`
- `year = 2026`

Use `type()` to verify that all three values are integers.

````python
apples = 10
students = 25
year = 2026

print(type(apples))
print(type(students))
print(type(year))
````

# Exercise 5 — Inspecting Integer Objects

Create two variables with integer values:

```python
a = 42
b = 42
```

Use `type()` and `id()` to inspect both variables.

````python
a = 42
b = 42

print(type(a))
print(type(b))

print(id(a))
print(id(b))

print(id(a) == id(b))
````
