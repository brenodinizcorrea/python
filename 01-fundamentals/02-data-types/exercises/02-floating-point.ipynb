# Exercise 1 — Creating Floats

Create three variables representing:

- a person's height: `1.75` meters;
- the price of a product: `29.90`;
- a temperature: `23.5` degrees.

Print the three values.

Then use `type()` to verify that all three are `float` objects.

`````python
person_height = 1.75
price_of_product = 29.90
temperature = 23.5

print(person_height)
print(price_of_product)
print(temperature)

print(type(person_height))
print(type(price_of_product))
print(type(temperature))
`````

## Exercise 5 — Floating-Point Precision in a Calculation

Create two variables:

```python
a = 0.1
b = 0.2
```

Calculate their sum and store the result in a third variable.

Then:

- Print the result.
- Compare the result with 0.3.
- Print the result using format(result, ".20f").

`````python
a = 0.1
b = 0.2

print(a + b)
print(0.3)

print(format((a + b), '.20f'))
print(format(0.3, '.20f'))
`````
