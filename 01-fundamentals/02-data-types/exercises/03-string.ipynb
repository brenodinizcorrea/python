# Exercise 2 — Inspecting Strings

Create the following variables:

```python
language = "Python"
country = "Brazil"
```

For each variable, print:

its value;
its type;
its `id()`.

Confirm that both objects have type str.

````python
language = 'Python'
country = 'Brazil'

print(language)
print(country)

print(type(language))
print(type(country))

print(id(language))
print(id(country))
````

# Exercise 3 — Reassignment

Create a variable containing the string `"Python"`.

Print its value and `id()`.

Then reassign the variable to `"Java"`.

Print its value and `id()` again.

Observe what changed after the reassignment.

Did the variable change its value, its object, or both?

````python
language = 'Python'

print(id(language))

language = 'Java'

print(id(language))
````
