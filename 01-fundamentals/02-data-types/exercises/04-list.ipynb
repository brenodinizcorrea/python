# Exercise 2 — Inspecting a List

Create the following list:

```python
numbers = [10, 20, 30]
```

Print:
- the list;
- its type;
- its id().

Then create another variable referring to the same list:

```
other = numbers
```

````python
numbers = [10, 20, 30]

print(numbers)
print(type(numbers))
print(id(numbers))

other = numbers

print(other)
print(type(other))
print(id(other))
````

# Exercise 3 — Mutating a List

Create the following list:

```python
numbers = [10, 20, 30]
```

Store its `id()` in a variable called `original_id`.

Then change the first element of the list to `100`

Print:

- the list before the change;
- the list after the change;
- the original `id()`;
- the current `id()`.

Did the list change its contents? Did it change its identity?

````python
numbers = [10, 20, 30]

original_id = id(numbers)

print(numbers)

numbers[0] = 100

print(numbers)
print(original_id)
print(id(numbers))
````

# Exercise 4 — Mutation vs. Reassignment

Create a list:

```python
numbers = [1, 2, 3]
```

First, mutate the list so that it becomes:

```
[10, 2, 3]
```

Observe its `id()`

Then create another list:

```
[100, 2, 3]
```

by reassigning the variable. Print the `id()` after each operation.

Which operation kept the same object?

Which operation made the variable refer to another object?

````python
numbers = [1, 2, 3]

print(id(numbers))

numbers[0] = 10

print(id(numbers))

numbers = [100, 2, 3]

print(id(numbers))
````

## Exercise 5 — Two Variables, One List

Create a list and assign it to two variables:

```python
a = [1, 2, 3]
b = a
```

Use `id()` to confirm that both variables refer to the same object.

Then change the first element through `a`.

Print both `a` and `b`.

````python
a = [1, 2, 3]
b = a

print(id(a))
print(id(b))

a[0] = 23

print(a)
print(b)

print(id(a))
print(id(b))
````
