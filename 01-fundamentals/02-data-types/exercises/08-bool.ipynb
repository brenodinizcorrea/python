# Exercise 1 — Creating Boolean Values

Create two variables:

- `is_logged_in` containing `True`;
- `is_admin` containing `False`.

Print both variables.

# Exercise 2 — Boolean Objects

Create three variables:

- `a` containing `True`;
- `b` containing `True`;
- `c` containing `False`.

Print the `id()` of all three variables.

Then use `is` to check:

- whether `a` and `b` refer to the same object;
- whether `a` and `c` refer to the same object.

```python
a = True
b = True
c = False

print(id(a))
print(id(b))
print(id(c))

print(a is b)
print(a is c)
```

# Exercise 3 — Boolean vs. Other Types

Create four variables:

- `a` containing `True`;
- `b` containing `"True"`;
- `c` containing `1`;
- `d` containing `0`.

Print the value and `type()` of each variable.

Observe how Python represents each value.

```python
a = True
b = 'True'
c = 1
d = 0

print(type(a))
print(type(b))
print(type(c))
print(type(d))
```

# Exercise 4 — Reassignment

Create a variable called `status` containing `True`.

Print its value and `id()`.

Then reassign `status` to `False`.

Print its value and `id()` again.

Did the boolean object change, or was the variable reassigned to another object?

```python
status = True

print(id(status))

status = False

print(id(status))
```

# Exercise 5 — Same Boolean Object

Create three variables:

- `first` containing `True`;
- `second` containing `True`;
- `third` containing `False`.

Print the `id()` of all three.

Then use `is` to verify that:

- `first` and `second` refer to the same object;
- `first` and `third` do not refer to the same object.

Finally, reassign `third` to `True` and check its `id()` again.

```python
first = True
second = True
third = False

print(id(first))
print(id(second))
print(id(third))

print(first is second)
print(first is third)

third = True

print(id(third))
```
