# Exercise 2 — Tuples as Objects

Create the following tuple:

```python
coordinates = (10, 20)
```

Print:

1. the tuple;
2. its type;
3. its `id()`.

Then create another variable referring to the same tuple:

```python
other = coordinates
```

Print the `id()` of both variables.

Do they refer to the same object?

````python
coordinates = (10, 20)

print(coordinates)
print(type(coordinates))
print(id(coordinates))

other = coordinates

print(id(other))
print(coordinates is other)
````

# Exercise 5 — Reassignment

Create the following tuple:

```python
values = (1, 2, 3)
```

Print its value and `id()`.

Then reassign the variable:

```python
values = (10, 20, 30)
```

Print the value and `id()` again.

Answer:

1. Did the original tuple change?
2. Did the variable start referring to another object?
3. How is this different from what happened when we mutated a list?

````python
values = (1, 2, 3)

print(id(values))

values = (10, 20, 30)

print(id(values))
````
