# Exercise 3 — Mutation

Create the following dictionary:

```python
person = {
    "name": "Alice",
    "age": 30
}
```

Store its `id()` in a variable called `original_id`.

Then change the value associated with `"age"` to `31`.

Print:

- the dictionary before the change;
- the dictionary after the change;
- the original `id()`;
- the current `id()`.

Did the contents change? Did the object change?

````python
person = {
    'name': 'Alice',
    'age': 30
}

original_id = id(person)

print(person)

person['age'] = 31

print(person)
print(original_id)
print(id(person))
````

# Exercise 5 — Two Variables, One Dictionary

Create the following dictionary:

```python
person = {
    "name": "Alice",
    "age": 30
}
```

Create another variable that refers to the same dictionary:

```python
other = person
```

Print the `id()` of both variables.

Then change the `"age"` value through `other`:

```python
other["age"] = 31
```

Print both `person` and `other`.

Answer:

1. Did both variables show the change?
2. Did both variables refer to the same object?
3. Why was the change visible through `person` even though the mutation was made through `other`?

````python
person = {
    'name': 'Alice',
    'age': 30
}

other = person

print(id(person))
print(id(other))

other['age'] = 31

print(person)
print(other)
````
