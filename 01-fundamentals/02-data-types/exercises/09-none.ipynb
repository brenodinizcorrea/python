# Exercise 1 — None and Other Values

Create five variables:

- `a` containing `None`;
- `b` containing `0`;
- `c` containing an empty string;
- `d` containing an empty list;
- `e` containing `False`.

Print the value and `type()` of each variable.

```python
a = None
b = 0
c = ''
d = []
e = False

print(type(a))
print(type(b))
print(type(c))
print(type(d))
print(type(e))
```

# Exercise 2 — The None Object

Create three variables:

- `first` containing `None`;
- `second` containing `None`;
- `third` containing `None`.

Print the `id()` of all three variables.

Then use `is` to verify whether they refer to the same object.

```python
first = None
second = None
third = None

print(id(first))
print(id(second))
print(id(third))

print(first is second)
print(second is third)
```

# Exercise 3 — Reassignment

Create a variable called `result` containing `None`.

Print its value and `id()`.

Then reassign `result` to the string `completed`.

Print its value and `id()` again.

Observe what happened to the variable and to the original `None` object.

```python
result = None

print(id(result))

result = 'completed'

print(id(result))
```

# Exercise 4 — Different Types

Create the following variables:

- `none_value` containing `None`;
- `number` containing `10`;
- `text` containing `Python`;
- `numbers` containing the list `[1, 2, 3]`.

Print the `type()` of each variable.

Then print the `id()` of each variable.

```python
none_value = None
number = 10
text = 'Python'
numbers = [1, 2, 3]

print(type(none_value))
print(type(number))
print(type(text))
print(type(numbers))

print(id(none_value))
print(id(number))
print(id(text))
print(id(numbers))
```

# Exercise 5 — Temporary Absence of a Value

Create a variable called `username` containing `None`.

Print its value and `type()`.

Then reassign `username` to the string `Alice`.

Print its value and `type()` again.

Finally, print the `id()` before and after the reassignment.

Observe the difference between the object originally referenced by `username` and the object referenced after the reassignment.

```python
username = None

print(type(username))
print(id(username))

username = 'Alice'

print(type(username))
print(id(username))
```
