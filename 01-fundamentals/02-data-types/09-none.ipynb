# None

## The Problem

Programs sometimes need to represent the absence of a value.

For example, suppose a program has a variable that will receive a value later.

At the beginning, there may be no value to store.

Python needs a way to explicitly represent this situation.

## Representing the Absence of a Value

`None` represents the absence of a value.

It does not mean:

- zero;
- an empty string;
- an empty list;
- `False`.

It represents a different situation: there is no value.

```python
a = None
b = 0
c = ""
d = []
e = False

print(a)
print(b)
print(c)
print(d)
print(e)
```

## None as an Object

`None` is not a special form of "nothing" outside Python's object model.

`None` is an object created from the `NoneType` type.

As with the other objects we have studied, we can inspect its type and identity.

```python
value = None

print(type(value))
print(id(value))
```

## There Is Only One None Object

Python uses a single `None` object.

Different variables assigned to `None` therefore refer to the same object.

```python
a = None
b = None
c = None

print(id(a))
print(id(b))
print(id(c))

print(a is b)
print(a is c)
```

## None Is Immutable

`None` is immutable.

The `None` object itself cannot be modified.

A variable can, however, be reassigned to another object.

```python
value = None

print(value)
print(id(value))

value = "Python"

print(value)
print(id(value))
```

The original `None` object was not modified.

The variable `value` was reassigned and now refers to a string object.

This is the same distinction we saw with other immutable types:

- **Mutation** changes an existing object.
- **Reassignment** makes a variable refer to another object.

For `None`, mutation is not possible.

## None Is Different from Other Empty Values

Several values can represent something that is empty or absent in a particular context.

However, they are different Python objects with different types.

`None` specifically represents the absence of a value.

```python
none_value = None
zero = 0
empty_string = ""
empty_list = []
false_value = False

print(type(none_value))
print(type(zero))
print(type(empty_string))
print(type(empty_list))
print(type(false_value))
```

## None Can Be Assigned Like Any Other Value

A variable does not need to receive its final value immediately.

It can start as `None` and later be reassigned.

```python
result = None

print(result)

result = "completed"

print(result)
```

# Summary

- `None` represents the absence of a value.
- `None` is an object.
- Its type is `NoneType`.
- Python uses a single `None` object.
- Multiple variables assigned to `None` refer to the same object.
- `None` is immutable.
- Reassigning a variable that contains `None` does not modify `None`.
- `None` is different from `0`, `""`, `[]`, and `False`.
- A variable can temporarily contain `None` and later be reassigned to another value.
