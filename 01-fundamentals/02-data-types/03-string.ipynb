# Strings

## Introduction

Programs need to represent more than numbers.

Names, messages, file names, commands, and many other kinds of information are
made of text. For example, a program may need to store a person's name and
display a message to the user.

Python needs a specific way to represent this kind of data.

## Creating Strings

Text is represented in Python using characters enclosed by quotation marks.

Both single and double quotation marks can be used.

```python
name = "Alice"
message = 'Hello, Alice!'

print(name)
print(message)
```

## Strings as Objects

The values above are not just pieces of text.

In Python, a string is an **object** created from the `str` type.

As with other objects, we can inspect its type and identity.

```python
text = "Python"

print(type(text))
print(id(text))
```

## Strings Are Immutable

Now suppose we have the following string:

```python
text = "Python"

text[0] = "J"
```

The operation raises an error.

A `str` object is **immutable**, which means that its contents cannot be changed
after the object has been created.

This does not mean that a variable cannot receive another string. It means that
the existing string object itself cannot be modified.

## Reassignment

If we cannot modify an existing string, how can we make a variable represent
different text?

We can assign a new string to the variable.

```python
text = "Python"

print(text)

text = "Java"

print(text)
```

In this example, `"Python"` was not changed.

The variable `text` was **reassigned** and now refers to another string object.

This is the distinction between:

- **mutation** — changing an existing object;
- **reassignment** — making a variable refer to another object.

For strings, mutation is not allowed because `str` objects are immutable.

### Observing Reassignment

We can use `id()` to observe that reassignment can make the variable refer to a
different object.

```python
text = "Python"

print(text)
print(id(text))

text = "Java"

print(text)
print(id(text))
```

## Summary

- `str` is the Python type used to represent text.
- Strings are objects.
- `str` objects are immutable.
- An existing string cannot be modified.
- Reassignment does not modify the existing string.
- Reassignment makes a variable refer to another object.
