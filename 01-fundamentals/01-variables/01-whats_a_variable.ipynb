# Introduction

When a program works with information, it often needs to use the same value more than once.

For example, a program may receive a person's name and later need to display it, compare it, or use it as part of another operation. Writing the value itself every time would make the program harder to understand and maintain.

This raises a simple question:

How can a program keep track of a value and refer to it later?

In this notebook, we will explore how Python uses variables to give values names, allowing programs to store information and refer to it throughout their execution.

Suppose a program needs to remember a person's name.

````python
'Breno'
````

# Giving a value a name

````python
name = 'Breno'
````

# Using Variables

````python
print(name)
````

# Reassigning a variable

````python
name = 'Other name'

print(name)
````

# Assigning one variable to another

A variable can be used when assigning a value to another variable.


````python
a = 10
b = a
````

We can see this by checking their identities:

````python
id(a)
````

````python
id(b)
````

However, assigning a new value to one variable does not change the other:

````python
a = 20
````

````python
a
````

````python
b
````

# Naming variables

Variable names should communicate what the value represents.

## Basic naming rules

- Names can contain letters, numbers, and `_`.
- A name cannot start with a number.
- Spaces are not allowed in variable names.
- Python is case-sensitive, so `age` and `Age` are different names.
- Reserved keywords, such as `if`, `for`, and `class`, cannot be used as variable names.

Python also follows a common naming convention called **snake_case** for variables and functions.

In `snake_case`, words are written in lowercase and separated by underscores:

```python
items_per_box = 12
total_items = 48
number_of_students = 30

## Good Examples

````python
name = "Breno"
age = 12
items_per_box = 2
total_items = 48
````

## Poor Examples

````python
x = "Breno"
a = 12
v = 2
ti = 48
````

# Summary

- Meaningful variable names make programs easier to understand and maintain.
