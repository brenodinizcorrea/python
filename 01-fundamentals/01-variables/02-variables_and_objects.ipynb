# The Problem

When we write:

````python
age = 45
````

But what exactly is age?

Is age the value 30, or is it a name that refers to something else?

In Python, a variable is a name that refers to an object.

Conceptually:

age ──→ 30

The value 30 is an object, and age is the variable that refers to it.

The same idea applies to other values:

````python
name = "Breno"
items = [1, 2, 3]
````

Conceptually:

name  ──→ "Breno"

items ──→ [1, 2, 3]

# Mutable and Immutable Objects

Two variables can refer to the same object in Python.

But what happens when that object is changed?

Does changing one variable also affect the other?

We can investigate this by comparing a mutable object, such as a `list`, with an immutable object, such as an `int`.

## Two Variables, One Integer

Let's start with two variables containing the same integer:

````python
a = 10
b = a

print(f'ID object a: {id(a)}')
print(f'ID object b: {id(b)}')
print(f'Value a: {a}')
print(f'Value b: {b}')
````

Now change b:

````python
b = b + 1

print(f'ID object a: {id(a)}')
print(f'ID object b: {id(b)}')
print(f'Value a: {a}')
print(f'Value b: {b}')
````

Why?

Because an int is immutable.

The integer 10 was not changed into 11. Instead, b = b + 1 produced the value 11, and b was reassigned to it.

## Two Variables, One List

Now let's repeat the experiment with a list:

````python
a = [1, 2, 3]
b = a

print(f'ID object a: {id(a)}')
print(f'ID object b: {id(b)}')
print(f'Value a: {a}')
print(f'Value b: {b}')
````

Now modify b:

````python
b.append(4)

print(f'ID object a: {id(a)}')
print(f'ID object b: {id(b)}')
print(f'Value a: {a}')
print(f'Value b: {b}')
````

Why?

Because a list is mutable.

The existing list object was modified, rather than b being reassigned to another object.

# Same Value, Different Objects

Two variables can contain the same value without referring to the same object.

Let's create two lists with the same values.

````python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)
````

The result is `True` because the two lists contain the same values.

But are they the same object?

We can compare their identities using `id()`.

````python
print(id(a))
print(id(b))
````

The two `id` values are different.

So we have two different objects with the same value:

```text
a ──→ [1, 2, 3]

b ──→ [1, 2, 3]

# Summary

- `int` objects are immutable.
- `list` objects are mutable.
- `id()` can be used to observe whether two variables refer to the same object.
- Modifying a mutable object changes what every variable referring to that object observes.
- Changing an immutable value results in a new value and reassignment.
- Two variables can contain equal values while referring to different objects.
- `==` compares values, while `is` checks whether two variables refer to the same object.
