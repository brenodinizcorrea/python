# Collection Methods

## The Problem

A program often needs to perform operations on values that are already stored in objects.

For example, a list may need to receive a new element, remove an existing one, or provide information about its contents.

How can Python provide operations that are specific to each type of object?

## What Is a Method?

A method is a function associated with an object.

Methods are called using:

`object.method()`

```python
numbers = [10, 20, 30]

numbers.append(40)

print(numbers)
```

## Modify vs. Return

A method's behavior depends on how it was implemented.

A method can:

- modify the object;
- return a value;
- modify the object and return a value;
- neither modify the object nor return a useful value.

There is no general rule that applies to all methods.

```python
numbers = [3, 1, 2]

result = numbers.sort()

print(numbers)
print(result)
```

`sort()` modifies the list and returns `None`.

```python
numbers = [10, 20, 30]

result = numbers.pop()

print(numbers)
print(result)
```

`pop()` modifies the list and returns the removed element.

```python
numbers = [10, 20, 20, 30]

result = numbers.count(20)

print(numbers)
print(result)
```

`count()` does not modify the list and returns a value.

## Methods and Object Types

Methods available on an object depend on its type.

Different types provide different methods.

```python
numbers = [10, 20, 30]

print(numbers.count(20))
print(numbers.index(30))
```

```python
text = "Python"

print(text.upper())
print(text.count("o"))
```

A `list` and a `str` provide different methods because they are different types of objects.

## List Methods

Lists are mutable collections, so some list methods can change their contents.

Some useful examples are:

```python
numbers = [30, 10, 20, 10]

numbers.append(40)

print(numbers)

removed = numbers.pop()

print(numbers)
print(removed)

numbers.sort()

print(numbers)

print(numbers.count(10))
```

## Set Methods

Sets are mutable collections of unique elements.

Some useful methods are:

```python
numbers = {10, 20, 30}

numbers.add(40)

print(numbers)

numbers.remove(20)

print(numbers)

numbers.discard(50)

print(numbers)
```

## Dictionary Methods

Dictionaries store key-value pairs.

Some useful methods are:

```python
person = {
    "name": "Alice",
    "age": 30
}

print(person.get("name"))
print(person.get("email"))

print(person.keys())
print(person.values())
print(person.items())
```

# Summary

- A method is a function associated with an object.
- Methods are called using `object.method()`.
- Methods are defined by the object's type.
- A method may modify the object, return a value, do both, or neither.
- Whether a method modifies an object or returns a value depends on its implementation.
- Mutable objects can be modified by methods.
- Not every method that changes an object returns the modified object.
- Collection methods provide operations specific to each collection type.
- You do not need to memorize every available method.
- Documentation can be used to check a method's behavior, arguments, return value, and possible errors.
