# Indexing and Slicing

## The Problem

A sequence may contain many values, but a program often needs to work with only one element or a specific portion of it.

For example, you may need to access the first value, the last value, or a range of values without processing the entire sequence.

## Indexing

An index identifies the position of an element in a sequence.

Python starts indexing at `0`.

````python
numbers = [10, 20, 30, 40]

print(numbers[0])
print(numbers[1])
print(numbers[3])
````

The indexes are:

```text
10 → 0
20 → 1
30 → 2
40 → 3

## Negative Indexing

Negative indexes access elements from the end of a sequence.

`-1` refers to the last element.

````python
numbers = [10, 20, 30, 40]

print(numbers[-1])
print(numbers[-2])
print(numbers[-4])
````

-4 → 10

-3 → 20

-2 → 30

-1 → 40

## Indexing Strings

Strings are sequences, so their characters can also be accessed by index.

````python
text = "Python"

print(text[0])
print(text[2])
print(text[-1])
````

## Slicing

Slicing accesses a portion of a sequence.

The basic syntax is:

`sequence[start:stop]`

The `start` index is included.

The `stop` index is excluded.

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])
````

## Omitting `start` and `stop`

The `start` or `stop` value can be omitted.

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[:3])
print(numbers[2:])
print(numbers[:])
````

- `[:3]` starts from the beginning.
- `[2:]` continues until the end.
- `[:]` selects the entire sequence.

## Slice Step

A third value can specify the step:

`sequence[start:stop:step]`

````python
numbers = [10, 20, 30, 40, 50, 60]

print(numbers[::2])
print(numbers[1::2])
````

## Reversing a Sequence

A negative step can traverse a sequence backwards.

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[::-1])
````

## Negative Step and `stop`

When using a negative step, Python moves from right to left.

The `stop` index is still excluded, but the direction changes how the indexes are traversed.

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[0::-1])
print(numbers[::-1])
````

`numbers[0::-1]` starts at index `0` and tries to move backwards.

There are no elements before index `0`, so the result contains only the first element.

To reverse the entire sequence, omit both `start` and `stop`:

`numbers[::-1]`

## Different Sequence Types

Indexing and slicing apply to sequence types such as:

- `str`
- `list`
- `tuple`

They do not apply directly to unordered collections such as `set`.

````python
text = "Python"
numbers = [10, 20, 30]
coordinates = (10, 20, 30)
set_ = names = {1, 2, 3}

print(text[0])
print(numbers[1])
print(coordinates[-1])
````

## Types That Do Not Support Indexing or Slicing

Indexing and slicing are features of sequence types.

The main built-in collection types that do not support them are:

- `set`
- `dict`

A `set` is unordered, so there is no index position.

A `dict` is accessed by keys rather than indexes.

````python
numbers = {10, 20, 30}

print(numbers[0])
````

````python
person = {
    "name": "Alice",
    "age": 30
}

print(person[0])
````

## Index Errors

An index must exist in the sequence.

Trying to access an index outside the valid range raises `IndexError`.

````python
numbers = [10, 20, 30]

print(numbers[5])
````

# Summary

- Python sequences are zero-indexed.
- The first element has index `0`.
- Negative indexes access elements from the end.
- `-1` refers to the last element.
- Indexing accesses a single element.
- Slicing accesses a portion of a sequence.
- Slicing uses `start:stop`.
- The `start` position is included.
- The `stop` position is excluded.
- Slicing can use `start:stop:step`.
- A negative step traverses a sequence backwards.
- With a negative step, omitting `start` and `stop` can be used to reverse a sequence.
- Indexing and slicing work with `str`, `list`, and `tuple`.
- `set` does not support indexing or slicing.
- `dict` does not support indexing or slicing; dictionaries are accessed by keys.
- An invalid sequence index raises `IndexError`.
- Attempting to index a type that does not support indexing raises `TypeError`.
