# Nested Collections

## The Problem

A program may need to represent data that has more than one level of structure.

For example, a list may contain several records, and each record may contain multiple values.

How can Python represent and access collections with multiple levels of data?

## Nested Lists

A collection can contain other collections as its elements.

```python
numbers = [
    [10, 20, 30],
    [40, 50, 60]
]

print(numbers)
```

Each inner list is an element of the outer list.

```python
numbers = [
    [10, 20, 30],
    [40, 50, 60]
]

print(numbers[0])
print(numbers[1])
```

Indexing can continue through multiple levels.

```python
numbers = [
    [10, 20, 30],
    [40, 50, 60]
]

print(numbers[0][1])
print(numbers[1][2])
```

## Nested Dictionaries

Dictionaries can contain other dictionaries as values.

```python
users = {
    "alice": {
        "age": 30,
        "country": "Brazil"
    },
    "bob": {
        "age": 25,
        "country": "Brazil"
    }
}

print(users)
```

```python
print(users["alice"])
print(users["alice"]["age"])
```

## Collections Inside Collections

Collections can contain other collections.

This allows data to be organized into multiple levels.

```python
data = [
    {"name": "Alice", "scores": [80, 90]},
    {"name": "Bob", "scores": [70, 85]}
]

print(data)
```

## General Rule

A collection can generally contain objects of any type, including other collections.

Lists, dictionaries, and tuples can contain lists, dictionaries, tuples, and sets.

```python
list_data = [
    [1, 2],
    {"name": "Alice"},
    (3, 4),
    {5, 6}
]

print(list_data)
```

```python
dict_data = {
    "numbers": [1, 2, 3],
    "user": {"name": "Alice"},
    "coordinates": (10, 20),
    "tags": {"python", "data"}
}

print(dict_data)
```

```python
tuple_data = (
    [1, 2],
    {"name": "Alice"},
    (3, 4),
    {5, 6}
)

print(tuple_data)
```

## What Is Hashable?

**Hashable** is a property of an object.

Python can calculate a stable **hash value** for a hashable object.

`set` and `dict` use hash values to locate elements efficiently, allowing average `O(1)` lookup.

A hashable object's hash value must remain stable while it is used in a `set` or as a `dict` key.

## Set Exception

A `set` can only contain **hashable** objects.

`list`, `dict`, and `set` are not hashable, so they cannot be elements of a `set`.

A `tuple` can be an element of a `set` only if all of its elements are hashable.

This is a rule of Python's design because sets rely on hashable elements.

```python
valid = {
    (10, 20),
    (30, 40)
}

print(valid)
```

```python
invalid = {
    [10, 20]
}
```

The first example works because the tuple contains only hashable elements.

The second raises `TypeError` because a `list` is not hashable.

## Nested Collections and Loops

Nested collections can be traversed using nested loops.

```python
groups = [
    [10, 20, 30],
    [40, 50, 60]
]

for group in groups:
    for number in group:
        print(number)
```

## Nested Slicing

Indexing and slicing can be combined when working with nested sequences.

```python
numbers = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

print(numbers[:2])
print(numbers[1][1:])
```

## Structure Matters

The same values can represent different structures depending on how collections are nested.

```python
flat = [10, 20, 30, 40]

nested = [
    [10, 20],
    [30, 40]
]

print(flat[2])
print(nested[1])
print(nested[1][0])
```

## Collections Inside Collections

Collections can contain other collections, regardless of their own type.

The important exception is that a `set` can only contain elements that are hashable.

Mutable collections such as `list`, `dict`, and `set` cannot be elements of a `set`.

Immutable collections such as `tuple` can be elements of a `set`, as long as all elements inside the tuple are also hashable.

# Summary

- A collection can contain other collections.
- Nested collections create multiple levels of structure.
- Lists, dictionaries, tuples, and sets can generally contain other collections.
- Indexing can be applied at each level of nested sequences.
- Slicing can be combined with indexing in nested sequences.
- Dictionaries can contain other dictionaries or collections as values.
- Nested collections can be traversed with nested loops.
- A `set` can only contain hashable objects.
- Hashable objects have a stable hash value that can be used by `set` and `dict` for efficient lookup.
- A `tuple` can be a `set` element only if all its elements are hashable.
- The structure of nested collections determines how their elements are accessed.
