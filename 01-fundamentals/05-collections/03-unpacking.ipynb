# Unpacking

## The Problem

A collection can contain several values that need to be assigned to separate variables.

Writing each assignment individually can be repetitive, especially when the structure of the collection already matches the variables we need.

## Unpacking

Unpacking assigns the elements of an iterable to multiple variables.

```python
numbers = [10, 20, 30]

a, b, c = numbers

print(a)
print(b)
print(c)
```

The number of variables must normally match the number of elements.

## Unpacking Strings

Strings are iterable, so their characters can also be unpacked.

```python
first, second, third = "Cat"

print(first)
print(second)
print(third)
```

## Unpacking Tuples

Tuples can be unpacked in the same way as lists.

```python
person = ("Alice", 30)

name, age = person

print(name)
print(age)
```

## Unpacking Dictionaries

Dictionaries can also be unpacked.

When a dictionary is unpacked directly, its keys are assigned to the variables.

To unpack values or key-value pairs, use `values()` or `items()`.

```python
person = {
    "name": "Alice",
    "age": 30
}

first, second = person

print(first)
print(second)
```

```python
person = {
    "name": "Alice",
    "age": 30
}

name, age = person.values()

print(name)
print(age)
```

```python
person = {
    "name": "Alice",
    "age": 30
}

first, second = person.items()

print(first)
print(second)
```

Direct dictionary unpacking uses the keys.

`values()` provides the values.

`items()` provides key-value pairs as tuples.

## Mismatched Number of Values

The number of variables and values must match unless extended unpacking is used.

```python
numbers = [10, 20, 30]

a, b = numbers
```

This raises `ValueError` because there are three values but only two variables.

## Extended Unpacking

The `*` operator can collect multiple remaining values into a list.

```python
numbers = [10, 20, 30, 40, 50]

first, *middle, last = numbers

print(first)
print(middle)
print(last)
```

Here:

- `first` receives the first element.
- `last` receives the last element.
- `middle` receives all remaining elements as a list.

## Ignoring Values

The underscore `_` is commonly used for a value that will not be used.

```python
person = ("Alice", 30, "Brazil")

name, _, country = person

print(name)
print(country)
```

## Unpacking in Loops

Unpacking can also happen during iteration when each element contains multiple values.

```python
people = [
    ("Alice", 30),
    ("Bob", 25),
    ("Charlie", 35)
]

for name, age in people:
    print(name, age)
```

# Summary

- Unpacking assigns elements of an iterable to multiple variables.
- Lists, tuples, strings, and dictionaries can be unpacked.
- Direct dictionary unpacking assigns its keys to the variables.
- `values()` can be used to unpack dictionary values.
- `items()` can be used to unpack dictionary key-value pairs.
- The number of variables normally must match the number of values.
- A mismatch raises `ValueError`.
- `*` can collect multiple remaining values into a list.
- `_` is commonly used for a value that will not be used.
- Unpacking can also be used directly in `for` loops.
