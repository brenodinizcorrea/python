# Exercise 1 — Matrix

Create a list called `matrix` containing three lists, each with three numbers.

Access and print:

- the first element of the first list;
- the last element of the second list;
- the middle element of the third list.

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(matrix[0][0])
print(matrix[1][-1])
print(matrix[-1][1])
```

# Exercise 3 — Orders

Create a list called `orders` containing three dictionaries.

Each dictionary must contain:

- `"product"`
- `"quantity"`
- `"price"`

Access and print the product and quantity of the second order.

```python
orders = [
    {
        'product': 'product_one',
        'quantity': 1,
        'price': 1.1
    },
    {
        'product': 'product_two',
        'quantity': 2,
        'price': 2.2
    },
    {
        'product': 'product_three',
        'quantity': 3,
        'price': 3.3
    }
]

print(orders)
print(orders[1]['product'])
print(orders[1]['quantity'])
```

# Exercise 4 — Temperatures

Create a list called `temperatures` containing three lists, each representing the temperatures recorded during one day.

Use nested loops to print every temperature.

```python
temperatures = [
    [29, 32, 27, 20],
    [27, 30, 25, 21],
    [30, 33, 31, 29],
]

for temperatures_day in temperatures:
    for single_temperature in temperatures_day:
        print(single_temperature)
```
