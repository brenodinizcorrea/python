# Exercise 1 — Adding Elements

Create a list called `tasks` containing:

"study", "work"

Add "exercise" to the end of the list using a list method.

Print the list.

```python
tasks = ['study', 'work']

print(tasks)

tasks.append('exercise')

print(tasks)
```

# Exercise 2 — Remove and Return

Create a list called `queue` containing:

"Alice", "Bob", "Charlie"

Remove the first element using a list method.

Store the removed element in a variable called `first`.

Print the list and `first`.

```python
queue = ['Alice', 'Bob', 'Charlie']

print(queue)

first = queue.pop(0)

print(queue)
print(first)
```

# Exercise 3 — Counting

Create a list called `scores` containing:

10, 20, 10, 30, 10, 40

Use a list method to determine how many times `10` appears.

Store the result in a variable called `occurrences`.

Print `scores` and `occurrences`.

Check whether the original list was modified.

```python
scores = [10, 20, 10, 30, 10, 40]

print(scores)

occurrences = scores.count(10)

print(scores)
print(occurrences)
```

# Exercise 4 — Unique Values

Create a set called `tags` containing:

"python", "data"

Add "sql" to the set.

Then try to add "python" again.

Print the set after each operation.

```python
tags = {'python', 'data'}
print(tags)

tags.add('sql')
print(tags)

tags.add('python')
print(tags)
```

# Exercise 5 — Safe Access

Create a dictionary called `product` with:

"name": "Laptop"
"price": 3000

Use a dictionary method to retrieve the value associated with `"stock"`.

Store the result in a variable called `stock`.

Print `stock`.

Do not add `"stock"` to the dictionary.

```python
product = {
    'name': 'Laptop',
    'price': 3000
}

stock = product.get('stock')

print(stock)
```

# Exercise 6 — Method Behavior

Create a list called `numbers` containing:

40, 10, 30, 20

Use a list method to put the numbers in ascending order.

Store whatever the method returns in a variable called `result`.

Print:

- `numbers`
- `result`
- `type(result)`

Determine whether the method modified the list, returned a value, or did both.

```python
numbers = [40, 10, 30, 20]

print(numbers)

result = numbers.sort()

print(numbers)
print(result)
print(type(result))
```
