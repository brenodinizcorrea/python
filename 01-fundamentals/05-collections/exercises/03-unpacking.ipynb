# Exercise 1 — Coordinates

Create a list called `coordinates` containing:

- 15
- 25
- 40

Unpack the three values into `x`, `y`, and `z`.

Print the three variables.

```python
coordinates = [15, 25, 40]

x, y, z = coordinates

print(x)
print(y)
print(z)
```

# Exercise 3 — Dictionary Keys

Create a dictionary called `settings` containing:

- `"language": "Python"`
- `"version": 3`

Unpack the dictionary directly into two variables.

Print the variables.

Then unpack its values into two different variables and print them.

```python
settings = {
    'language': 'Python',
    'version': 3
}

first, second = settings

print(first)
print(second)

language, version = settings.values()

print(language)
print(version)
```

# Exercise 4 — First, Middle, Last

Create a list called `scores` containing:

10, 20, 30, 40, 50, 60

Use extended unpacking to assign:

- the first value to `first`;
- all middle values to `middle`;
- the last value to `last`.

Print the three variables.

```python
scores = [10, 20, 30, 40, 50, 60]

first, *middle, last = scores

print(first)
print(middle)
print(last)
```

# Exercise 5 — Ignore Data

Create a tuple called `user` containing:

"Lucas", 28, "Brazil"

Unpack the tuple into three variables.

The middle value will not be used.

Use `_` for it.

Print only the user's name and country.

```python
user = ('Lucas', 28, 'Brazil')

name, _, country = user

print(name)
print(country)
```

# Exercise 6 — Product Data

Create a list called `products` containing three tuples.

Each tuple must contain:

- a product name;
- a price.

Use a `for` loop to unpack each tuple into `name` and `price`.

Print the product name and price for each iteration.

```python
products = [
    ('Notebook', 2000),
    ('Apple', 2.5),
    ('Pen', 15.4)
]

for name, price in products:
    print(name, price)
```
