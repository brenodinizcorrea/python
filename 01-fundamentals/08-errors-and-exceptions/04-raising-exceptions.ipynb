# Raising Exceptions

## The Problem

Imagine that a function receives an age. The program may be able to process any integer, but some values do not make sense for the situation.

Instead of allowing invalid data to continue through the program, we may want to stop the operation and clearly report what is wrong.

## `raise`

The `raise` statement allows you to deliberately trigger an exception.

This is useful when the program detects a condition that should not be allowed to continue.

```python
age = -5

if age < 0:
    raise ValueError("Age cannot be negative.")
```

The exception type describes the kind of problem, while the message provides additional context about what went wrong.

## Raising Different Exceptions

Different conditions may require different exception types.

Choose an exception that accurately represents the problem being reported.

```python
number = 0

if number == 0:
    raise ZeroDivisionError("Number cannot be zero.")
```

## `raise` with `try/except`

An exception raised with `raise` can be handled by a matching `except` block, just like an exception generated automatically by Python.

```python
try:
    raise ValueError("Invalid value.")
except ValueError:
    print("The value is invalid.")
```

## Validating Input

A common use of `raise` is validating data before allowing the program to continue.

The validation code checks a condition and raises an exception when the value is not acceptable.

```python
age = -5

if age < 0:
    raise ValueError("Age cannot be negative.")

print("Age is valid.")
```

If the validation fails, the `raise` statement stops the current execution flow unless the exception is handled.

```python
age = -5

try:
    if age < 0:
        raise ValueError("Age cannot be negative.")
except ValueError as error:
    print(error)
```

```python
example = ValueError("Age cannot be negative.")

print(example)
```

## Raising an Exception from a Function

A function can use `raise` to reject invalid arguments.

This allows the function itself to define which values it accepts.

```python
def calculate_discount(price):
    if price < 0:
        raise ValueError("Price cannot be negative.")
    
    return price * 0.9

print(calculate_discount(100))
```

```python
def calculate_discount(price):
    if price < 0:
        raise ValueError("Price cannot be negative.")
    
    return price * 0.9

print(calculate_discount(-100))
```

# Summary

- `raise` deliberately triggers an exception.
- The exception type communicates the kind of problem.
- An exception message provides additional context.
- `raise` is commonly used to validate values and enforce rules.
- Raised exceptions can be handled with `try/except`.
- Python can raise exceptions automatically when operations fail.
