# Exercise 1 — Handle an Exception

Create a `try/except` block that converts `value` to an integer using `int()`. Catch `ValueError` and print `"Invalid number."`.

Given: `value = "twenty"`

Expected output: `Invalid number.`

````python
try:
    value = int('twenty')
    print(value)
except ValueError:
    print('Invalid number.')
````

# Exercise 3 — Handle Multiple Errors

Create a `try/except` structure that converts `value` to an integer and divides `100` by it. Handle `ValueError` by printing `"Invalid number."` and `ZeroDivisionError` by printing `"Cannot divide by zero."`.

Given: `value = "0"`

Expected output: `Cannot divide by zero.`

````python
try:
    value = '0'
    value = int(value)

    print(100 / value)

except ValueError:
    print('Invalid number.')

except ZeroDivisionError:
    print('Cannot divide by zero.')
````

# Exercise 4 — Continue After Handling an Error

Write a `try/except` block that attempts to convert `age` to an integer. If a `ValueError` occurs, print `"Invalid age."`. After the exception is handled, print `"Program continues."`.

Given: `age = "twenty"`

Expected output:
`Invalid age.`
`Program continues.`

````python
try:
    age = 'twenty'
    age = int(age)

except ValueError:
    print('Invalid age.')

print('Program continues.')
````

# Exercise 5 — Fix the Error

Reproduce the broken code below. It raises an exception because the code tries to access an index that does not exist. Fix it by handling the correct exception with `try/except` and print `"Index not found."`.

Given:

```python
numbers = [10, 20, 30]
print(numbers[5])
```

````python
try:
    numbers = [10, 20, 30]
    print(numbers[5])

except IndexError:
    print('Index not found.')
````
