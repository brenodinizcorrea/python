# Exercise 1 — Use `else` After a Successful Conversion

Create a `try/except/else` structure that converts `value` to an integer using `int()`.

Print `"Invalid number."` if a `ValueError` occurs. Print `"Valid number."` only if the conversion succeeds.

Given: `value = "25"`

Expected output: `Valid number.`

`````python
value = '25'

try:
    value = int(value)
except ValueError:
    print('Invalid number.')
else:
    print('Valid number.')
`````
