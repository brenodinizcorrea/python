# Exercise 1 — Validate a Temperature

Create a variable named `temperature` with the value `-20`.

Use `raise` to trigger a `ValueError` with the message `"Temperature is below the allowed limit."` when `temperature` is below `-10`.

Expected output:
`ValueError: Temperature is below the allowed limit.`

`````python
temperature = -20

if temperature < -10:
    raise ValueError('Temperature is below the allowed limit.')
`````

Expected output:
`ValueError: Quantity must be greater than zero.`

# Exercise 3 — Capture the Exception Message

Create a variable named `score` with the value `-3`.

Raise a `ValueError` with the message `"Score cannot be negative."` when `score` is less than zero.

Use `try/except` to catch the exception as `error` and print the message stored in `error`.

Expected output: `Score cannot be negative.`

`````python
score = -3

try:
    if score < 0:
        raise ValueError('Score cannot be negative.')
except ValueError as error:
    print(error)
`````

# Exercise 4 — Validate a Function Argument

Define a function named `withdraw` that receives `balance` and `amount`.

Raise a `ValueError` with the message `"Insufficient balance."` when `amount` is greater than `balance`.

Return the remaining balance when the withdrawal is valid.

Given: `balance = 500` and `amount = 150`

Expected output: `350`

`````python
def withdraw(balance, amount):

    if amount > balance:
        raise ValueError('Insufficient balance.')

    return balance - amount

print(withdraw(500, 150))
`````
