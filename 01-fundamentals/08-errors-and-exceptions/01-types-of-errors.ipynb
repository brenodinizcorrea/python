# Types of Errors

## The Problem

Imagine that you are running a Python program and, instead of producing the expected result, the execution suddenly stops.

Sometimes the problem is caused by a variable that does not exist. Sometimes an operation is incompatible with the data, or a value is invalid.

The challenge is that these situations can look similar when we first encounter them.

How can we understand what went wrong?

## An Unexpected Result

Consider the following code:

```python
age = 30
name = "Ana"

print(age + name)
```

The program does not produce a result.

Instead, Python reports a `TypeError`.

The error itself gives us information about the problem.

## NameError

What happens when Python cannot find the name we are trying to use?

```python
print(name_one)
```

This produces a `NameError`.

The name `name_one` has not been defined.

Now define it before using it:

```python
name_one = "Ana"

print(name_one)
```

## TypeError

A `TypeError` occurs when an operation is incompatible with the types of the objects involved.

```python
age = 30
name = "Ana"

print(age + name)
```

Here, `age` is an `int` and `name` is a `str`.

The `+` operation cannot combine these two types in this way.

Using compatible types makes the operation valid:

```python
age = 30
years = 5

print(age + years)
```

## ValueError

What happens when the type is appropriate, but the value itself cannot be used as expected?

```python
number = int("10")

print(number)
```

```python
number = int("abc")

print(number)
```

Both values are strings, so the type is acceptable for `int()`.

The difference is their content.

`"10"` represents an integer, while `"abc"` does not.

This is the key distinction:

- `TypeError` → incompatible type.
- `ValueError` → invalid value.

## IndexError

Now consider a list.

What happens when we request an element that is outside the list?

```python
numbers = [10, 20, 30]

print(numbers[3])
```

This produces an `IndexError`.

The list has indexes `0`, `1`, and `2`, so index `3` does not exist.

## KeyError

A similar problem can happen with dictionaries.

What happens when we request a key that does not exist?

```python
person = {
    "name": "Ana",
    "age": 30
}

print(person["city"])
```

This produces a `KeyError`.

The key `"city"` does not exist in the dictionary.

**# Summary**

- `NameError` → a name is not defined.
- `TypeError` → an operation is incompatible with the types involved.
- `ValueError` → a value is invalid for the operation.
- `IndexError` → a list index does not exist.
- `KeyError` → a dictionary key does not exist.
- Error messages provide information about what went wrong.
