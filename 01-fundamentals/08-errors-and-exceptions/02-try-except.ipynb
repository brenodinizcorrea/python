# Try and Except

## The Problem

Imagine that your program asks users to enter their age. Most users enter a number, but someone may type `"twenty"` instead.

The program should be able to deal with this situation without stopping completely.

```python
age = int("twenty")

print(f"You are {age} years old.")
```

## Handling an Exception

The previous code stops when the conversion fails.

We need a way to attempt an operation and define what should happen if that operation fails.

## `try`

The `try` block contains code that may raise an exception.

Python attempts to execute this code normally.

```python
try:
    age = int("25")
    print(f"You are {age} years old.")
except:
    print("Invalid age.")
```

The conversion succeeds, so the code inside `except` is not executed.

## `except`

The `except` block runs when an exception occurs inside the corresponding `try` block.

This allows the program to respond to the problem instead of stopping immediately.

```python
try:
    age = int("twenty")
    print(f"You are {age} years old.")
except:
    print("Invalid age.")
```

The conversion fails, so Python skips the remaining code inside `try` and executes `except`.

## Handling a Specific Exception

An `except` block can specify which exception it is designed to handle.

This makes the program's response more precise.

```python
try:
    age = int("twenty")
except ValueError:
    print("Please enter a valid number.")
```

## Multiple Exceptions

Different operations can produce different exceptions.

Multiple `except` blocks allow the program to respond differently to each known problem.

```python
try:
    number = int("0")
    result = 100 / number
except ValueError:
    print("The value is not a valid integer.")
except ZeroDivisionError:
    print("The number cannot be zero.")
```

## `try` vs. `except`

The two blocks have different responsibilities.

- `try` contains the operation that may fail.
- `except` contains the response to a matching exception.

```python
try:
    number = int("10")
    result = 100 / number
except ValueError:
    print("Invalid number.")
else:
    print(result)
```

## When No Exception Occurs

Sometimes the operation inside `try` succeeds.

In that case, Python skips the `except` block and continues with the rest of the program.

```python
try:
    number = int("10")
except ValueError:
    print("Invalid number.")

print("Program continues.")
```

## When the Exception Is Not Handled

A `try` block does not automatically handle every possible exception.

If the exception does not match any `except` block, Python still reports the exception.

```python
try:
    numbers = [10, 20, 30]
    print(numbers[5])
except ValueError:
    print("Invalid value.")
```

The exception is an `IndexError`, so the `ValueError` handler does not match it.

# Summary

- A `try` block contains code that may raise an exception.
- An `except` block handles a matching exception.
- Code after a failed operation inside `try` is not executed.
- Specific exceptions can be handled with specific `except` blocks.
- Multiple `except` blocks can handle different exception types.
- An unmatched exception is not handled by the `except` blocks.
