# Else and Finally

## The Problem

Imagine that your program tries to process a value that may or may not be valid. When the operation succeeds, you may want to perform one action; when it fails, you may want to handle the error differently.

In either case, some final action may need to happen after the process finishes.

## `else`

The `else` block runs only when no exception occurs inside the `try` block.

It is useful when an action should happen only after the operation succeeds.

```python
try:
    value = int("25")
except ValueError:
    print("Invalid number.")
else:
    print("Valid number.")
```

## `else` with an Exception

When an exception occurs, Python skips the `else` block and executes the matching `except` block.

```python
try:
    value = int("twenty")
except ValueError:
    print("Invalid number.")
else:
    print("Valid number.")
```

## `else` vs. Code After `try/except`

The `else` block runs only when no exception occurs.

Code placed after the entire `try/except` structure runs after the structure finishes, whether an exception was handled or not.

```python
try:
    value = int("25")
except ValueError:
    print("Invalid number.")
else:
    print("Conversion succeeded.")

print("Program continues.")
```

## `else` Is Optional

The `else` block runs only when no exception occurs inside the `try` block.

However, it is not required to execute code after a successful operation. The same behavior can often be achieved by placing that code directly inside the `try` block.

The main advantage of `else` is **separation**: it makes a clear distinction between the operation that may raise an exception and the code that should run only after that operation succeeds.

```python
try:
    value = int("25")
except ValueError:
    print("Invalid number.")
else:
    print("Valid number.")
```

```python
try:
    value = int("25")
    print("Valid number.")
except ValueError:
    print("Invalid number.")
```

## `finally`

The `finally` block always runs after the `try` process finishes.

It runs whether an exception occurs or not.

```python
try:
    value = int("25")
except ValueError:
    print("Invalid number.")
finally:
    print("Processing finished.")
```

## `finally` with an Exception

When an exception occurs, Python executes the matching `except` block and then executes `finally`.

```python
try:
    value = int("twenty")
except ValueError:
    print("Invalid number.")
finally:
    print("Processing finished.")
```

## `try`, `except`, `else`, and `finally`

The four blocks have different roles.

- `try` contains the operation that may fail.
- `except` handles a matching exception.
- `else` runs when no exception occurs.
- `finally` always runs.

```python
try:
    value = int("25")
except ValueError:
    print("Invalid number.")
else:
    print("Valid number.")
finally:
    print("Processing finished.")
```

## Execution Order

When no exception occurs, the order is `try` → `else` → `finally`.

```python
try:
    print("try")
except ValueError:
    print("except")
else:
    print("else")
finally:
    print("finally")
```

When an exception occurs and is handled, the order is `try` → `except` → `finally`.

```python
try:
    print("try")
    int("twenty")
except ValueError:
    print("except")
else:
    print("else")
finally:
    print("finally")
```

# Summary

- `else` runs only when the `try` block completes without an exception.
- `except` runs when a matching exception occurs.
- `finally` runs regardless of whether an exception occurs.
- Code after `try/except` also runs after a handled exception.
- `try`, `except`, `else`, and `finally` each have a different role in controlling execution.
