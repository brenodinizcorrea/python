# Exercise 1 — Access Permission

Create a variable `age` containing 22 and a variable `has_permission` containing `False`.

If the person is at least 18 years old and has permission, print `Access granted`.

Otherwise, print `Access denied`.

```python
age = 22
has_permission = False

if age >= 18 and has_permission:
    print('Access granted')
else:
    print('Access denied')
```

# Exercise 2 — Temperature

Create a variable `temperature` containing 12.

Use `if`, `elif`, and `else` to print:

- `Cold` when the temperature is below 15;
- `Mild` when the temperature is between 15 and 25;
- `Hot` when the temperature is above 25.

```python
temperature = 12

if temperature < 15:
    print('Cold')
elif temperature >= 15 and temperature <= 25:
    print('Mild')
else:
    print('Hot')
```

# Exercise 3 — Password Status

Create a variable `password_length` containing 7.

Use conditional statements to print:

- `Too short` when the length is less than 8;
- `Acceptable` when the length is between 8 and 11;
- `Strong` when the length is 12 or greater.

```python
password_length = 7

if password_length < 8:
    print('Too Short')
elif password_length >= 8 and password_length <= 11:
    print('Acceptable')
else:
    print('Strong')
```

# Exercise 5 — Number Classification

Create a variable `number` containing -5.

Use `if`, `elif`, and `else` to print:

- `Positive` when the number is greater than 0;
- `Negative` when the number is less than 0;
- `Zero` when the number is equal to 0.

```python
number = -5

if number > 0:
    print('Positive')
elif number < 0:
    print('Negative')
else:
    print('Zero')
```
