# Exercise 1 — Countdown

Create a variable `number` containing 10.

Use a `while` loop to print the numbers from 10 down to 1.

After each iteration, decrease `number` by 1.

```python
number = 10

while number >= 1:
    print(number)
    number -= 1
```

# Exercise 2 — Accumulation

Create a variable `total` containing 0 and a variable `number` containing 1.

Use a `while` loop to add `number` to `total` while `number` is less than or equal to 5.

Increase `number` by 1 after each iteration.

Print `total` after the loop finishes.

```python
total = 0
number = 1

while number <= 5:
    total += number
    number += 1

print(total)
```

# Exercise 3 — Doubling

Create a variable `number` containing 2.

Use a `while` loop to repeatedly multiply `number` by 2 while its value is less than 100.

Print `number` during each iteration.

```python
number = 2

while number < 100:
    print(number)
    number *= 2
```

# Exercise 4 — Repeated Subtraction

Create a variable `number` containing 50 and a variable `count` containing 0.

Use a `while` loop to repeatedly subtract 7 from `number` while `number` is greater than or equal to 7.

Increase `count` by 1 after each subtraction.

Print `number` and `count` after the loop finishes.

```python
number = 50
count = 0

while number >= 7:
    number -= 7
    count += 1
    
print(number)
print(count)
```

# Exercise 5 — Condition with `or`

Create two variables:

`number` containing 3

`limit` containing 20

Use a `while` loop to print `number` and multiply it by 2 after each iteration.

The loop should continue while `number` is less than `limit` or `number` is equal to `limit`.

Print the final value of `number` after the loop.

```python
number = 3
limit = 20

while number <= limit:
    print(number)
    number *= 2

print(number)
```

# Exercise 6 — Loop with a List

Create a list called `numbers` containing:

`4`, `8`, `12`, `16`, `20`

Create a variable `index` containing 0.

Use a `while` loop to print each item in the list.

Increase `index` by 1 after each iteration.

Do not use a `for` loop.

```python
numbers = [4, 8, 12, 16, 20]

index = 0

while index < len(numbers):
    print(numbers[index])
    index += 1
```
