# Exercise 1 — Stop Searching

Create a list called `numbers` containing:

3, 8, 12, 5, 19, 7

Use a `for` loop to go through the list.

When the number 19 is found, stop the loop.

Print each number before checking whether it is 19.

```python
numbers = [3, 8, 12, 5, 19, 7]

for item in numbers:
    print(item)
    if item == 19:
        break
```

# Exercise 2 — Skip Multiples

Create a list called `numbers` containing:

2, 5, 8, 11, 14, 17, 20

Use a `for` loop to print only the numbers that are not divisible by 2.

Use `continue` to skip the numbers that are divisible by 2.

```python
numbers = [2, 5, 8, 11, 14, 17, 20]

for item in numbers:
    if item % 2 == 0:
        continue

    print(item)
```

# Exercise 3 — Limit

Create a variable `number` containing 1.

Use a `while` loop to increase `number` by 1 on each iteration.

Stop the loop when `number` reaches 7.

Use `break`.

Print `number` during each iteration.

```python
number = 1

while True:
    print(number)
    if number == 7:
        break

    number += 1
```

# Exercise 4 — Inner Loop

Create two lists:

`groups` containing 1, 2, 3

`numbers` containing 1, 2, 3, 4

Use nested `for` loops.

Inside the inner loop, use `break` when `number` is equal to 3.

Print `group` and `number` before the `break` condition.

Observe which loop continues after the inner loop is terminated.

```python
groups = [1, 2, 3]
numbers = [1, 2, 3, 4]

for single_group in groups:
    for single_number in numbers:
        print(single_group, single_number)
        if single_number == 3:
            break
```

# Exercise 5 — Empty Block

Create a variable `age` containing 20.

Create an `if` statement that checks whether `age` is greater than or equal to 18.

Leave the `if` block empty using `pass`.

After the `if` statement, print `age`.

```python
age = 20

if age >= 18:
    pass

print(age)
```
