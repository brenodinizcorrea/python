# Exercise 1 — List Iteration

Create a list called `prices` containing:

`10.5`, `20.0`, `15.5`, `30.0`

Use a `for` loop to print each price.

```python
prices = [10.5, 20.0, 15.5, 30.0]

for item in prices:
    print(item)
```

# Exercise 2 — String Iteration

Create a variable called `word` containing `Python`.

Use a `for` loop to print each character.

Then use another `for` loop to print each character followed by the text `!`.

```python
word = 'Python'

for item in word:
    print(item)

for item in word:
    print(item + '!')
```

# Exercise 3 — Filtering Values

Create a list called `numbers` containing:

`4`, `11`, `7`, `20`, `3`, `15`

Use a `for` loop and an `if` statement to print only the numbers greater than 10.

```python
numbers = [4, 11, 7, 20, 3, 15]

for item in numbers:
    if item > 10:
        print(item)
```

# Exercise 4 — Range

Use `range()` with a `for` loop to print the even numbers from 2 through 10.

Do not create a list containing the numbers manually.

```python
for item in range(2, 11, 2):
    print(item)
```

# Exercise 5 — Dictionary

Create a dictionary called `product` with the following data:

- `name`: `Notebook`
- `price`: `2500`
- `category`: `Computer`

Use a `for` loop with `items()` to print each key and value.

The output should contain the key and its corresponding value on each iteration.

```python
product = {
    'name': 'Notebook',
    'price': 2500,
    'category': 'Computer'
}

for key, value in product.items():
    print(key, value)
```

# Exercise 6 — Nested Loops

Create two lists:

- `colors` containing `Red`, `Blue`;
- `sizes` containing `S`, `M`, `L`.

Use nested `for` loops to print every possible combination of color and size.

```python
colors = ['Red', 'Blue']
sizes = ['S', 'M', 'L']

for single_color in colors:
    for single_size in sizes:
        print(single_color, single_size)
```
