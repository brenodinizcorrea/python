# Break, Continue and Pass

## The Problem

Sometimes a loop should not process every iteration in the same way.

A program may need to stop the loop early, skip a particular iteration, or leave a block empty temporarily.

## `break`

`break` immediately terminates the loop.

Execution continues with the first statement after the loop.

```python
for number in range(1, 10):
    if number == 5:
        break

    print(number)

print("Done")
```

The loop stops when `number` reaches `5`.

The values `5` through `9` are not processed.

## `break` with `while`

`break` can also terminate a `while` loop.

```python
number = 1

while number <= 10:
    if number == 6:
        break

    print(number)
    number += 1
```

## `break` in Nested Loops

When loops are nested, `break` affects only the loop in which it appears.

It does not automatically terminate the outer loop.

```python
for outer in range(3):

    print("outer:", outer)

    for inner in range(5):

        if inner == 2:
            break

        print("inner:", inner)
```

Here, `break` belongs to the inner loop.

Therefore, it terminates the inner loop, while the outer loop continues with its next iteration.

```python
outer = 1

while outer <= 3:

    inner = 1

    while inner <= 5:

        if inner == 3:
            break

        print(outer, inner)
        inner += 1

    outer += 1
```

## `continue`

`continue` skips the rest of the current iteration.

The loop then moves to the next iteration.

```python
for number in range(1, 6):
    if number == 3:
        continue

    print(number)
```

The iteration where `number` is `3` is skipped.

The loop continues with `4`.

## `continue` in Nested Loops

When loops are nested, `continue` affects only the loop in which it appears.

This applies to both `for` and `while` loops.

```python
for outer in range(2):

    for inner in range(4):

        if inner == 2:
            continue

        print(outer, inner)
```

Here, `continue` affects the inner `for` loop.

The current iteration of the inner loop is skipped, but the outer loop continues normally.

```python
outer = 1

while outer <= 2:

    inner = 1

    while inner <= 4:

        if inner == 2:
            inner += 1
            continue

        print(outer, inner)
        inner += 1

    outer += 1
```

Here, `continue` affects the inner `while` loop.

The outer `while` continues normally.

## `pass`

`pass` does nothing.

It is useful when Python requires a statement but you do not want to execute anything yet.

```python
number = 10

if number > 5:
    pass

print(number)
```

## `pass` in a Loop

`pass` can also appear inside loops.

It does not skip an iteration or terminate the loop.

```python
for number in range(1, 4):
    pass

print("Done")
```

The loop still executes all of its iterations.

`pass` simply means:

**"Do nothing here."**

## `pass` in Nested Loops

When loops are nested, `pass` does not affect the execution of other loops.

It simply does nothing at the point where it appears.

This applies to both `for` and `while` loops.

```python
for outer in range(2):

    for inner in range(4):

        if inner == 2:
            pass

        print(outer, inner)
```

Here, `pass` does not skip the iteration or terminate the inner loop.

The `print()` statement still executes.

```python
outer = 1

while outer <= 2:

    inner = 1

    while inner <= 4:

        if inner == 2:
            pass

        print(outer, inner)
        inner += 1

    outer += 1
```

The same behavior occurs with nested `while` loops.

`pass` simply means that Python should do nothing at that point and continue normally.

# Summary

- `break` immediately terminates the loop in which it appears.
- In nested loops, `break` affects only the loop in which it appears.
- `break` does not automatically terminate outer loops.
- `continue` skips the current iteration of the loop in which it appears.
- In nested loops, `continue` affects only the loop in which it appears.
- `continue` does not automatically affect outer loops.
- `pass` does nothing at the point where it appears.
- In nested loops, `pass` does not affect the execution of other loops.
- `break` changes a loop's termination.
- `continue` changes what happens during the current iteration.
- `pass` does not change the loop's execution.
- These behaviors apply to both `for` and `while` loops.
