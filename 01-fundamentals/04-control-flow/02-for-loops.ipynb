# For Loops

## The Problem

Programs often need to repeat an operation for each value in a collection.

## The Iteration Variable

The variable after `for` receives the current item at each iteration.

Its name is chosen by the programmer.

```python
numbers = [10, 20, 30]

for number in numbers:
    print(number)
```

In this example, `number` refers to:

1. `10`
2. `20`
3. `30`

The loop finishes after all items have been processed.

## Iterating Over Different Iterables

A `for` loop can iterate over different iterable objects.

For example, strings:

```python
text = "Python"

for character in text:
    print(character)
```

A tuple:

```python
coordinates = (10, 20, 30)

for coordinate in coordinates:
    print(coordinate)
```

And a set:

```python
numbers = {10, 20, 30}

for number in numbers:
    print(number)
```

## Iterating Over a Dictionary

When iterating directly over a dictionary, the loop iterates over its keys.

```python
person = {
    "name": "Alice",
    "age": 30
}

for key in person:
    print(key)
```

To access both keys and values, we can use `items()`.

```python
person = {
    "name": "Alice",
    "age": 30
}

for key, value in person.items():
    print(key, value)
```

## `range()`

`range()` produces a sequence of numbers that can be used with a `for` loop.

The first value is included and the stop value is excluded.

The three arguments represent:

`start`, `stop`, and `step`.

```python
for number in range(5):
    print(number)
```

```python
for number in range(2, 6):
    print(number)
```

```python
for number in range(0, 10, 2):
    print(number)
```

## `for` and Conditions

A `for` loop can contain conditional statements.

```python
numbers = [1, 2, 3, 4, 5]

for number in numbers:
    if number > 3:
        print(number)
```

## Nested `for` Loops

A `for` loop can contain another `for` loop.

The inner loop runs completely for each iteration of the outer loop.

```python
rows = [1, 2]
columns = [1, 2, 3]

for row in rows:
    for column in columns:
        print(row, column)
```

# Summary

- A `for` loop repeats a block of code for each item in an iterable.
- The iteration variable receives the current item.
- The iteration variable can have any valid variable name.
- `for` can iterate over strings, lists, tuples, sets, dictionaries, and other iterables.
- Iterating directly over a dictionary produces its keys.
- `items()` can be used to iterate over dictionary keys and values.
- `range()` produces a sequence of numbers commonly used with `for`.
- `range()` includes the start value and excludes the stop value.
- A `for` loop can contain other statements, including `if`.
- `for` loops can be nested.
