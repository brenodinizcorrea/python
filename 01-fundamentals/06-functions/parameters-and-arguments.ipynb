# Parameters and Arguments

## The Problem

Consider a function that calculates the total price of an order.

If the values are defined directly inside the function, the same code cannot be reused for orders with different prices or quantities without changing the function itself.

Creating a separate function for each order would lead to duplicated code and make the program harder to maintain.

The function needs a way to receive different values each time it is called.

## Parameters

A parameter is a variable defined in a function's definition.

It represents a value that the function expects to receive.

```python
def greet(name):
    print("Hello", name)
```

## Arguments

An argument is the value provided to a function when it is called.

```python
greet("Alice")
greet("Bob")
```

Here:

- `name` is the parameter.
- `"Alice"` and `"Bob"` are arguments.

## Multiple Parameters

A function can have multiple parameters.

Arguments are assigned to parameters according to their position.

```python
def add(a, b):
    return a + b

result = add(10, 20)

print(result)
```

## Positional Arguments

When arguments are passed without parameter names, their position determines which parameter receives each value.

```python
def describe_product(name, price):
    print(name)
    print(price)

describe_product("Laptop", 2500)
```

## Keyword Arguments

Arguments can also be passed using parameter names.

With keyword arguments, the order does not need to match the parameter order.

```python
describe_product(
    price=2500,
    name="Laptop"
)
```

## Mixing Positional and Keyword Arguments

Positional and keyword arguments can be used in the same function call.

However, positional arguments must come before keyword arguments.

```python
def describe_product(name, price, category):
    print(name, price, category)
```

```python
describe_product("Laptop", price=2500, category="Computer")
```

```python
describe_product(name="Laptop", 2500, category="Computer")
```

## Missing and Extra Arguments

The number of provided arguments must match the required parameters.

Providing too few or too many arguments raises `TypeError`.

```python
def multiply(a, b):
    return a * b

multiply(10)
```

```python
multiply(10, 20, 30)
```

## Type Hints

Type hints (also called type annotations) indicate the expected types of parameters, variables, and return values.

They improve readability, documentation, and support from development tools.

```python
def calculate_total(price: float, quantity: int) -> float:
    return price * quantity

total = calculate_total(25.5, 3)

print(total)
```

Type hints do not enforce the types at runtime.

Python still allows the function to be called with other types.

The annotations describe the expected types; they do not perform type validation.

```python
def add(a: int, b: int) -> int:
    return a + b

print(add("Hello", " World"))
```

# Summary

- Parameters are variables defined in a function's definition.
- Arguments are values provided when calling a function.
- Parameters allow functions to receive values.
- A function can have multiple parameters.
- Positional arguments are assigned according to their position.
- Keyword arguments are assigned using parameter names.
- Keyword arguments do not need to follow parameter order.
- When mixing positional and keyword arguments, positional arguments must come first.
- The number of arguments must match the required parameters.
- Too few or too many arguments raise `TypeError`.
