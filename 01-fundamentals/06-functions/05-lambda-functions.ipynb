# Lambda Functions

## The Problem

Imagine that you need a small function for a single operation, such as multiplying a value by two.

Creating a complete named function works, but sometimes the operation is simple and will only be used in one specific place.

In these situations, defining a function with `def`, assigning it a name, and writing a separate block of code can add unnecessary structure.

## Lambda Expression

A `lambda` creates a small anonymous function.

It can receive parameters and return the result of a single expression.

```python
double = lambda x: x * 2

print(double(5))
```

This is equivalent to:

```python
def double(x):
    return x * 2

print(double(5))
```

## Multiple Parameters

A lambda can receive multiple parameters.

```python
add = lambda a, b: a + b

print(add(10, 20))
```

## Lambda and `return`

A lambda does not use the `return` keyword.

The expression after `:` is automatically returned.

```python
square = lambda x: x ** 2

print(square(4))
```

## Lambda as an Argument

A lambda can be passed directly to another function.

This is useful when a function needs a small operation temporarily.

```python
numbers = [1, 2, 3, 4]

result = list(map(lambda x: x * 2, numbers))

print(result)
```

```python
def double(x):

    return x * 2

numbers = [1, 2, 3, 4]

result = list(map(double, numbers))

print(result)
```

## Lambda Limitations

A lambda is restricted to a single expression.

It cannot contain a sequence of statements such as multiple independent operations.

```python
square = lambda x: x ** 2
```

For more complex logic, a regular function defined with `def` is generally clearer.

# Summary

- `lambda` creates a small anonymous function.
- A lambda can receive parameters.
- The expression after `:` is automatically returned.
- A lambda does not use `return`.
- A lambda can receive multiple parameters.
- A lambda can be assigned to a variable.
- A lambda can be passed as an argument to another function.
- A lambda is limited to a single expression.
- Regular functions are generally clearer for more complex logic.
