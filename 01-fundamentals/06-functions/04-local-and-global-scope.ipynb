# Local and Global Scope

## The Problem

Imagine that a program needs to keep track of a user's score.

The score is used in different parts of the program, while some temporary values are only needed while a specific function is running.

If every variable were accessible everywhere, different parts of the program could accidentally modify values that they should not control.

The program needs a way to determine where a variable can be accessed and how long that variable exists.

## Local Variables

A variable created inside a function is local to that function.

It can be accessed inside the function where it was created.

```python
def calculate_total():
    total = 100
    print(total)

calculate_total()
```

The variable `total` belongs to the function's local scope.

```python
def calculate_total():
    total = 100

calculate_total()

print(total)
```

The code raises `NameError` because `total` does not exist in the global scope.

## Global Variables

A variable created outside a function belongs to the global scope.

It can be accessed from inside a function.

```python
tax = 0.1

def calculate_tax(price):
    return price * tax

print(calculate_tax(100))
```

## Same Name in Different Scopes

A local variable can have the same name as a global variable.

The local variable takes precedence inside the function.

```python
value = 10

def show_value():
    value = 20
    print(value)

show_value()

print(value)
```

The function prints `20`, while the global variable remains `10`.

The two variables are different objects bound to different names in different scopes.

## Modifying a Global Variable

A function can read a global variable without declaring it as `global`.

However, assigning a value to that name inside the function creates a local variable instead.

```python
score = 10

def update_score():
    score = 20

update_score()

print(score)
```

The global `score` remains `10`.

The assignment inside the function created a local `score`.

## The `global` Keyword

The `global` keyword tells Python that an assignment should refer to a variable in the global scope.

```python
score = 10

def update_score():
    global score
    score = 20

update_score()

print(score)
```

The global `score` is now changed to `20`.

## Function Parameters Are Local

Function parameters are local to the function.

They can be used inside the function but are not automatically available outside it.

```python
def greet(name):
    print("Hello", name)

greet("Alice")

print(name)
```

The parameter `name` belongs to the function's local scope.

## Scope and Nested Functions

A function can be defined inside another function.

The inner function can access variables from its enclosing function.

```python
def outer():
    message = "Hello"

    def inner():
        print(message)

    inner()
    print(message)

outer()
```

The inner function can access `message` because it belongs to the enclosing scope.

More advanced rules for nested scopes will be explored later when needed.

# Summary

- Scope determines where a variable can be accessed.
- Variables created inside functions are local variables.
- Local variables are not directly accessible outside their function.
- Variables created outside functions belong to the global scope.
- Functions can read global variables.
- Assigning to a name inside a function normally creates a local variable.
- A local variable takes precedence over a global variable with the same name.
- The `global` keyword allows a function to assign to a global variable.
- Function parameters are local to the function.
- A nested function can access variables from its enclosing function.
