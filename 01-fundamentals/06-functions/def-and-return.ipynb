# Functions

## The Problem

Programs often need to perform the same operation in different parts of the code.

Repeating the same instructions makes code longer and harder to maintain.

How can we define a reusable block of code and execute it whenever needed?

## Defining a Function

A function is a reusable block of code that performs a specific task.

A function is defined using `def`.

```python
def greet():
    print("Hello!")
```

Defining a function does not execute its code.

The function runs when it is called.

```python
greet()
```

## Calling a Function

A function is called using its name followed by parentheses.

```python
def greet():
    print("Hello!")

greet()
greet()
```

The same function can be called multiple times.

## Return

A function can return a value using `return`.

The returned value can be stored in a variable or used in another expression.

```python
def add():
    return 10 + 20

result = add()

print(result)
```

## `print()` vs. `return`

`print()` displays a value.

`return` sends a value back to the code that called the function.

```python
def show_number():
    print(10)

def get_number():
    return 10

a = show_number()
b = get_number()

print(a)
print(b)
```

`show_number()` displays `10`, but its return value is `None`.

`get_number()` returns `10`, which can be assigned to `b`.

## Returning Multiple Values

A function can return multiple values.

Python returns them as a tuple.

```python
def get_coordinates():
    return 10, 20

result = get_coordinates()

print(result)
print(type(result))
```

The tuple can then be unpacked into variables:

```python
x, y = get_coordinates()

print(x)
print(y)
```

The number of variables must normally match the number of values in the tuple.

```python
x = get_coordinates()
```

This is valid because `x` receives the entire tuple.

```python
x, y, z = get_coordinates()
```

But raises `ValueError` because the tuple contains two values and three variables are expected.

## Early Return

`return` immediately terminates the function.

Code after `return` is not executed.

```python
def check_number():
    return 10

    print("This will not run")

result = check_number()

print(result)
```

## No Explicit Return

If a function does not explicitly return a value, its return value is `None`.

```python
def greet():
    print("Hello!")

result = greet()

print(result)
```

## Functions Are Objects

A function is also an object in Python.

Using the function name without `()` refers to the function object.

Using `()` calls the function.

```python
def greet():
    print("Hello")

print(greet)

greet()
```

`greet` refers to the function object.

`greet()` executes the function.

# Summary

- A function is a reusable block of code.
- Functions are defined using `def`.
- Defining a function does not execute it.
- A function is executed when it is called.
- A function can be called multiple times.
- `return` sends a value back to the calling code.
- `print()` displays a value but does not return it.
- `return` immediately terminates the function.
- A function without an explicit `return` returns `None`.
- A function can return multiple values, which Python packs into a tuple.
- A function is also an object in Python.
