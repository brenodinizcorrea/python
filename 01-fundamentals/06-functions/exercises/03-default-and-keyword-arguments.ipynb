# Exercise 1 — Shipping Fee

Create a function called `calculate_total` with two parameters:

- `price`
- `shipping`, with a default value of 15

The function must return the price plus the shipping fee.

Call the function once without providing `shipping` and once providing a different shipping fee.

```python
def calculate_total(price, shipping=15):

    return price + shipping

print(calculate_total(200))
print(calculate_total(200, 37))
```

# Exercise 2 — User Profile

Create a function called `create_profile` with:

- `name` as a required parameter;
- `country` with `"Brazil"` as the default;
- `active` with `True` as the default.

Print all three values.

Call the function using only the required argument, and then override both default values.

```python
def create_profile(name, country='Brazil', active=True):

    print(name, country, active)

create_profile('Joao')
create_profile('Joao', 'Argentina', active=False)
```

# Exercise 3 — Greeting

Create a function called `greet` with:

- `name` as a required parameter;
- `message` with `"Hello"` as the default.

Call the function three times:

1. using only `name`;
2. overriding `message` positionally;
3. overriding `message` using a keyword argument.

```python
def greet(name, message='Hello'):

    print(name, message)

greet('Joao')
greet('Joao', 'Hi')
greet('Joao', message='Hi')
```

# Exercise 5 — Shopping List

Create a function called `add_product` with:

- `product` as a required parameter;
- `products` with an empty list as the default.

The function must add the product to `products` and return the list.

Call the function twice without providing `products`.

Observe the result of both calls and explain why the second call contains the product from the first call.

```python
def add_product(product, products=[]):

    products.append(product)

    return products

print(add_product('product_A'))
print(add_product('product_B'))
```
