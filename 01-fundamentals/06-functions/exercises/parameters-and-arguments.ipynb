# Exercise 1 — Discount

Create a function called `calculate_discount` that receives a price as a parameter.

The function must return the price with a 10% discount.

Call the function with a price of 250 and print the result.

```python
def calculate_discount(price):

    return price * 0.9

result = calculate_discount(250)

print(result)
```

# Exercise 2 — Rectangle

Create a function called `calculate_perimeter` that receives `width` and `height`.

The function must return the perimeter of the rectangle.

Call the function with 12 and 7 and print the result.

```python
def calculate_perimeter(widht, height):

    return (widht * 2) + (height * 2)

result = calculate_perimeter(12, 7)

print(result)
```

# Exercise 3 — User

Create a function called `create_user` with three parameters:

- `name`
- `age`
- `country`

The function must print all three values.

Call it using positional arguments.

```python
def create_user(name, age, country):

    print(name, age, country)

create_user('Joao', 23, 'Brasil')
```

- `name`
