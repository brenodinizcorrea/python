# Exercise 2 — Same Variable Name

Create a global variable called `message` with the value `"Global"`.

Create a function called `show_message` that creates a local variable also called `message`, with the value `"Local"`.

Print `message` inside the function and outside the function.

```python
message = 'Global'

def show_message():

    message = 'Local'

    print(message)

show_message()

print(message)
```

# Exercise 3 — Global Variable

Create a global variable called `tax` with the value `0.1`.

Create a function called `calculate_tax` that receives a `price` and uses the global `tax` to calculate and return the tax amount.

Call the function with `500`.

```python
tax = 0.1

def calculate_tax(price):

    return price * tax

print(calculate_tax(500))
```

# Exercise 4 — List Scope

Create a global list called `numbers` containing three numbers.

Create a function called `change_numbers`.

Inside the function:

1. assign a new list to a local variable called `numbers`;
2. print the local `numbers`.

After calling the function, print the global `numbers`.

Then modify the exercise so that the function uses `.append()` on `numbers` instead of assigning a new list.

Observe the difference.

```python
numbers = [1, 2, 3]

def change_numbers():

    numbers = [4, 5, 6]

    print(numbers)

change_numbers()

print(numbers)
```

```python
numbers = [1, 2, 3]

def change_numbers():

    numbers.extend([4, 5, 6])

    print(numbers)

change_numbers()

print(numbers)
```

# Exercise 5 — Global Counter

Create a global variable called `counter` with the value `0`.

Create a function called `increment` that increases `counter` by `1`.

Call the function three times and print `counter`.

Use the appropriate Python mechanism to make the function modify the global variable.

```python
counter = 0

def increment():

    global counter

    counter += 1

increment()
increment()
increment()

print(counter)
```
