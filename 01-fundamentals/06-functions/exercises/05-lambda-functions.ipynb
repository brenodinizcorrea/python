# Exercise 1 — Double

Create a lambda function that receives a number and returns its double.

Assign it to `double`.

Use it with the value `15` and print the result.

````python
double = lambda x: x * 2

print(double(15))
````

# Exercise 2 — Calculate Price

Create a lambda function called `calculate_price` that receives:

- `price`
- `quantity`

It must return the total price.

Use it with `25` and `4`.

````python
calculate_price = lambda price, quantity: price * quantity

print(calculate_price(25, 4))
````

# Exercise 3 — Compare Functions

Create two functions that receive a number and return its square:

1. one using `def`;
2. one using `lambda`.

Call both with `8` and print their results.

````python
def square_def(number):

    return number ** 2

print(square_def(8))

square_def = lambda number: number ** 2

print(square_def(8))
````

# Exercise 4 — Transform Values

Create a list:

```python
numbers = [2, 4, 6, 8]
```

Use map() with a lambda to multiply every number by 3.

Convert the result to a list and print it.

````python
numbers = [2, 4, 6, 8]

new_numbers = list(map(lambda x: x * 3, numbers))

print(numbers)
print(new_numbers)
````
