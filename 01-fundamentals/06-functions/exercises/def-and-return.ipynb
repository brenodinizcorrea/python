## Exercise 3 — Function Result

Create a function called `show_message` that prints:

"Processing..."

Call the function and store its result in a variable called `result`.

Print `result`.

Also print its type.

```python
def show_message():

    print('Processing...')

result = show_message()

print(result)
print(type(result))
```

## Exercise 4 — Rectangle

Create a function called `get_dimensions` that returns two values:

12 and 8.

Unpack the returned values into `width` and `height`.

Print both variables.

```python
def get_dimensions():

    return 12, 8

width, height = get_dimensions()

print(width, height)
```

## Exercise 5 — Function Object

Create a function called `calculate` that returns `100`.

Without calling the function, assign the function itself to a variable called `operation`.

Print `operation`.

Then call `operation()` and print its result.

```python
def calculate():

    return 100

operation = calculate
print(operation)

print(operation())
```
