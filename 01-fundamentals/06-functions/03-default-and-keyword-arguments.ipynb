# Default and Keyword Arguments

## The Problem

Imagine that you are creating a function to calculate the final price of an order.

Most orders use the same shipping fee, but occasionally an order may require a different fee.

If the function requires the shipping fee every time, the same value must be repeatedly provided even when nothing changes.

It would be useful for the function to have a standard value that is automatically used when no different value is provided.

## Default Parameters

A parameter can have a default value.

If no argument is provided for that parameter, Python uses its default value.

```python
def calculate_shipping(value, fee=20):
    return value + fee

print(calculate_shipping(100))
print(calculate_shipping(100, 35))
```

## Default Values and Positional Arguments

A default parameter can be overridden by passing a positional argument.

```python
def greet(name, greeting="Hello"):
    print(greeting, name)

greet("Alice")
greet("Alice", "Good morning")
```

## Default Values and Keyword Arguments

A default parameter can also be overridden using a keyword argument.

```python
greet("Alice", greeting="Welcome")
```

## Multiple Default Parameters

A function can have multiple parameters with default values.

```python
def create_account(name, country="Brazil", active=True):
    print(name, country, active)

create_account("Alice")
create_account("Bob", "Canada", False)
```

## Parameter Order

Parameters without default values must come before parameters with default values.

This is invalid:

```python
def calculate(a=10, b):
    return a + b
```

## Default Values Are Evaluated Once

Default parameter values are evaluated when the function is defined, not each time it is called.

This behavior becomes especially important when using mutable objects as default values.

```python
def add_item(item, items=[], c=20):
    items.append(item)
    return items


print(add_item.__defaults__)

print(add_item("A"))
print(add_item.__defaults__)

print(add_item("B"))
print(add_item.__defaults__)
```

# Summary

- Parameters can have default values.
- A default value is used when no argument is provided.
- Default values can be overridden with positional arguments.
- Default values can be overridden with keyword arguments.
- Parameters without default values must come before parameters with default values.
- Default parameter values are evaluated when the function is defined.
- Mutable default values can be reused across function calls.
