# Reading and Writing Files

## The Problem

Your program needs to save the result of a calculation so it can be used later.

The result could be stored in a text file on the computer.

Later, the program may need to open that file, read its content, replace it, or add new content.

## Opening a File

Python uses `open()` to open a file.

It returns a file object that allows the program to interact with the file.

```python
from pathlib import Path

file_path = Path("data/example.txt")

file_path.parent.mkdir(exist_ok=True)

with open(file_path, "w") as file:
    file.write("Python is useful.")
```

```python
file = open(file_path, "r")

print(file)
```

The second argument defines the mode used to open the file.

The `r` mode is used for reading.

```python
file.close()

file = open(file_path, "r")

content = file.read()

print(content)

file.close()
```

## Reading Content

`read()` reads the entire content of the file and returns it as a string.

```python
with open(file_path, "r") as file:
    content = file.read()

print(content)
```

When a file contains multiple lines, `readline()` reads one line at a time.

```python
with open(file_path, "w") as file:
    file.write("First line\n")
    file.write("Second line\n")
    file.write("Third line\n")

with open(file_path, "r") as file:
    print(file.readline())
    print(file.readline())
```

`readlines()` reads all lines and returns them as a list.

```python
with open(file_path, "r") as file:
    lines = file.readlines()

print(lines)
```

A file object can also be iterated over directly.

Each iteration returns the next line.

```python
with open(file_path, "r") as file:
    for line in file:
        print(line.strip())
```

## Writing Content

The `w` mode is used for writing.

Writing to an existing file replaces its previous content.

```python
with open(file_path, "w") as file:
    file.write("New content.")
```

```python
with open(file_path, "r") as file:
    print(file.read())
```

The `a` mode is used to append content.

New content is added after the existing content.

```python
with open(file_path, "a") as file:
    file.write("\nAdditional content.")
```

```python
with open(file_path, "r") as file:
    print(file.read())
```

## Why Close the File?

When we use `open()` without `with`, the file remains open until we close it.

For short scripts, forgetting to close a file may appear to cause no problem.

However, leaving files open can consume system resources and may prevent expected file operations, especially when many files are opened.

For this reason, files opened manually should be closed with `close()`.

```python
file = open(file_path, "r")

print(file.read())

file.close()
```

## Managing the File

A file should be closed after we finish using it.

The `with` statement manages this automatically.

When the block ends, Python closes the file.

```python
with open(file_path, "r") as file:
    print(file.read())

print(file.closed)
```

The basic workflow is:

`Path` → represents the file location

`open()` → opens the file

`read()` / `write()` → interact with its content

`with` → manages the file safely

## Summary

- The main file modes are:

1. `r` — read
2. `w` — write and replace
3. `a` — append

- The main reading methods are:

1. `read()` — read all content
2. `readline()` — read one line
3. `readlines()` — read all lines as a list

- `write()` writes content to a file.
- Using `with` ensures that the file is closed automatically.
