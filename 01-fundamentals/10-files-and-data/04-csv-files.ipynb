# CSV Files

## The Problem

Your program receives data from a CSV file containing information such as products, customers, or sales.

The data is stored as text, but you need to work with individual rows and columns in Python.

For example, you may need to read each product, calculate values from numeric columns, or create a new CSV file with processed data.

How can Python work with tabular data stored in CSV files?

## CSV Files

CSV stands for Comma-Separated Values.

A CSV file represents tabular data.

Each line represents a row, and values inside a row are separated by a delimiter.

```python
csv_content = """name,age,city
Ana,25,Vitoria
Bruno,30,Serra"""

print(csv_content)
```

The file contains structured data, but its content is still text.

The `csv` module provides tools to interpret this text as rows and columns.

```python
import csv
```

## Reading CSV Rows

`csv.reader` reads a CSV file row by row.

Each row is returned as a list.

```python
from pathlib import Path

file_path = Path("data/04-sales.csv")

file_path.parent.mkdir(exist_ok=True)

file_path.write_text(
    "product,quantity,price\n"
    "Laptop,2,3500\n"
    "Mouse,5,80\n"
    "Keyboard,3,250"
)
```

```python
with open(file_path, "r") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

A row such as:

`["Laptop", "2", "3500"]`

is represented as a list.

CSV values are read as strings.

## Writing CSV Rows

`csv.writer` writes Python sequences as CSV rows.

Each row can be provided as a list.

```python
output_path = Path("data/04-new_sales.csv")

with open(output_path, "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["product", "quantity", "price"])
    writer.writerow(("Monitor", 2, 1200))
    writer.writerow({"Headphones", 4, 300})
```

```python
with open(output_path, "r") as file:
    print(file.read())
```

## Reading Rows as Dictionaries

Lists access values by position.

For CSV files with headers, accessing values by column name can make the code clearer.

`csv.DictReader` reads each row as a dictionary.

```python
with open(file_path, "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row)
```

```python
with open(file_path, "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row["product"])
```

`DictReader` uses the first row as dictionary keys.

For example:

`{"product": "Laptop", "quantity": "2", "price": "3500"}`

## Writing Rows as Dictionaries

`csv.DictWriter` writes dictionaries to CSV.

The `fieldnames` argument defines the CSV columns and their order.

```python
dictionary_path = Path("data/04-dictionary_sales.csv")

field_names = ["product", "quantity", "price"]

with open(dictionary_path, "w", newline="") as file:
    writer = csv.DictWriter(
        file,
        fieldnames=field_names
    )

    writer.writeheader()

    writer.writerow({
        "product": "Monitor",
        "quantity": 2,
        "price": 1200
    })

    writer.writerow({
        "product": "Headphones",
        "quantity": 4,
        "price": 300
    })
```

```python
with open(dictionary_path, "r") as file:
    print(file.read())
```

## Reader or DictReader?

`csv.reader`

CSV row → list

Use it when accessing values by position is appropriate.

`csv.DictReader`

CSV row → dictionary

Use it when accessing values by column name makes the code clearer.

## Writer or DictWriter?

`csv.writer`

Python sequence → CSV row

`csv.DictWriter`

Python dictionary → CSV row

## Custom Delimiters

CSV files do not always use commas to separate values.

The `delimiter` argument defines which character separates the values.

For example, a CSV file may use `;` instead of `,`.

```python
from pathlib import Path

file_path = Path("data/04-products_semicolon.csv")

file_path.parent.mkdir(exist_ok=True)

file_path.write_text(
    "product;quantity;price\n"
    "Laptop;2;3500\n"
    "Mouse;5;80\n"
    "Keyboard;3;250"
)
```

```python
with open(file_path, "r") as file:
    reader = csv.reader(file, delimiter=";")

    for row in reader:
        print(row)
```

```python
with open(file_path, "r") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

The same `delimiter` argument can be used when writing CSV files.

It also works with `DictReader` and `DictWriter`.

```python
output_path = Path("data/04-products_semicolon_output.csv")

with open(output_path, "w", newline="") as file:
    writer = csv.writer(file, delimiter=";")

    writer.writerow(["product", "quantity", "price"])
    writer.writerow(["Monitor", 2, 1200])
    writer.writerow(["Headphones", 4, 300])
```

## Summary

- The `csv` module provides tools for working with tabular data stored in CSV files.
- `csv.reader` → reads CSV rows as lists
- `csv.writer` → writes sequences as CSV rows
- `csv.DictReader` → reads CSV rows as dictionaries
- `csv.DictWriter` → writes dictionaries as CSV rows
- `delimiter` → defines the character used to separate values
