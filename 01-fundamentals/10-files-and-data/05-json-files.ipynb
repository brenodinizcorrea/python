# JSON Files

## The Problem

Your program receives structured data from an API or a configuration file.

The data may contain objects, lists, numbers, strings, and nested structures.

The data is represented as JSON text, but your Python program needs to work with Python objects.

How can Python convert between JSON and Python data?

## What Is JSON?

JSON is a standardized text format for representing structured data.

It defines how values and structures are written so that different systems can exchange data.

JSON can represent objects, arrays, strings, numbers, booleans, and null values.

```python
json_content = """{
    "name": "Ana",
    "age": 25,
    "products": ["Laptop", "Mouse"]
}"""

print(json_content)
```

A JSON object uses key-value pairs.

A JSON array is an ordered collection of values.

JSON syntax is similar to some Python structures, but JSON is a data format, not Python syntax.

## JSON and Python

When JSON is interpreted by Python, its structures are converted to corresponding Python objects.

JSON object → Python `dict`

JSON array → Python `list`

JSON string → Python `str`

JSON number → Python `int` or `float`

JSON `true` / `false` → Python `True` / `False`

JSON `null` → Python `None`

```python
json_content = "[2, 3]"

print(json_content)
```

## The `json` Module

Python provides the `json` module in the standard library.

It converts between JSON text and Python objects.

```python
import json
```

## `loads()` — JSON String to Python

`json.loads()` reads JSON from a string and converts it into a Python object.

The `s` stands for string.

```python
json_content = '{"name": "Ana", "age": 25}'

data = json.loads(json_content)

print(data)
print(type(data))

print(data["name"])
print(data["age"])
```

```python
json_content = '[2, 3]'

data = json.loads(json_content)

print(data)
print(type(data))
```

```python
json_content = '2'

data = json.loads(json_content)

print(data)
print(type(data))
```

```python
json_content = 'true'

data = json.loads(json_content)

print(data)
print(type(data))
```

```python
json_content = 'null'

data = json.loads(json_content)

print(data)
print(type(data))
```

## `dumps()` — Python to JSON String

`json.dumps()` converts a Python object into a JSON string.

The `s` stands for string.

```python
data = {
    "name": "Bruno",
    "age": 30,
    "city": "Serra"
}

json_content = json.dumps(data)

print(json_content)
print(type(json_content))
```

```python
data = [2, 3]

json_content = json.dumps(data)

print(json_content)
print(type(json_content))
```

The difference is:

`loads()` → JSON string → Python object

`dumps()` → Python object → JSON string

## `load()` and `dump()`

When working directly with files:

`json.load()` reads JSON from a file.

`json.dump()` writes a Python object as JSON to a file.

```python
from pathlib import Path

file_path = Path("data/05-dict.json")

file_path.parent.mkdir(exist_ok=True)

data = {
    "name": "Carlos",
    "age": 35,
    "city": "Curitiba"
}

with open(file_path, "w") as file:
    json.dump(data, file)
```

```python
from pathlib import Path

file_path = Path("data/05-dict.json")

with open(file_path, "r") as file:
    data = json.load(file)

print(data)
print(type(data))
print(data["name"])
```

```python
from pathlib import Path

file_path = Path("data/05-list.json")

file_path.parent.mkdir(exist_ok=True)

data = [2, 3]

with open(file_path, "w") as file:
    json.dump(data, file)
```

```python
from pathlib import Path

file_path = Path("data/05-list.json")

with open(file_path, "r") as file:
    data = json.load(file)

print(data)
print(type(data))
print(data[0])
```

The four functions can be understood as two pairs:

`loads()` / `dumps()` → work with strings

`load()` / `dump()` → work with files

## JSON Files

JSON is often stored in files with the `.json` extension.

The extension is a convention that identifies the expected format of the file.

Python does not require the `.json` extension to read JSON.

The content is what must follow the JSON format.

```python
from pathlib import Path

file_path = Path("data/05-list-text.txt")

file_path.parent.mkdir(exist_ok=True)

data = [2, 3]

with open(file_path, "w") as file:
    json.dump(data, file)
```

```python
from pathlib import Path

file_path = Path("data/05-list-text.txt")

with open(file_path, "r") as file:
    data = json.load(file)

print(data)
print(type(data))
print(data[0])
```

## Nested JSON

JSON structures can contain other structures.

An object can contain arrays, and arrays can contain objects.

This allows JSON to represent complex hierarchical data.

```python
json_content = """{
    "name": "Ana",
    "address": {
        "city": "Salvador",
        "state": "BA"
    },
    "orders": [
        {
            "product": "Laptop",
            "quantity": 1
        },
        {
            "product": "Mouse",
            "quantity": 2
        }
    ]
}"""

print(json_content)
```

```python
data = json.loads(json_content)

print(data["address"]["city"])
print(data["orders"][0]["product"])
print(data["orders"][1]["quantity"])
```

After conversion, the nested JSON structure becomes nested Python dictionaries and lists.

The access syntax follows the structure:

`dict → dict`

`dict → list → dict`

## Formatting JSON

The `indent` argument formats JSON with indentation, making nested structures easier to read.

```python
print(json.dumps(data, indent=4))
```

## Summary

- JSON is a standardized text format for representing structured data.
- JSON objects correspond to Python dictionaries.
- JSON arrays correspond to Python lists.
- `loads()` → JSON string to Python object.
- `dumps()` → Python object to JSON string.
- `load()` → JSON file to Python object.
- `dump()` → Python object to JSON file.
- JSON can contain nested objects and arrays.
- `.json` is a conventional file extension; the content defines whether the data is valid JSON.
