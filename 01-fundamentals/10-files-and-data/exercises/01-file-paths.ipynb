# Exercise 1 — Create a Path

Create a `Path` object representing the relative path `data/sales.csv`.

Print the path and its file name.

# Exercise 2 — Relative vs Absolute

Create two `Path` objects:

- `data/sales.csv`
- `/tmp/data/sales.csv`

Print whether each path is absolute.

```python
import pathlib

first_path = pathlib.Path('data/sales.csv')
second_path = pathlib.Path('/tmp/data/sales.csv')

print(f'Path: {first_path}')
print(f'Is Absolute: {first_path.is_absolute()}')

print(f'Path: {second_path}')
print(f'Is Absolute: {second_path.is_absolute()}')
```

# Exercise 4 — Build a Path

Build the following path using `Path` and the `/` operator:

`data/reports/2026/august/sales.csv`

Do not write the complete path as a single string.

Print the resulting path.

```python
import pathlib

data_path = pathlib.Path('data')
reports_path = pathlib.Path('reports')
year_path = pathlib.Path('2026')
month_path = pathlib.Path('august')
file_path = pathlib.Path('sales.csv')

final_path = data_path / reports_path / year_path / month_path / file_path

print(final_path)
```

# Exercise 6 — Absolute Path Resets the Path

Create the following paths:

- `base = Path("data/reports")`
- `absolute = Path("/tmp")`
- `file = "sales.csv"`

Combine them using `/`:

`base / absolute / file`

Print the resulting path.

Then create another path using only relative components:

`base / "2026" / file`

Print the result as well.

Observe why the first path does not start with `data/reports`.

```python
import pathlib

base = pathlib.Path("data/reports")
absolute  = pathlib.Path("/tmp")
file  = pathlib.Path("sales.csv")

combined_path = base / absolute / file

print(combined_path)

another_combined_path = base / '2026' / file

print(another_combined_path)
```
