## Exercise 2 — Returning a Context Value

Create a context manager called `Connection`.

Its `__enter__()` method should print:

```text
Connection opened
```

Then return the string:

```text
Connected
```

Use `as` to assign the returned value to a variable called `connection`.

Print the value inside the `with` block.

When leaving the context, print:

```text
Connection closed
```

````python
class Connection():

    def __enter__(self):
        print('Connection opened')
        return 'Connected'

    def __exit__(self, exception_type, exception_value, traceback):
        print('Connection closed')

with Connection() as connection:
    print(connection)
````

## Exercise 3 — Automatic Exit

Create a context manager called `Resource`.

Its `__enter__()` method should print:

```text
Resource acquired
```

Its `__exit__()` method should print:

```text
Resource released
```

Inside the `with` block, print:

```text
Using resource
```

After the `with` block, print:

```text
After context
```

Observe when `__exit__()` is executed.

````python
class Resource():

    def __enter__(self):
        print('Resource acquired')

    def __exit__(self, exception_type, exception_value, traceback):
        print('Resource released')

with Resource():
    print('Using resource')

print('After context')
````

## Exercise 4 — Exit After an Exception

Create a context manager called `Example`.

Its `__enter__()` method should print:

Entering

Its `__exit__()` method should print:

Exiting

Inside the `with` block:

1. Print:
   Working

2. Raise a `ValueError` with the message:
   Something went wrong

Run the code and observe whether `__exit__()` is executed.

````python
class Example():

    def __enter__(self):
        print('Entering')

    def __exit__(self, exception_type, exception_value, traceback):
        print('Exiting')

with Example() as example:
    print('Working')
    raise ValueError('Something went wrong')
````

## Exercise 5 — Inspecting an Exception

Create a context manager called `ErrorLogger`.

Its `__enter__()` method should print:

```print
Starting
```

Its `__exit__()` method should print the exception type and exception value.

Inside the `with` block, raise:

```print
ValueError("Invalid value")
```

Run the code and inspect the values received by `__exit__()`.

````python
class ErrorLogger():

    def __enter__(self):
        print('Starting')

    def __exit__(self, exception_type, exception_value, traceback):
        print(exception_type)
        print(exception_value)
        print(traceback)
        print('Exiting')

with ErrorLogger() as errorlogger:
    raise ValueError('Invalid value')
````
