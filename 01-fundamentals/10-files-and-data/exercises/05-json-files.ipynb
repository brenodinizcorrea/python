# Exercise 1 — `loads()`

The variable `json_content` contains:

```python
{"product": "Laptop", "price": 3500, "quantity": 2}
```

Use `json.loads()` to convert the JSON string into a Python object.

Print the object and the product name.

````python
import json

json_content = '''{
    "product": "Laptop", 
    "price": 3500, 
    "quantity": 2
}'''

json_converted = json.loads(json_content)

print(json_content)
print(json_converted)
print(type(json_converted))
print(json_converted['product'])
````

# Exercise 2 — `dumps()`

Create this Python dictionary:

```python
{
    "name": "Bruno",
    "age": 30,
    "city": "Serra"
}
```

Use `json.dumps()` to convert it into a JSON string.

Print the result and its type.

````python
import json

json_dict = {
    "name": "Bruno",
    "age": 30,
    "city": "Belo Horizonte"
}

json_string = json.dumps(json_dict)

print(json_dict)
print(type(json_dict))
print(json_string)
print(type(json_string))
````

# Exercise 3 — JSON File

Create `data/product.json`.

Use `json.dump()` to write:

```python
{
    "product": "Monitor",
    "price": 1200,
    "quantity": 2
}
```

Then use `json.load()` to read the file.

Print the product name and price.

````python
import pathlib
import json

file_path = pathlib.Path('data/05-product.json')

file_path.parent.mkdir(exist_ok=True)

json_dict = {
    "product": "Monitor",
    "price": 1200,
    "quantity": 2
}

with open(file_path, 'w') as file:
    json.dump(json_dict, file)

with open(file_path, 'r') as file:
    content = json.load(file)

print(content)
print(content['product'])
print(content['price'])
````

# Exercise 4 — Read Nested JSON

Create the file `data/customer.json` with this JSON structure:

```python
{
    "customer": {
        "name": "Ana",
        "city": "Salvador"
    },
    "orders": [
        {
            "product": "Laptop",
            "quantity": 1
        },
        {
            "product": "Mouse",
            "quantity": 2
        }
    ]
}
```

Use `json.dump()` to write the data to the file.

Then use `json.load()` to read the file.

Print the customer name and the product of the first order.

````python
import pathlib
import json

orders = {
    "customer": {
        "name": "Ana",
        "city": "Vitoria"
    },
    "orders": [
        {
            "product": "Laptop",
            "quantity": 1
        },
        {
            "product": "Mouse",
            "quantity": 2
        }
    ]
}

file_path = pathlib.Path('data/05-customers.json')

file_path.parent.mkdir(exist_ok=True)

with open(file_path, 'w') as file:
    json.dump(orders, file)

with open(file_path, 'r') as file:
    content = json.load(file)

print(content['customer']['name'])
print(content['orders'][0]['product'])
````

# Exercise 5 — Nested List

Create this Python dictionary:

```python
{
    "customer": "Carlos",
    "orders": [
        {"product": "Laptop", "quantity": 1},
        {"product": "Mouse", "quantity": 2},
        {"product": "Keyboard", "quantity": 1}
    ]
}
```

Use a `for` loop to print each product and quantity.

Then use `json.dumps()` with `indent=4` to print the complete structure as formatted JSON.

````python
import json

orders = {
    "customer": "Carlos",
    "orders": [
        {"product": "Laptop", "quantity": 1},
        {"product": "Mouse", "quantity": 2},
        {"product": "Keyboard", "quantity": 1}
    ]
}

for item in orders['orders']:
    print(item['product'])

print(json.dumps(orders, indent=4))
````
