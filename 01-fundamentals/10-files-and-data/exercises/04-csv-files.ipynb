# Exercise 1 — Read CSV Rows

A file called `data/products.csv` contains:

```python
product,quantity,price
Laptop,2,3500
Mouse,5,80
Keyboard,3,250
```

Use `csv.reader` to open the file and print each row.

Then print the product name from each data row using its position in the list.

````python
import pathlib
import csv

file_path = pathlib.Path('data/04-products.csv')

file_path.parent.mkdir(exist_ok=True)

with open(file_path, 'w') as file:
    file.write(
        'product,quantity,price\n'
        'Laptop,2,3500\n'
        'Mouse,5,80\n'
        'Keyboard,3,250\n'
    )

with open(file_path, 'r') as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)

        if row[0] != 'product':
            print(row[0])
````

# Exercise 2 — Custom Delimiter

A file called `data/products_semicolon.csv` contains:

```python
product;quantity;price
Laptop;2;3500
Mouse;5;80
Keyboard;3;250
```

Use `csv.reader` to read the file.

The values are separated by `;`, not by `,`.

Print each row and then print the product name from each data row.

````python
import pathlib
import csv

filepath = pathlib.Path('data/04-products_semicolon.csv')

filepath.parent.mkdir(exist_ok=True)

with open(filepath, 'w') as file:
    file.write(
        'product;quantity;price\n'
        'Laptop;2;3500\n'
        'Mouse;5;80\n'
        'Keyboard;3;250\n'
    )

with open(filepath, 'r') as file:
    reader = csv.reader(file, delimiter=';')

    for row in reader:
        print(row)
        print(row[0])
````

# Exercise 3 — Write CSV Rows

Create a file called `data/orders.csv`.

Write the following data using `csv.writer`:

```python
product,quantity
Monitor,2
Headphones,4
Keyboard,3
```

After writing the file, read it and print its content to verify the result.

````python
import pathlib
import csv

file_path = pathlib.Path('data/04-orders.csv')

file_path.parent.mkdir(exist_ok=True)

with open(file_path, 'w') as file:
    writer = csv.writer(file)

    writer.writerow(['product', 'quantity'])
    writer.writerow(['Monitor', '2'])
    writer.writerow(['Headphones', '4'])
    writer.writerow(['Keyboard', '3'])
````

# Exercise 4 — Read with `DictReader`

The file `data/products.csv` contains product information with a header.

Use `csv.DictReader` to read the file.

For each product, print:

```python
Product: <product name>
Price: <price>
```

Access the values using their column names, not their positions.

````python
import csv
import pathlib

file_path = pathlib.Path('data/04-products.csv')

with open(file_path, 'r') as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row)
````

# Exercise 5 — Write with `DictWriter`

Create `data/customers.csv` with the following columns:

```python
name,email,city
```

Write these customers using `csv.DictWriter`:

```python
Ana,ana@example.com,Salvador
Bruno,bruno@example.com,Manaus
Carlos,carlos@example.com,Curitiba
```

Include the header in the file.

````python
import pathlib
import csv

file_path = pathlib.Path('data/04-customers.csv')

fieldnames = ['name', 'email', 'city']

file_path.parent.mkdir(exist_ok=True)

with open(file_path, 'w') as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()

    writer.writerow({
        'name': 'Ana',
        'email': 'ana@example.com',
        'city': 'Salvador'
    })
    writer.writerow({
        'name': 'Bruno',
        'email': 'bruno@example.com',
        'city': 'Manaus'
    })
    writer.writerow({
        'name': 'Carlos',
        'email': 'carlos@example.com',
        'city': 'Curitiba'
    })
````
