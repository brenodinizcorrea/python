# Exercise 1 — Read a File

Create a `Path` for `data/message.txt`.

Open the file in read mode and print its entire content.

Use `with` to manage the file.

````python
import pathlib

file_path = pathlib.Path('data/message.txt')

with open(file_path, 'r') as file:
    print(file)
    print(file.read())
````

# Exercise 2 — Read Lines

Create a `Path` for `data/products.txt`.

Open the file and print each line without the newline character at the end.

Use a `for` loop to iterate over the file.

````python
import pathlib

file_path = pathlib.Path('data/products.txt')

with open(file_path, 'r') as file:
    for single_product in file:
        single_product = single_product.strip()
        print(single_product)
````

# Exercise 3 — Write a File

Create a `Path` for `data/report.txt`.

Open the file in write mode and save these three lines:

```text
Sales: 150
Returns: 12
Net sales: 138
```

Then open the file in read mode and print its content to verify the result.

````python
import pathlib

file_path = pathlib.Path('data/report.txt')

with open(file_path, 'w') as file:
    file.write('Sales: 150\nReturns: 12\nNet sales: 138')

with open(file_path, 'r') as file:
    print(file.read())
````

# Exercise 4 — Append Content

Use the file `data/report.txt` from the previous exercise.

Open it in append mode and add this line:

```text
Status: completed
```

Then read the file and print its complete content.

````python
import pathlib

file_path = pathlib.Path('data/report.txt')

with open(file_path, 'a') as file:
    file.write('\nStatus: completed')

with open(file_path, 'r') as file:
    print(file.read())
````

# Exercise 5 — Replace File Content

Create a `Path` for `data/status.txt`.

First, write this content to the file:

```text
Status: pending
```

Then replace its content using write mode with:

```text
Status: completed
```

Finally, read the file and print its content.

Use `with` for every file operation.

````python
import pathlib

file_path = pathlib.Path('data/status.txt')

with open(file_path, 'w') as file:
    file.write('Status: pending')

with open(file_path, 'r') as file:
    print(file.read())

with open(file_path, 'w') as file:
    file.write('Status: completed')

with open(file_path, 'r') as file:
    print(file.read())
````
