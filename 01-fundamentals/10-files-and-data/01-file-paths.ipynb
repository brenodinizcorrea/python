# File Paths

## The Problem

Programs often need to work with files stored on the computer.

For example, a program may need to find a CSV file, check whether a directory exists, or read a text file.

Before interacting with a file, Python needs to know where that file is located.

## Where is Python running?

Before looking for a file, we need to know the directory from which Python is currently working.

`Path.cwd()` returns the current working directory.

````python
from pathlib import Path

current_directory = Path.cwd()

print(current_directory)
````

This directory is the starting point for relative paths.

If we refer to `data/sales.csv`, Python looks for `data` inside the current working directory.

````python
from pathlib import Path

file_path = Path("data/sales.csv")

print(file_path)
print(file_path.exists())
````

A `Path` represents a location in the filesystem.

Creating a `Path` does not create or open the file.

It only gives Python an object that represents that location.

## Finding something in the current directory

We can use a path to refer to a directory and inspect what is actually there.

`iterdir()` returns the files and directories contained in a directory.

````python
from pathlib import Path

current_directory = Path.cwd()

print(current_directory)

for item in current_directory.iterdir():
    print(item)
````

Now we can interact with real locations on the computer.

Each item returned by `iterdir()` is a `Path` object representing a file or directory.

````python
from pathlib import Path

current_directory = Path.cwd()

for item in current_directory.iterdir():
    print(item.name)
    print("Directory:", item.is_dir())
    print("File:", item.is_file())
    print()
````

````python

file_path = Path("data/sales.csv")

print(file_path.name)
print("Directory:", file_path.is_dir())
print("File:", file_path.is_file())
````

## Navigating the filesystem

Paths form a hierarchy.

For example:

```text
project/
└── data/
    └── sales.csv
```

A path can be built by moving through this hierarchy.

````python
from pathlib import Path

project_path = Path("project")
data_path = project_path / "data"
sales_file = data_path / "sales.csv"

print(sales_file)
````

````python
from pathlib import Path

project_path = Path("project")
data_path = Path("data")
sales_file = Path("sales.csv")

data_path = project_path / data_path
sales_file = data_path / sales_file

print(sales_file)
````

The `/` operator combines `Path` objects.

This is preferable to manually concatenating strings because `Path` understands filesystem paths.

## Navigating up

A path also lets us move toward its parent directories.

`.parent` returns the directory that contains the current path.

````python
from pathlib import Path

sales_file = Path("project/data/sales.csv")

print(sales_file)
print(sales_file.parent)
print(sales_file.parent.parent)
````

## Relative paths

`project/data/sales.csv` is a relative path.

It describes the location starting from the current working directory.

Relative paths are useful when working with files inside a project.

````python
from pathlib import Path

relative_path = Path("project/data/sales.csv")

print(relative_path)
print(relative_path.is_absolute())
````

## Absolute paths

An absolute path describes the complete location from the filesystem root.

`Path.cwd()` gives us an absolute path for the current working directory.

````python
from pathlib import Path

absolute_path = Path.cwd()

print(absolute_path)
print(absolute_path.is_absolute())
````

We can combine an absolute starting point with a relative path to obtain the complete location of a file.

````python
from pathlib import Path

current_directory = Path.cwd()
relative_path = Path("project/data/sales.csv")

absolute_path = current_directory / relative_path

print(absolute_path)
````

## Useful path information

A `Path` can also tell us about the different parts of a location.

- `.name` → final name
- `.stem` → name without the extension
- `.suffix` → extension
- `.parent` → containing directory

````python
from pathlib import Path

path = Path("project/data/sales.csv")

print(path.name)
print(path.stem)
print(path.suffix)
print(path.parent)
````

## Summary

- Working with files starts with representing their locations.
- `Path` represents filesystem paths as Python objects.
- We can use `Path` to:

1. find the current working directory
2.  build paths
3.  navigate through directories
4.  inspect files and directories
5.  work with relative and absolute paths

- A path describes where something is. It does not necessarily mean that the file or directory exists.
