# List Comprehensions

## The Problem

Imagine that you receive a list of prices and need to create another list containing the prices after applying a 10% discount.

A straightforward solution is to create an empty list, iterate over the original values, calculate each discount, and add the result to the new list.

This works, but for simple transformations the code requires several lines to express a relatively simple operation.

Python provides a more concise way to create a new list from the elements of an iterable.

## The Traditional Approach

A `for` loop can be used to create a new list step by step.

`````python
prices = [100, 200, 300, 400]

discounted_prices = []

for price in prices:
    discounted_prices.append(price * 0.9)

print(discounted_prices)
`````

## List Comprehension

A list comprehension creates a new list by describing the transformation directly inside the list expression.

`````python
prices = [100, 200, 300, 400]

discounted_prices = [price * 0.9 for price in prices]

print(discounted_prices)
`````

## Structure

A basic list comprehension has three main parts:

```text
[expression for item in iterable]
````

The expression determines what will be placed in the new list.

The for part determines how the elements are traversed.

`````python
numbers = [1, 2, 3, 4]

squares = [number ** 2 for number in numbers]

print(squares)
`````

## Transforming Values

The expression can transform each element before placing it in the new list.

`````python
names = ["alice", "bob", "charlie"]

uppercase_names = [name.upper() for name in names]

print(uppercase_names)
`````

## Filtering Values

A list comprehension can also include an `if` condition to select which elements are included.

`````python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = [number for number in numbers if number % 2 == 0]

print(even_numbers)
`````

## Transforming and Filtering

The transformation and filtering can be combined.

`````python
numbers = [1, 2, 3, 4, 5, 6]

squares_of_even_numbers = [
    number ** 2
    for number in numbers
    if number % 2 == 0
]

print(squares_of_even_numbers)
`````

## List Comprehension vs. `for`

A list comprehension is not a different kind of loop.

It provides another syntax for creating a list while iterating over an iterable.

For simple transformations and filters, comprehensions can make the intent easier to read.

For more complex logic, a regular `for` loop may be clearer.

# Summary

- A list comprehension creates a new list from an iterable.
- The expression determines the value added to the new list.
- The `for` part determines how the iterable is traversed.
- An `if` condition can filter which elements are included.
- Transformation and filtering can be combined.
- List comprehensions provide a concise alternative to some `for` loops.
- A regular `for` loop can be clearer when the logic becomes more complex.
