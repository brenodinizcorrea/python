# Iterables and Iterators

## The Problem

Imagine that you have a collection of values and want to process each value one at a time.

You already know that a `for` loop can do this:

```python
numbers = [10, 20, 30]

for number in numbers:
    print(number)
```

The loop works automatically, but there is an important mechanism hidden behind it.

Python needs a way to obtain the next element from `numbers` until there are no more elements.

To understand how this works, we need to understand two concepts: iterables and iterators.

## Iterables

An **iterable** is an object whose elements can be traversed one at a time.

Common examples include:

- lists
- tuples
- strings
- sets
- dictionaries
- ranges

If an object can be used directly in a `for` loop, it is generally an iterable.

````python
numbers = [10, 20, 30]

for number in numbers:
    print(number)
````

Here, `numbers` is an iterable.

It contains multiple values that Python can traverse one at a time.

## `iter()`

Python provides the `iter()` function to obtain an **iterator** from an iterable.

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(iterator)
````

The list is the **iterable**.

The object returned by `iter()` is the **iterator**.

The iterator is responsible for providing the elements one at a time.

## `next()`

Python provides the `next()` function to request the next value from an iterator.

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
````

Each call to `next()` advances the iterator and returns the next element.

## Iterator Keeps Its Position

An iterator remembers where it is in the iteration.

Calling `next()` again does not start from the beginning.

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
````

````python
numbers = {'a':10, 'b':20, 'c':30}

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
````

The iterator has already consumed `10`.

The next call therefore returns `20`, not `10` again.

## When the Iterator Ends

After all elements have been returned, there is nothing left for the iterator to provide.

Calling `next()` again raises `StopIteration`.

`StopIteration` is Python's way of indicating that the iterator has no more values.

````python
numbers = [10, 20]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
````

## How `for` Uses Iterators

The `for` loop hides this process.

When Python receives an iterable in a `for` loop, it obtains an iterator and repeatedly requests its next value until the iterator is exhausted.

````python
numbers = [10, 20, 30]

for number in numbers:
    print(number)
````

Conceptually, Python performs the equivalent of:

1. Obtain an iterator from `numbers`.
2. Get the next value.
3. Execute the loop body.
4. Get the next value.
5. Repeat until `StopIteration`.

The `for` loop handles this automatically.

## Iterable vs. Iterator

The distinction is:

- **Iterable** → an object that can be traversed.
- **Iterator** → an object that provides the values one at a time and keeps track of the current position.

For example:

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(type(numbers))
print(type(iterator))
````

````python
numbers = {'a':10, 'b':20, 'c':30}

iterator = iter(numbers)

print(type(numbers))
print(type(iterator))
````

The list and the iterator are different objects with different roles.

```text
numbers
   │
   │ iter()
   ↓
iterator
   │
   ├── next() → 10
   ├── next() → 20
   └── next() → 30

## Iterators Are Also Iterables

An iterator is also an iterable.

Calling `iter()` on an iterator returns the iterator itself.

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(iterator is iter(iterator))
````

## `range()` and Iteration

`range` is another iterable that can be traversed with a `for` loop.

````python
numbers = range(5)

for number in numbers:
    print(number)
````

The important point is not how `range` is implemented internally, but that it can provide its values during iteration.

# Summary

- An iterable is an object whose elements can be traversed.
- Lists, tuples, strings, sets, dictionaries, and ranges are examples of iterables.
- `iter()` obtains an iterator from an iterable.
- An iterator provides values one at a time.
- `next()` requests the next value from an iterator.
- An iterator keeps track of its current position.
- An exhausted iterator raises `StopIteration`.
- A `for` loop automatically handles the iteration process.
- An iterator is also an iterable.
- Iteration allows Python to process values one at a time.
