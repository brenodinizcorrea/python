# Set and Dictionary Comprehensions

## The Problem

Imagine that you receive a list of numbers containing repeated values and need to create a collection containing only the unique values.

In another situation, you receive a list of products and need to create a dictionary containing each product and its calculated price.

A `for` loop can solve both problems, but the resulting code can become unnecessarily verbose when the operation is simple.

Python provides comprehension syntax for creating sets and dictionaries in the same concise way we used with lists.

## Set Comprehensions

A set comprehension creates a new set by iterating over an iterable.

Its basic structure is:

[expression for item in iterable]

The resulting collection is a `set`, so duplicate values are removed.

```python
numbers = [1, 2, 2, 3, 4, 4, 5]

unique_numbers = {number for number in numbers}

print(unique_numbers)
```

## Transforming Values

The expression can transform each element before adding it to the set.

```python
numbers = [1, 2, 3, 4, 5]

squared_numbers = {number ** 2 for number in numbers}

print(squared_numbers)
```

## Filtering Sets

A set comprehension can use an `if` condition to filter elements.

```python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = {
    number
    for number in numbers
    if number % 2 == 0
}

print(even_numbers)
```

## Dictionary Comprehensions

A dictionary comprehension creates a new dictionary.

Its basic structure is:

{key: value for item in iterable}

```python
numbers = [1, 2, 3, 4]

squares = {
    number: number ** 2
    for number in numbers
}

print(squares)
```

## Transforming Keys and Values

The key and value expressions can be created from the current element.

```python
prices = {
    "apple": 5,
    "banana": 3,
    "orange": 4
}

discounted_prices = {
    product: price * 0.9
    for product, price in prices.items()
}

print(discounted_prices)
```

## Filtering Dictionaries

A dictionary comprehension can also use an `if` condition to select which entries are included.

```python
prices = {
    "apple": 5,
    "banana": 3,
    "orange": 8,
    "mango": 10
}

expensive_products = {
    product: price
    for product, price in prices.items()
    if price >= 8
}

print(expensive_products)
```

## List vs. Set vs. Dictionary

The comprehension syntax follows the same general idea:

- List → `[]`
- Set → `{}`
- Dictionary → `{key: value}`

The brackets determine the type of collection being created.

```python
numbers = [1, 2, 3, 4]

list_result = [number * 2 for number in numbers]

set_result = {number * 2 for number in numbers}

dict_result = {
    number: number * 2
    for number in numbers
}

print(list_result)
print(set_result)
print(dict_result)
```

# Summary

- Set comprehensions create new sets from iterables.
- Set comprehensions automatically eliminate duplicate values.
- Dictionary comprehensions create new dictionaries.
- Dictionary comprehensions produce key-value pairs.
- Set and dictionary comprehensions can transform values.
- Set and dictionary comprehensions can filter elements using `if`.
- The collection syntax determines whether the result is a list, set, or dictionary.
- Comprehensions are useful when the transformation or filtering logic is simple.
