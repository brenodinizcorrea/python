# Exercício 1 — Criando e percorrendo um iterator

Dada a lista:

    numbers = [10, 20, 30, 40]

Crie um iterator a partir de `numbers`.

Depois:

1. Use `next()` para obter o primeiro elemento.
2. Use `next()` novamente para obter o segundo.
3. Use `next()` novamente para obter o terceiro.
4. Armazene cada resultado em uma variável.
5. Imprima as três variáveis.

```python
numbers = [10, 20, 30, 40]

iterator = iter(numbers)

first = next(iterator)
second = next(iterator)
third = next(iterator)

print(first)
print(second)
print(third)
```

# Exercício 2 — Consumindo um iterator

Crie um iterator a partir da lista:

    names = ["Ana", "Bruno", "Carlos", "Daniel"]

Use `next()` quatro vezes para percorrer todos os elementos.

Depois, tente executar um quinto `next()`.

Observe o que acontece quando o iterator não possui mais elementos.

# Exercício 4 — Iterator de um dictionary

Dado:

    person = {
        "name": "Ana",
        "age": 30,
        "city": "Vitória"
    }

Crie um iterator usando:

    iterator = iter(person)

Use `next()` para percorrer o dictionary.

Depois, faça o mesmo utilizando:

    iterator = iter(person.keys())

Compare os resultados.

```python
person = {
    'name': 'Ana',
    'age': 30,
    'city': 'Vitória'
}

iterator = iter(person)

print(next(iterator))
print(next(iterator))
print(next(iterator))

iterator = iter(person.keys())

print(next(iterator))
print(next(iterator))
print(next(iterator))
```

# Exercício 5 — Keys e values

Dado:

    person = {
        "name": "Ana",
        "age": 30,
        "city": "Vitória"
    }

Crie três iterators:

    key_iterator = iter(person.keys())
    value_iterator = iter(person.values())
    item_iterator = iter(person.items())

Use `next()` para percorrer cada um deles.

Observe a diferença entre os elementos produzidos por cada iterator.

```python
person = {
    'name': 'Ana',
    'age': 30,
    'city': 'Vitória'
}

key_iterator = iter(person.keys())
value_iterator = iter(person.values())
item_iterator = iter(person.items())

print(next(key_iterator))
print(next(key_iterator))
print(next(key_iterator))

print(next(value_iterator))
print(next(value_iterator))
print(next(value_iterator))

print(next(item_iterator))
print(next(item_iterator))
print(next(item_iterator))
```

# Exercício 6 — Identificando o tipo dos iterators

Crie iterators a partir dos seguintes objetos:

    numbers = [10, 20, 30]
    names = ("Ana", "Bruno", "Carlos")
    word = "Python"
    numbers_set = {10, 20, 30}
    person = {"name": "Ana", "age": 30}

Para cada um:

1. Crie um iterator usando `iter()`.
2. Use `type()` para descobrir seu tipo.
3. Imprima o tipo encontrado.

O objetivo é observar que diferentes iterables podem produzir diferentes tipos de iterator.

```python
numbers = [10, 20, 30]
names = ("Ana", "Bruno", "Carlos")
word = "Python"
numbers_set = {10, 20, 30}
person = {"name": "Ana", "age": 30}

list_iterator = iter(numbers)
tuple_iterator = iter(names)
string_iterator = iter(word)
set_iterator = iter(numbers_set)
dict_iterator = iter(person)

print(type(list_iterator))
print(type(tuple_iterator))
print(type(string_iterator))
print(type(set_iterator))
print(type(dict_iterator))
```

# Exercício 7 — Dois iterators independentes

Dada:

    numbers = [10, 20, 30, 40]

Crie dois iterators diferentes a partir da mesma lista:

    iterator_a = iter(numbers)
    iterator_b = iter(numbers)

Use `next()` algumas vezes em `iterator_a` e depois em `iterator_b`.

Observe que os dois possuem posições independentes.

```python
numbers = [10, 20, 30, 40]

iterator_a = iter(numbers)
iterator_b = iter(numbers)

print(next(iterator_a))
print(next(iterator_a))
print(next(iterator_a))

print(next(iterator_b))
```
