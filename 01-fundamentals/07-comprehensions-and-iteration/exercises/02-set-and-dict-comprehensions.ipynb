# Exercise 1 — Unique Numbers

Given:

numbers = [1, 2, 2, 3, 4, 4, 5, 5, 6]

Create a set called `unique_numbers` containing the unique values.

Print the result.

```python
numbers = [1, 2, 2, 3, 4, 4, 5, 5, 6]

unique_numbers = {single_number for single_number in numbers}

print(unique_numbers)
```

# Exercise 2 — Squared Values

Given:

numbers = {1, 2, 3, 4, 5}

Create a new set called `squared_numbers` containing the square of each number.

Print the result.

```python
numbers = {1, 2, 3, 4, 5}

squared_numbers = {single_number ** 2 for single_number in numbers}

print(squared_numbers)
```

# Exercise 3 — Even Numbers

Given:

numbers = [1, 2, 3, 4, 5, 6, 7, 8]

Create a set called `even_numbers` containing only the even values.

Print the result.

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

even_numbers = {single_number for single_number in numbers if single_number % 2 == 0}

print(even_numbers)
```

# Exercise 4 — Number Dictionary

Given:

numbers = [1, 2, 3, 4, 5]

Create a dictionary called `squares` where:

- each number is a key;
- its square is the corresponding value.

Expected structure:

{
    
    1: 1,

    2: 4,
    
    ...
}

Print the result.

```python
numbers = [1, 2, 3, 4, 5]

squares = {single_number: single_number ** 2 for single_number in numbers}

print(squares)
```

# Exercise 5 — Expensive Products

Given:

prices = {

    
    "apple": 5,
    
    "banana": 3,
    
    "orange": 8,
    
    "mango": 10,
    
    "grape": 6
}

Create a dictionary called `expensive_products` containing only products with a price greater than or equal to 8.

Print the result.

```python
prices = {
    "apple": 5,
    "banana": 3,
    "orange": 8,
    "mango": 10,
    "grape": 6
}

expensive_products = {
    key: value 
    for key, value in prices.items()
    if value >= 8
}

print(expensive_products)
```
