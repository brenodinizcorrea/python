# Exercise 1 — Double Values

Given:

numbers = [3, 6, 9, 12, 15]

Create a new list called `doubled` containing each number multiplied by 2.

Print `doubled`.

```python
numbers = [3, 6, 9, 12, 15]

doubled = [item * 2 for item in numbers]

print(doubled)
```

# Exercise 2 — Normalize Names

Given:

names = ["alice", "bob", "charlie", "diana"]

Create a new list called `normalized_names` containing all names in uppercase.

Print `normalized_names`.

```python
names = ['alice', 'bob', 'charlie', 'diana']

normalized_names = [single_name.upper() for single_name in names]

print(normalized_names)
```

# Exercise 3 — Positive Numbers

Given:

numbers = [-5, 3, -2, 8, 0, 10, -1]

Create a new list called `positive_numbers` containing only numbers greater than 0.

Print `positive_numbers`.

```python
numbers = [-5, 3, -2, 8, 10, -1]

positive_numbers = [single_number for single_number in numbers if single_number > 0]

print(positive_numbers)
```

# Exercise 4 — Even Numbers

Given:

numbers = [1, 2, 3, 4, 5, 6, 7, 8]

Create a new list called `squared_even_numbers`.

It must contain the square of every even number.

Print the result.

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

squared_even_numbers = [single_number ** 2 for single_number in numbers if single_number % 2 == 0]

print(squared_even_numbers)
```

# Exercise 5 — Approved Orders

Given:

orders = [120, 80, 250, 40, 300, 150]

Create a new list called `large_orders`.

Only orders greater than or equal to 150 should be included.

Then create another list called `discounted_orders` containing those same orders after applying a 10% discount.

Print both lists.

```python
orders = [120, 80, 250, 40, 300, 150]

large_orders = [single_order for single_order in orders if single_order >= 150]

discounted_orders = [single_order * 0.9 for single_order in large_orders]

print(large_orders)
print(discounted_orders)
```
