# Comparison Operators

## The Problem

Programs often need to determine how values relate to each other.

For example, suppose a program stores a person's age:

age = 20

How can Python determine whether the person is at least 18 years old?

## Comparisons Produce Boolean Values

A comparison is an expression.

When Python evaluates a comparison, the result is a boolean value:

- `True`
- `False`

For example:

```python
10 > 5
```

```python
result = 10 > 5

print(result)
print(type(result))
```

## Equal

The `==` operator checks whether two values are equal.

It is important to distinguish `==` from `=`.

- `=` assigns a value to a variable.
- `==` compares two values.

```python
a = 10
b = 10

result = a == b

print(result)
```

```python
a = 10
b = 20

result = a == b

print(result)
```

## Not Equal

The `!=` operator checks whether two values are different.

```python
a = 10
b = 20

result = a != b

print(result)
```

```python
a = 10
b = 10

print(a != b)
```

## Greater Than

The `>` operator checks whether the value on the left is greater than the value on the right.

```python
10 > 5
```

```python
5 > 10
```

## Less Than

The `<` operator checks whether the value on the left is less than the value on the right.

```python
5 < 10
```

## Greater Than or Equal

The `>=` operator checks whether the value on the left is greater than or equal to the value on the right.

```python
age = 18

print(age >= 18)
```

```python
print(18 >= 18)
print(20 >= 18)
print(17 >= 18)
```

## Less Than or Equal

The `<=` operator checks whether the value on the left is less than or equal to the value on the right.

```python
print(18 <= 18)
print(17 <= 18)
print(20 <= 18)
```

## Comparing Different Numeric Types

Python can compare integers and floating-point numbers.

The comparison produces a boolean value.

```python
a = 10
b = 10.0

print(a == b)
print(a != b)
```

## Comparing Different Types

Python can compare values that belong to different types, but the behavior depends on the comparison operator.

For equality and inequality, Python can compare values of different types.

For example, an integer and a string containing the same characters are not considered equal.

```python
a = 10
b = "10"

print(a == b)
print(a != b)
```

However, the ordering operators cannot compare an `int` and a `str`.

Using `>`, `<`, `>=`, or `<=` between these types raises a `TypeError`.

Because Python does not define a numerical ordering between an integer and a string.

```python
a = 10
b = "10"

print(a > b)
```

## Comparing Strings

Comparison operators can also be used with strings.

For equality, Python checks whether the strings contain the same sequence of characters.

```python
first = "Python"
second = "Python"
third = "Java"

print(first == second)
print(first == third)
```

## The Comparison Does Not Change the Original Values

A comparison evaluates existing values.

It does not modify the objects being compared.

```python
a = 10
b = 20

result = a < b

print(a)
print(b)
print(result)
```

# Summary

- A comparison is an expression.
- A comparison produces a boolean value.
- `==` checks whether two values are equal.
- `!=` checks whether two values are different.
- `>` checks whether the left value is greater.
- `<` checks whether the left value is smaller.
- `>=` checks whether the left value is greater than or equal.
- `<=` checks whether the left value is smaller than or equal.
- `=` assigns a value; `==` compares values.
- The result of a comparison can be assigned to a variable.
- Comparing values does not modify the original objects.
- `==` and `!=` can compare values of different types.
- Ordering operators such as `>`, `<`, `>=`, and `<=` require compatible types.
- Comparing an `int` with a `str` using an ordering operator raises a `TypeError`.
