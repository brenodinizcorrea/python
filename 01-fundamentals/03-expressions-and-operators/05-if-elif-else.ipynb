# Conditional Statements

## The Problem

Programs often need to make decisions.

For example, a program may need to display a message only when a person is an adult.

## `if`

An `if` statement evaluates a condition.

If the condition produces `True`, the indented code is executed.

If the condition produces `False`, the indented code is skipped.

```python
age = 25

if age >= 18:
    print("Adult")
```

## Indentation

Python uses indentation to define which statements belong to an `if` block.

The statements inside the block must be indented.

```python
age = 20

if age >= 18:
    print("Adult")
    print("Can enter")
```

## `else`

`else` defines what should happen when the `if` condition is `False`.

```python
age = 15

if age >= 18:
    print("Adult")
else:
    print("Minor")
```

## `elif`

`elif` allows the program to test another condition when the previous condition was `False`.

Multiple `elif` statements can be used.

```python
age = 15

if age >= 18:
    print("Adult")
elif age >= 13:
    print("Teenager")
else:
    print("Child")
```

## Evaluation Order

Python evaluates conditions from top to bottom.

When it finds a condition that is `True`, its block is executed and the remaining conditions in that `if` structure are skipped.

```python
age = 20

if age >= 18:
    print("Adult")
elif age >= 13:
    print("Teenager")
else:
    print("Child")
```

## Logical Expressions as Conditions

A condition can also be a logical expression.

```python
age = 20
has_ticket = True

if age >= 18 and has_ticket:
    print("Can enter")
```

## `if` Without `else`

An `if` statement does not require an `else`.

If the condition is `False`, Python simply continues with the next statement.

```python
temperature = 25

if temperature > 30:
    print("Hot")

print("Done")
```

# Summary

- `if` conditionally executes a block of code.
- The condition is an expression.
- The block is executed when the condition produces `True`.
- Indentation defines the statements inside the block.
- `else` defines an alternative block when the `if` condition is `False`.
- `elif` allows additional conditions to be tested.
- Conditions are evaluated from top to bottom.
- Once a condition is `True`, the remaining branches are skipped.
- An `if` statement can be used without `elif` or `else`.
- Comparison and logical expressions can be used as conditions.
