# Expressions and Statements

## The Concept Definition

A Python program contains different kinds of code.

Some pieces of code produce values.

Other pieces of code perform actions or control how the program is executed.

To understand Python more clearly, it is useful to distinguish between **expressions** and **statements**.

## Expressions

An **expression** is a piece of code that Python can evaluate to produce a value.

```python
10
```

The expression produces the value `10`.

```python
'Python'
```

This expression produces the string `"Python"`.

```python
True
```

This expression produces the boolean value `True`.

## Variables as Expressions

A variable can also be used as an expression.

When Python evaluates a variable, it obtains the value associated with that variable.

```python
age = 25

age
```

The expression `age` produces the value `25`.

## Arithmetic Expressions

Arithmetic operations are expressions because they produce values.

```python
10 + 5
```

```python
10 * 3
```

Each expression is evaluated and produces a result.

## Comparison Expressions

Comparisons are also expressions.

They produce boolean values.

```python
10 > 5
```

```python
10 == 10
```

The result of each comparison is a value of type `bool`.

## Logical Expressions

Logical operations are expressions as well.

They produce boolean values.

```python
True and False
```

```python
True or False
```

## Expressions Can Be Combined

Expressions can contain other expressions.

```python
age = 25

age >= 18
```

Here, `age` is an expression and `age >= 18` is another expression that uses it.

## Expression Inside a Statement

Expressions frequently appear inside statements.

```python
age = 25
```

This is an **assignment statement**.

The statement assigns a value to a variable.

The expression on the right side is evaluated first.

```python
age = 10 + 5
```

Python first evaluates the expression `10 + 5`.

The result is `15`.

The assignment then makes `age` refer to that value.

## Statements Do Something

A statement generally represents an action or a structure that Python executes.

```python
age = 25
```

The statement changes the program's state by creating or updating a variable reference.

## `print()` and Expressions

`print()` is a function call.

It can receive an expression as an argument.

```python
age = 25

print(age)
```

Here:

- `age` is an expression;
- `print(age)` is a function call statement.

The expression `age` is evaluated first, producing `25`.

That value is then passed to `print()`.

## Expression vs. Statement

Consider the following:

```python
age = 25
```

The expression is:

`25`

The statement is:

`age = 25`

Now consider:

```python
is_adult = age >= 18
```

The expression is:

`age >= 18`

The statement is:

`is_adult = age >= 18`

## A Simple Mental Model

You can think of an expression as answering:

**"What value does this produce?"**

And a statement as answering:

**"What should Python do?"**

```python
10 + 5
```

What value?
→ 15

```python
age = 10 + 5
```

What should Python do?
→ Assign the resulting value to `age`.

# Summary

- An expression is a piece of code that Python can evaluate to produce a value.
- Values such as `10`, `"Python"`, and `True` can be expressions.
- A variable can be used as an expression.
- Arithmetic operations are expressions.
- Comparisons are expressions that produce boolean values.
- Logical operations are expressions that produce boolean values.
- Expressions can be combined to form larger expressions.
- A statement is a piece of code that performs an action or defines a program structure.
- Assignment is a statement.
- Expressions commonly appear inside statements.
- In `is_adult = age >= 18`, `age >= 18` is an expression and the complete line is an assignment statement.
