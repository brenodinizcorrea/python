# Arithmetic Operators

## The Problem

A program often needs to calculate values.

For example, suppose an order contains 4 boxes with 12 items in each box.

We already know how Python represents these quantities:

``` python
boxes = 4
items_per_box = 12
```

But how can Python calculate the total number of items?

## Expressions

An **expression** is a piece of code that Python evaluates to produce a value.

For example:

10 + 5

Python evaluates this expression and produces the value `15`.

````python
10 + 5
````

The result of an expression can be assigned to a variable.

````python
result = 10 + 5

print(result)
````

## Arithmetic Operators

Python provides operators for performing arithmetic.

The basic arithmetic operators are:

- `+` addition
- `-` subtraction
- `*` multiplication
- `/` division
- `//` floor division
- `%` remainder
- `**` exponentiation

## Addition

The `+` operator adds two values.

````python
10 + 5
````

## Subtraction

The `-` operator subtracts one value from another.

````python
10 - 5
````

## Multiplication

The `*` operator multiplies two values.

````python
10 * 5
````

## Division

The `/` operator performs regular division.

The result is a `float`, even when the mathematical result is a whole number.

````python
10 / 2
````

We can verify the type of the result.

````python
result = 10 / 2

print(result)
print(type(result))
````

## Floor Division

The `//` operator performs floor division.

It returns the result of the division rounded down to the nearest integer.

````python
19 // 5
````

Regular division and floor division therefore produce different results.

````python
print(19 / 5)
print(19 // 5)
````

## Remainder

The `%` operator returns the remainder of a division.

````python
19 % 5
````

## Exponentiation

The `**` operator performs exponentiation.

For example, `2 ** 3` means 2 raised to the power of 3.

````python
2 ** 3
````

## Arithmetic with Different Numeric Types

Arithmetic operators can be used with both `int` and `float`.

When a calculation involves a `float`, the result is generally a `float`.

````python
price = 19.90
quantity = 2

total = price * quantity

print(total)
print(type(total))
````

## Summary

- An **expression** is evaluated by Python to produce a value.
- Arithmetic operators allow us to perform calculations.
- `+` performs addition.
- `-` performs subtraction.
- `*` performs multiplication.
- `/` performs regular division and produces a `float`.
- `//` performs floor division.
- `%` produces the remainder of a division.
- `**` performs exponentiation.
- The result of an expression can be assigned to a variable.
- Expressions can be combined to create new expressions.
