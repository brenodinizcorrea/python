# Exercise 1 — `and`

Create two variables:

- `a` containing `True`;
- `b` containing `False`.

Create a variable called `result` containing the result of `a and b`.

Print `a`, `b`, and `result`.

Then print the type of `result`.

```python
a = True
b = False

result = a and b

print(a)
print(b)
print(result)
print(type(result))
```

# Exercise 2 — `or`

Create three variables:

- `first` containing `False`;
- `second` containing `False`;
- `third` containing `True`.

Create two variables:

- `result_1` containing `first or second`;
- `result_2` containing `second or third`.

Print both results.

Then print the type of each result.

```python
first = False
second = False
third = True

result_1 = first or second
result_2 = second or third

print(result_1)
print(result_2)

print(type(result_1))
print(type(result_2))
```

# Exercise 4 — Comparisons and `and`

Create two variables:

- `age` containing 25;
- `has_ticket` containing `True`.

Create two variables:

- `is_adult` containing the result of checking whether `age` is greater than or equal to 18;
- `can_enter` containing the result of combining `is_adult` and `has_ticket` with `and`.

Print `is_adult` and `can_enter`.

Then print the type of both results.

```python
age = 25
has_ticket = True

is_adult = age >= 18
can_enter = is_adult and has_ticket

print(is_adult)
print(can_enter)

print(type(is_adult))
print(type(can_enter))
```

# Exercise 5 — Combining Logical Expressions

Create three variables:

- `age` containing 16;
- `has_parent` containing `True`;
- `has_ticket` containing `True`.

Create a variable called `is_adult` containing the result of checking whether `age` is greater than or equal to 18.

Create a variable called `can_enter` that is `True` when the person is an adult and has a ticket, or when the person has a parent and a ticket.

Print `is_adult` and `can_enter`.

Then print the type of both results.

```python
age = 16
has_parent = True
has_ticket = True

is_adult = age >= 18
can_enter = (is_adult and has_ticket) or (has_parent and has_ticket)

print(is_adult)
print(can_enter)

print(type(is_adult))
print(type(can_enter))
```
