# Exercise 1 — Expressions and Variables

Create two variables:

- `boxes` containing 5;
- `items_per_box` containing 12.

Create a variable called `total_items` containing the result of multiplying the two variables.

Print `total_items`.

# Exercise 2 — Basic Arithmetic Operators

Create two variables:

- `a` containing 20;
- `b` containing 6.

Create one variable for the result of each of the following operations:

- addition;
- subtraction;
- multiplication;
- regular division.

Print all four results.

Then print the `type()` of the result of the regular division.

```python
a = 20
b = 6

addition = a + b
subtraction = a - b
multiplication = a * b
regular_division = a / b

print(addition)
print(subtraction)
print(multiplication)
print(regular_division)

print(type(addition))
print(type(subtraction))
print(type(multiplication))
print(type(regular_division))
```

# Exercise 3 — Division and Remainder

Create two variables:

- `total` containing 17;
- `groups` containing 5.

Create three variables:

- `regular_division` containing the result of `/`;
- `floor_division` containing the result of `//`;
- `remainder` containing the result of `%`.

Print the three results.

Then print the `type()` of each result.

```python
total = 17
groups = 5

regular_division = total / groups
floor_division = total // groups
remainer = total % groups

print(regular_division)
print(floor_division)
print(remainer)

print(type(regular_division))
print(type(floor_division))
print(type(remainer))
```

# Exercise 4 — Exponentiation

# Exercise 5 — Combining Expressions

Create the following variables:

- `price` containing 25.50;
- `quantity` containing 4;
- `discount` containing 10.

Create a variable called `total` that calculates the price multiplied by the quantity.

Then create a variable called `final_total` that subtracts `discount` from `total`.

Print `total` and `final_total`.

Finally, print the `type()` of both results.

```python
price = 25.50
quantity = 4
discount = 10

total = quantity * price

final_total = total - discount

print(total)
print(final_total)

print(type(total))
print(type(final_total))
```
