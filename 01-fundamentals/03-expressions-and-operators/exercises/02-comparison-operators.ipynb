# Exercise 1 — Equality and Inequality

Create two variables:

- `a` containing 10;
- `b` containing 10.

Create two variables:

- `equal` containing the result of comparing `a` and `b` with `==`;
- `different` containing the result of comparing `a` and `b` with `!=`.

Print both results.

Then print their types.

```python
a = 10
b = 10

equal = a == b
different = a != b

print(equal)
print(different)

print(type(equal))
print(type(different))
```

# Exercise 2 — Ordering Comparisons

Create two variables:

- `age` containing 25;
- `minimum_age` containing 18.

Create variables containing the results of:

- `age > minimum_age`;
- `age < minimum_age`;
- `age >= minimum_age`;
- `age <= minimum_age`.

Print all four results.

```python
age = 25
minimum_age = 18

greater_than = age > minimum_age
less_than = age < minimum_age
greater_or_equal_than = age >= minimum_age
less_or_equal_than = age <= minimum_age

print(greater_than)
print(less_than)
print(greater_or_equal_than)
print(less_or_equal_than)
```

# Exercise 3 — Comparing Different Numeric Types

Create two variables:

- `a` containing 10;
- `b` containing 10.0.

Print the result of:

- `a == b`;
- `a != b`.

Then print the type of both variables.

Finally, print the type of each comparison result.

```python
a = 10
b = 10.0

print(a == b)
print(a != b)

print(type(a))
print(type(b))

print(type(a == b))
print(type(a != b))
```

# Exercise 4 — Different Types

Create two variables:

- `number` containing 10;
- `text` containing the string `10`.

Print:

- the type of `number`;
- the type of `text`;
- the result of `number == text`;
- the result of `number != text`.

Do not convert either value to another type.

# Exercise 5 — Comparison and Reassignment

Create a variable called `age` containing 20.

Create a variable called `is_adult` containing the result of checking whether `age` is greater than or equal to 18.

Print `age`, `is_adult`, and the type of `is_adult`.

Then reassign `age` to 15.

Print `age` and `is_adult` again.

Observe whether `is_adult` changes automatically when `age` is reassigned.

```python
age = 20
is_adult = age >= 18

print(age)
print(is_adult)
print(type(is_adult))

age = 15

print(age)
print(is_adult)
```
