# Logical Operators

## The Problem

Programs often need to combine multiple conditions.

For example, suppose a program needs to determine whether a person can enter an event.

A person may need to satisfy more than one condition:

- be at least 18 years old;
- have a valid ticket.

## Combining Boolean Values

Python provides logical operators for combining boolean values.

The three basic logical operators are:

- `and`
- `or`
- `not`

These operators work with boolean values and produce a boolean result.

## `and`

The `and` operator produces `True` only when both values are `True`.

```python
a = True
b = True

result = a and b

print(result)
```

If one of the values is `False`, the result is `False`.

```python
a = True
b = False

print(a and b)
```

If both values are `False`, the result is also `False`.

```python
a = False
b = False

print(a and b)
```

## `or`

The `or` operator produces `True` when at least one of the values is `True`.

It produces `False` only when both values are `False`.

```python
a = True
b = True

print(a or b)
```

```python
a = True
b = False

print(a or b)
```

```python
a = False
b = True

print(a or b)
```

```python
a = False
b = False

print(a or b)
```

## `not`

The `not` operator reverses a boolean value.

- `not True` produces `False`.
- `not False` produces `True`.

```python
print(not True)
print(not False)
```

The result of `not` is also a boolean value.

```python
result = not True

print(result)
print(type(result))
```

## Combining Comparison and Logical Operators

Logical operators can combine the results of comparison expressions.

```python
age = 25
has_ticket = True

is_adult = age >= 18

can_enter = is_adult and has_ticket

print(can_enter)
```

## Using `or` with Comparisons

Logical operators can also combine alternative conditions.

For example, suppose a program accepts two types of identification.

```python
has_id = True
has_passport = False

can_identify = has_id or has_passport

print(can_identify)
```

## Using `not` with a Comparison

The `not` operator can reverse the result of a comparison.

```python
age = 15

is_adult = age >= 18

print(is_adult)
print(not is_adult)
```

## Logical Expressions Are Expressions

Logical operations are also expressions.

They are evaluated by Python and produce a value.

That value can be assigned to a variable.

```python
a = True
b = False

result = a and b

print(result)
print(type(result))
```

## Combining More Than Two Values

Logical operators can be used more than once in the same expression.

```python
a = True
b = True
c = False

result = a and b and c

print(result)
```

We can also combine different logical operators in the same expression.

```python
a = True
b = False
c = True

result = a and b or c

print(result)
```

## Logical Operators Do Not Modify the Original Values

Logical operators evaluate values and produce a result.

They do not modify the boolean values used in the expression.

```python
a = True
b = False

result = a and b

print(a)
print(b)
print(result)
```

# Summary

- Logical operators combine boolean values.
- `and` produces `True` only when both values are `True`.
- `or` produces `True` when at least one value is `True`.
- `not` reverses a boolean value.
- Logical operators produce boolean results.
- Comparison expressions can be combined using logical operators.
- The result of a logical expression can be assigned to a variable.
- Logical operations do not modify the original values.
